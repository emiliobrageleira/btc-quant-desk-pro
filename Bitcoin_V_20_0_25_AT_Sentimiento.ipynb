```python
# ============================================================
# QUANT DESK PRO | Google Colab
# Dashboard + LONG/SHORT + Riesgo + Backtest realista
# V.18 · datos externos (taker, funding, macro, F&G, contexto diario)
#
# Cambio sobre V.18.0:
#  - El botón "Dashboard HTML" ahora se abre directamente en una
#    pestaña nueva del navegador (si falla, se descarga como antes).
#
# V.18.3: nuevas temporalidades 5 min, 30 min, 45 min, 1 h, 4 h, 6 h,
#  12 h y 24 h. Cada una descarga el origen adecuado (Yahoo/Binance) y,
#  cuando no existe vela nativa, agrega OHLCV en bloques UTC.
#
# V.18.4: junto al selector de temporalidad se muestra la hora actual de
#  España + la duración de la temporalidad elegida = hora resultante, con
#  el día de la semana y la fecha (se actualiza en vivo).
# V.18.5: el mismo texto aparece también bajo los campos Entrada / Stop loss /
#  Take profit de los paneles LONG y SHORT.
# V.18.7: en la cabecera de los paneles LONG/SHORT, entre el título y el
#  Beneficio/riesgo bruto, se muestran los % de tendencia del modelo:
#  P(TP) del lado y su desviación en pp respecto a la tasa base.
# V.18.8: barra fina bajo esa cabecera, proporcional a P(TP) y completada
#  hasta el 100 % con otro color.
# V.18.9: los títulos de los paneles pasan a "LONG = COMPRAR" y
#  "SHORT = VENDER".
# V.18.10: en los paneles LONG/SHORT se invierte el orden de los campos:
#  Entrada · Take profit · Stop loss.
# V.18.11: parpadea el título (LONG = COMPRAR / SHORT = VENDER) del lado
#  con mayor P(TP).
# V.18.12: el botón "Dashboard HTML" usa una gama de verdes animada y una
#  flecha que se mueve invitando a pulsarlo.
# V.18.13: flecha blanca y texto del botón "ABRIR EN EL NAVEGADOR".
# V.18.14: nuevo diseño del botón: pastilla verde con brillo que barre, halo
#  pulsante, latido y mano animada; texto "VER DASHBOARD COMPLETO".
# V.18.15: el botón tiene el mismo tamaño que los demás botones de exportar.
# V.18.16: al terminar de cargar/ejecutar, el documento vuelve al inicio.
# V.18.17: al pulsar el panel LONG = COMPRAR o SHORT = VENDER se abre en el
#  centro de la pantalla un cuadro del tamaño de la zona pulsada con su contenido.
# V.18.18: ese cuadro se coloca pegado a la parte superior de la pantalla visible.
# V.18.19: indicador de carga (rueda + barra) en el centro de la pantalla visible
#  mientras se cargan o actualizan los datos.
# V.18.20: el indicador es el símbolo de Bitcoin (₿) girando 360° en horizontal.
# V.18.21: las zonas LONG = COMPRAR / SHORT = VENDER suben dos posiciones: ahora
#  van justo debajo de [precio | alerta] y antes de las tarjetas KPI.
# V.18.22: el indicador de carga muestra el porcentaje (0-100 %) y una barra.
# V.18.28: gráfico de líneas con la previsión de tendencia, justo debajo de
#  LONG = COMPRAR / SHORT = VENDER y antes de las tarjetas KPI (notebook y
#  HTML exportado), según la temporalidad elegida.
# V.18.29: sin cambios de lógica; el número de versión se ve en la cabecera
#  ('Quant Desk Pro · V.18.29') para distinguirlo de copias antiguas. Orden:
#  [precio | alerta] → Gestión del riesgo y tamaño de posición (LONG = COMPRAR
#  / SHORT = VENDER) → Previsión de tendencia → tarjetas KPI → mercado.
# V.18.30: la previsión de tendencia sube 6 posiciones: ahora tiene su propio
#  contenedor entre "Umbral de señal" y "Acciones" (por encima de Acciones,
#  Exportar, estado, alerta, [precio | alerta] y Gestión del riesgo). En el HTML
#  exportado va al principio del documento.
# V.18.31: el gráfico de previsión marca con líneas finas horizontales el techo
#  máximo y el suelo mínimo de la temporalidad elegida y sombrea la zona de
#  consolidación (activa o la más reciente) de esa temporalidad.
# V.18.32: los límites superior e inferior de la consolidación se proyectan con
#  líneas finas horizontales a lo largo de todo el gráfico (histórico y proyección).
# V.18.33: el gráfico de previsión incluye en su parte inferior el histograma de
#  volumen de las velas mostradas (verde = vela alcista, rojo = bajista).
# V.18.34: líneas horizontales en los techos y suelos donde coinciden varias
#  velas (máximos o mínimos locales casi al mismo precio) de la temporalidad
#  elegida, con el número de toques en la etiqueta.
# V.18.35: las dos tarjetas de carga (arranque y actualización) son idénticas y
#  el paso de una a otra no se nota: misma tarjeta reutilizada, aparece ya en su
#  sitio (sin saltar), se desliza suave si la pantalla cambia de tamaño y, si
#  hubiera dos copias, la nueva toma el relevo con el mismo % y tiempo.
# V.18.36: consulta diaria de la tendencia. Los datos diarios (cierres del activo,
#  macro y Fear & Greed) caducan tras el cierre diario (00:05 UTC) y los modelos se
#  recalculan una vez al día; un indicador muestra la tendencia diaria, el último
#  cierre y si hay que actualizar, y el botón "Forzar consulta diaria" lo fuerza.
# V.18.37: se eliminan del código todos los costes (se opera sin comisiones): el
#  equilibrio es 1 / (R:R + 1), el umbral automático parte de ahí y el backtest, las
#  tarjetas, las tablas y los textos ya no incluyen comisión, slippage ni funding como
#  coste. (El funding real de Binance sigue siendo un dato de entrada del modelo.)
# V.18.38: el gráfico de previsión incluye una capa opcional de velas japonesas
#  (OHLC de las velas mostradas), desactivada al inicio; se activa o desactiva
#  pulsando "Velas japonesas" en la leyenda del gráfico.
# V.18.39: autodiagnóstico en cada actualización (panel tras las tarjetas KPI y
#  resumen en la línea de estado): frescura de la última vela, datos diarios,
#  modelos, calidad OHLC, fuentes externas, señal, previsión y ventaja del modelo.
# V.18.40: con avisos (dorado) o errores (rojo) aparece junto a la línea de estado el
#  botón «Forzar autodiagnóstico»: repite la revisión con la hora actual, consulta
#  al proveedor si hay velas más nuevas y repinta el panel (que ahora se muestra
#  bajo la línea de estado en el notebook y sigue tras los KPI en el HTML exportado).
# V.18.41: bajo el precio del encabezado, el precio de hace 5 min, 30 min, 45 min,
#  1 h, 4 h, 6 h, 12 h y 24 h (las temporalidades seleccionables) con la variación
#  del precio actual frente a cada uno; la temporalidad elegida va resaltada.
# V.18.42: la fila de precios pasados añade 2 días, 1 semana, 1 mes, 6 meses y 1 año.
# V.18.43: la fila añade además el máximo anual y el mínimo anual (cierres diarios
#  de los últimos 365 días, con su fecha).
# V.18.44: la fila añade la mayor subida (alcista) y la mayor caída (bajista) entre
#  dos días del año en curso, en %, con las fechas y precios de inicio y fin.
#
# V.18.45: ANÁLISIS TÉCNICO AMPLIADO. 22 variables nuevas (TA_FEATURES: Estocástico,
#  StochRSI, CCI, MFI, CMF, OBV, VWAP, Supertrend, Ichimoku, Donchian, Aroon, DI±,
#  compresión Bollinger/Keltner, velas japonesas y divergencia RSI) que entran al modelo
#  (se desactivan con USE_TA = False) y un panel «Análisis técnico» con la lectura de
#  cada indicador y el consenso alcista/bajista/neutro, tras las tarjetas KPI.
#
# V.18.46: ANÁLISIS DE SENTIMIENTO. (1) 4 variables nuevas para el modelo (SENT_FEATURES:
#  atención en Wikipedia y tono de la prensa vía GDELT; se desactivan con USE_SENTIMENT = False)
#  que se suman a Fear & Greed y al funding. (2) Panel «Análisis de sentimiento» tras el análisis
#  técnico: titulares recientes (CoinDesk, Cointelegraph, Decrypt, Bitcoin Magazine, Google News)
#  puntuados con VADER + léxico cripto, índice compuesto -100…+100 y comparación con la señal.
#
# Novedades V.18 sobre V.17:
#  - Flujo taker (compra/venta agresora) de Binance.
#  - Funding rate REAL (Binance Futures, con Bybit de respaldo).
#  - Contexto diario (EMA200, máx/mín 50d, RSI diario) y régimen (ADX).
#  - Macro: S&P500, DXY, VIX, bono 10a, correlación con S&P500.
#  - Fear & Greed Index y activo de referencia (BTC/ETH).
#  - Comparativa 'solo precio' vs 'con datos extra' en la validación.
#  - Cada fuente es opcional: si falla, el panel lo indica.
#
# (V.17: rediseño visual)
# ============================================================

# ------------------------------------------------------------
# INDICADOR DE CARGA (moneda + porcentaje) desde el inicio del todo.
# Es el mismo indicador de siempre: aquí sólo se instala y se muestra
# antes de instalar las librerías; render_dashboard() lo actualiza y lo cierra.
# ------------------------------------------------------------
_LOADING_INSTALL_JS = r"""
(function () {
  if (window.__qdLoad && window.__qdLoad.v >= 2) { return; }
  var st = document.createElement('style');
  st.textContent =
    '#qd-load{position:fixed;inset:0;z-index:100000;pointer-events:none;opacity:0;' +
    'background:rgba(255,255,255,.97)}' +
    '#qd-load.qd-fade{transition:opacity .2s ease-out}#qd-load.qd-on{opacity:1}' +
    '#qd-load-card.qd-glide{transition:top .3s ease}' +
    '#qd-load-card{position:absolute;left:50%;top:50%;transform:translate(-50%,-50%);' +
    'width:min(92vw,520px);box-sizing:border-box;padding:16px 28px 14px;border-radius:16px;text-align:center;' +
    'background:#111a26;border:1px solid rgba(247,147,26,.45);color:#E8EEF6;' +
    'font-family:system-ui,-apple-system,Segoe UI,Roboto,sans-serif;' +
    'box-shadow:0 18px 50px rgba(0,0,0,.55),0 0 0 6px rgba(247,147,26,.08)}' +
    '#qd-coin-wrap{width:56px;height:56px;margin:0 auto 8px;perspective:600px}' +
    '#qd-coin{position:relative;width:100%;height:100%;transform-style:preserve-3d;' +
    'animation:qdlcoin 1.6s linear infinite}' +
    '.qd-face{position:absolute;inset:0;border-radius:50%;display:flex;align-items:center;' +
    'justify-content:center;backface-visibility:hidden;-webkit-backface-visibility:hidden;' +
    'font-family:Arial,Helvetica,sans-serif;font-weight:800;font-size:34px;line-height:1;color:#FFFFFF;' +
    'text-shadow:0 2px 3px rgba(120,60,0,.55);' +
    'background:radial-gradient(circle at 32% 28%,#FFD27A 0%,#F7931A 55%,#C76F00 100%);' +
    'border:3px solid #FFC15A;box-shadow:inset 0 0 0 2px rgba(199,111,0,.55),0 0 18px rgba(247,147,26,.45)}' +
    '.qd-face.qd-back{transform:rotateY(180deg)}' +
    '#qd-load-msg{font-size:15px;font-weight:700;letter-spacing:.2px;height:24px;line-height:1.3;' +
    'display:flex;align-items:center;justify-content:center;overflow:hidden}' +
    '#qd-load-pct{font-size:26px;font-weight:800;margin-top:4px;color:#F7931A;' +
    'font-variant-numeric:tabular-nums;line-height:1.1}' +
    '#qd-load-bar{height:6px;margin-top:8px;border-radius:99px;overflow:hidden;' +
    'background:rgba(247,147,26,.18)}' +
    '#qd-load-bar i{display:block;height:100%;width:0%;border-radius:99px;' +
    'background:linear-gradient(90deg,#C76F00,#F7931A,#FFD27A);transition:width .12s linear}' +
    '#qd-load-sub{font-size:12px;color:#93A3B8;margin-top:6px}' +
    '#qd-load-time{margin-top:8px;font-size:12.5px;line-height:1.5;color:#C9D4E2;' +
    'font-variant-numeric:tabular-nums}' +
    '#qd-load-time span{display:inline-block;margin:0 10px}#qd-load-time b{color:#FFD27A;font-weight:700}' +
    '#qd-load-tip{height:42px;overflow:hidden;margin-top:10px;padding-top:8px;border-top:1px solid rgba(255,255,255,.14);' +
    'font-size:13px;line-height:1.5;font-style:italic;color:#FFD27A;animation:qdltip .5s ease-out}' +
    '@keyframes qdltip{from{opacity:0;transform:translateY(5px)}to{opacity:1;transform:translateY(0)}}' +
    '@keyframes qdlcoin{from{transform:rotateY(0deg)}to{transform:rotateY(360deg)}}' +
    '';
  document.head.appendChild(st);

  var timer = null, io = null, el = null, tick = null, removeT = null;
  var active = false, pendingMsg = '', target = 0, shown = 0, creepCap = 6;
  var t0 = 0, kind = 'first', hist = 0;   // inicio, tipo de carga y duración de la vez anterior
  var tipT = null, tipLast = -1;
  var myId = Math.random().toString(36).slice(2), chan = null;
  var placed = false, resumed = false, live = null;
  var TIPS = [
    "Bitcoin cotiza las 24 horas, los 7 días de la semana: el mercado no tiene hora de cierre.",
    "La oferta de Bitcoin está limitada a 21 millones de unidades y cada halving reduce a la mitad la emisión de nuevos bloques.",
    "La volatilidad es una característica estructural del Bitcoin: abre oportunidades, pero exige una gestión de riesgo rigurosa.",
    "El funding rate de los perpetuos refleja el posicionamiento: un funding muy positivo indica un mercado apalancado en largos.",
    "El flujo taker mide quién cruza el spread: más compras agresivas indican presión compradora inmediata.",
    "Las liquidaciones en cadena de posiciones apalancadas pueden amplificar los movimientos bruscos de precio.",
    "El índice de Miedo y Codicia resume el sentimiento del mercado; los extremos ayudan a contextualizar, no a predecir.",
    "El Bitcoin reacciona con frecuencia a la liquidez global, a los tipos de interés y a la fortaleza del dólar.",
    "Los flujos de los ETF de Bitcoin al contado son uno de los datos institucionales más seguidos por el mercado.",
    "Los fines de semana la liquidez suele ser menor y los movimientos pueden resultar más bruscos.",
    "Cada señal se evalúa con la probabilidad de alcanzar el take profit antes que el stop loss (triple barrera).",
    "El backtest vela a vela valida la estrategia con datos pasados; un buen resultado histórico no garantiza el futuro.",
    "Dimensionar la posición según el riesgo por operación protege el capital en las rachas adversas.",
    "Un mercado lateral y uno con tendencia exigen estrategias distintas: la temporalidad elegida condiciona la lectura."
  ];
  function nextTip() {
    if (!el) { return; }
    var box = el.querySelector('#qd-load-tip');
    if (!box) { return; }
    var i = Math.floor(Math.random() * TIPS.length);
    if (TIPS.length > 1 && i === tipLast) { i = (i + 1 + Math.floor(Math.random() * (TIPS.length - 1))) % TIPS.length; }
    tipLast = i;
    box.textContent = TIPS[i];
    box.style.animation = 'none'; void box.offsetWidth; box.style.animation = '';   // reinicia el fundido
  }

  // Si el indicador vive en dos sitios (dos iframes), la copia nueva continúa con el
  // mismo % / tiempo / consejo y la antigua se retira en cuanto la nueva está colocada.
  try {
    chan = new BroadcastChannel('qd-load');
    chan.onmessage = function (ev) {
      if (ev.data && ev.data.t === 'takeover' && ev.data.id !== myId && el) { cleanup(); }
    };
  } catch (err) { chan = null; }
  function saveLive() {
    try {
      var tb = el && el.querySelector('#qd-load-tip');
      localStorage.setItem('qd_load_live', JSON.stringify({
        u: Date.now(), t0: t0, shown: shown, target: target, kind: kind,
        msg: pendingMsg, tip: tb ? tb.textContent : ''}));
    } catch (err) {}
  }
  function place(top) {
    var card = el && el.querySelector('#qd-load-card');
    if (!card) { return; }
    if (typeof top === 'number') { card.style.top = top + 'px'; }
    if (placed) { return; }
    placed = true;
    if (!resumed) { el.classList.add('qd-fade'); }   // fundido sólo si no hay otra tarjeta ya visible
    void el.offsetWidth;
    el.classList.add('qd-on');
    setTimeout(function () { if (el) { var c = el.querySelector('#qd-load-card'); if (c) { c.classList.add('qd-glide'); } } }, 80);
    try { if (chan) { chan.postMessage({t: 'takeover', id: myId}); } } catch (err) {}
  }

  function paint() {
    if (!el) { return; }
    var p = Math.max(0, Math.min(100, shown));
    el.querySelector('#qd-load-pct').textContent = Math.floor(p) + '%';
    el.querySelector('#qd-load-bar i').style.width = p + '%';
    // Contador de segundos y tiempo aproximado de carga
    var e = Math.max(0, (Date.now() - t0) / 1000);
    var est = hist || (kind === 'first' ? 45 : 20);
    if (shown >= 10) {
      var byPct = e * 100 / shown;
      est = hist ? (0.4 * hist + 0.6 * byPct) : byPct;
    }
    est = (shown >= 100) ? e : Math.max(est, e + 1);
    var a = el.querySelector('#qd-load-el'), b = el.querySelector('#qd-load-est');
    if (a) { a.innerHTML = '\u23F1 Tiempo transcurrido: <b>' + Math.floor(e) + ' s</b>'; }
    if (b) { b.innerHTML = 'Tiempo aproximado de carga: <b>~' + Math.round(est) + ' s</b>'; }
  }

  function setMsg(msg) {
    if (msg) { pendingMsg = msg; }
    if (el) { el.querySelector('#qd-load-msg').textContent = pendingMsg; }
  }

  function ensure() {
    if (el) { return; }
    el = document.createElement('div');
    el.id = 'qd-load';
    el.innerHTML = '<div id="qd-load-card"><div id="qd-coin-wrap"><div id="qd-coin">' +
      '<div class="qd-face">\u20BF</div><div class="qd-face qd-back">\u20BF</div></div></div>' +
      '<div id="qd-load-msg"></div><div id="qd-load-pct">0%</div>' +
      '<div id="qd-load-bar"><i></i></div>' +
      '<div id="qd-load-time"><span id="qd-load-el"></span><span id="qd-load-est"></span></div>' +
      '<div id="qd-load-tip"></div>' +
      '<div id="qd-load-sub">Por favor, espera un momento…</div></div>';
    document.body.appendChild(el);
    setMsg();
    paint();
    if (resumed && live && live.tip) { el.querySelector('#qd-load-tip').textContent = live.tip; } else { nextTip(); }
    tipT = setInterval(nextTip, 10000);
    // Centro del tramo VISIBLE (si la salida vive en un iframe alto, como en Colab)
    try {
      io = new IntersectionObserver(function (entries) {
        var e = entries[entries.length - 1];
        var card = document.getElementById('qd-load-card');
        if (card && e && e.isIntersecting && e.intersectionRect) {
          place(Math.round(e.intersectionRect.top + e.intersectionRect.height / 2));
        }
      }, {threshold: (function () { var t = []; for (var i = 0; i <= 200; i++) { t.push(i / 200); } return t; })()});
      io.observe(document.documentElement);
    } catch (err) {}
    setTimeout(function () { place(null); }, 300);   // respaldo si no hay IntersectionObserver
    // El número sube suavemente hacia el avance real; entre avisos avanza despacio
    tick = setInterval(function () {
      if (shown < target) {
        shown = Math.min(target, shown + Math.max(0.5, (target - shown) * 0.2));
      } else if (shown < creepCap && shown < 99) {
        shown += 0.08;
      }
      paint();
      saveLive();
    }, 100);
  }

  function cleanup() {
    if (tick) { clearInterval(tick); tick = null; }
    if (tipT) { clearInterval(tipT); tipT = null; }
    if (io) { io.disconnect(); io = null; }
    if (removeT) { clearTimeout(removeT); removeT = null; }
    if (el) { el.remove(); el = null; }
    active = false; placed = false; resumed = false;
  }

  window.__qdLoad = {
    v: 2,
    show: function (msg) {
      var revive = !!removeT;                     // la tarjeta aún enseña el 100 %: se reutiliza
      if (removeT) { clearTimeout(removeT); removeT = null; }
      if (msg) { pendingMsg = msg; }
      if (active && !revive) { setMsg(msg); return; }   // llamada anidada: sólo cambia el texto
      active = true; target = 0; shown = 0; creepCap = 6;
      t0 = Date.now();
      kind = (window.__qdLoadN || 0) === 0 ? 'first' : 'refresh';
      window.__qdLoadN = (window.__qdLoadN || 0) + 1;
      // ¿hay otra copia del indicador en marcha (otro iframe)? Entonces se continúa donde iba.
      live = null; resumed = false;
      try {
        var lv = JSON.parse(localStorage.getItem('qd_load_live') || 'null');
        if (lv && Date.now() - lv.u < 2500 && !el) { live = lv; resumed = true; t0 = lv.t0; shown = lv.shown; target = lv.target; kind = lv.kind; if (!msg) { pendingMsg = lv.msg; } }
      } catch (err) {}
      hist = 0;
      try { var hv = parseFloat(localStorage.getItem('qd_load_' + kind)); if (isFinite(hv) && hv > 0) { hist = hv; } } catch (err) {}
      if (el) { setMsg(); paint(); return; }      // misma tarjeta: no se recrea, no hay salto
      timer = setTimeout(function () { timer = null; ensure(); }, resumed ? 0 : 120);  // evita parpadeos en cargas instantáneas
    },
    progress: function (p, msg, creepTo) {
      if (!active) { return; }
      if (typeof p === 'number') { target = Math.max(target, Math.min(99, p)); }
      creepCap = (typeof creepTo === 'number') ? Math.min(97, creepTo) : Math.min(97, target + 6);
      setMsg(msg);
    },
    hide: function () {
      if (timer) { clearTimeout(timer); timer = null; }
      if (!el) { try { localStorage.removeItem('qd_load_live'); } catch (err) {} cleanup(); return; }
      try { localStorage.setItem('qd_load_' + kind, String(Math.round((Date.now() - t0) / 1000))); localStorage.removeItem('qd_load_live'); } catch (err) {}
      target = 100; shown = 100; paint();         // se ve el 100 % un instante
      setMsg('Listo');
      removeT = setTimeout(cleanup, 450);
    }
  };
})();
"""

from IPython.display import display as _boot_display, HTML as _BootHTML, Javascript as _BootJS

# Al arrancar, la salida de la celda mide casi 0 px y la tarjeta quedaría recortada.
# Este hueco temporal le da altura para verse entera, como el indicador normal.
try:
    _boot_display(_BootHTML(
        "<div id='qd-boot-spacer' style='height:280px'></div>"))
except Exception:
    pass

try:
    _boot_display(_BootJS(
        _LOADING_INSTALL_JS +
        "\nwindow.__qdLoad.show('Cargando librerías y preparando el código…');"))
except Exception:
    pass

%pip -q install -U yfinance plotly vaderSentiment

import os
import asyncio
import json
import threading
import time as _time
import contextlib
import urllib.parse
import urllib.request
import warnings
from datetime import datetime, timedelta, time as dtime
from zoneinfo import ZoneInfo
from types import SimpleNamespace

import numpy as np
import pandas as pd
import yfinance as yf
import plotly.graph_objects as go
import ipywidgets as widgets

from IPython import get_ipython
from IPython.display import display, HTML, Javascript
from plotly.subplots import make_subplots
from sklearn.ensemble import HistGradientBoostingClassifier
from sklearn.metrics import log_loss, brier_score_loss, roc_auc_score
from sklearn.model_selection import TimeSeriesSplit

warnings.filterwarnings("ignore")

VERSION = "V.18.46 · análisis de sentimiento"

# ---------------- CONFIGURACIÓN ----------------

SYMBOLS = ["BTC-USD", "ETH-USD", "SOL-USD"]
DEFAULT_SYMBOL = "BTC-USD"
PERIOD = "700d"          # Histórico por defecto (velas 1h)
INTERVAL = "1h"           # Temporalidad efectiva del dashboard (ver TIMEFRAMES)

# Cada temporalidad define de dónde salen las velas y cómo se ajusta el modelo.
#   label     : texto del desplegable
#   minutes   : duración de la vela
#   yf        : (intervalo Yahoo, periodo Yahoo) de la descarga base
#   agg       : minutos de la vela base si hay que agregar (None = nativa)
#   taker     : intervalo Binance de las velas de flujo taker y su duración (min)
#   atr       : distancia del stop en múltiplos de ATR (LONG y SHORT)
#   rr        : ratio beneficio/riesgo (take profit = stop * rr)
#   hold      : velas máximas en posición (y horizonte de etiquetado)
#   refit_h   : cada cuántas HORAS se reentrena el walk-forward
#   min_bars  : velas cerradas mínimas para aceptar la descarga
#   wf_min    : arranque mínimo (en velas) del walk-forward
#   vol       : (ventana, mínimo) de la mediana de volatilidad (Vol_Regime)
# Límites de Yahoo: 5m/15m/30m sólo los últimos 60 días; 1h hasta 730 días.
TIMEFRAMES = {
    "5m":  dict(label="5 minutos", minutes=5,    yf=("5m", "60d"),  agg=None,
                taker=("5m", 5),    atr=0.75, rr=1.5, hold=48, refit_h=168, min_bars=1500,
                wf_min=1000, vol=(720, 240)),
    "30m": dict(label="30 minutos", minutes=30,  yf=("30m", "60d"), agg=None,
                taker=("30m", 30),  atr=1.0, rr=1.5, hold=48, refit_h=168, min_bars=1500,
                wf_min=1000, vol=(720, 240)),
    "45m": dict(label="45 minutos", minutes=45,  yf=("15m", "60d"), agg=15,
                taker=("15m", 15),  atr=1.25, rr=1.5, hold=32, refit_h=168, min_bars=1200,
                wf_min=700, vol=(720, 240)),
    "1h":  dict(label="1 hora", minutes=60,      yf=("1h", PERIOD), agg=None,
                taker=("1h", 60),   atr=1.5, rr=2.0, hold=24, refit_h=168, min_bars=1500,
                wf_min=1000, vol=(720, 240)),
    "4h":  dict(label="4 horas", minutes=240,    yf=("1h", PERIOD), agg=60,
                taker=("1h", 60),   atr=1.0, rr=2.0, hold=16, refit_h=168, min_bars=1500,
                wf_min=1000, vol=(720, 240)),
    "6h":  dict(label="6 horas", minutes=360,    yf=("1h", PERIOD), agg=60,
                taker=("1h", 60),   atr=1.0, rr=2.0, hold=16, refit_h=168, min_bars=1500,
                wf_min=1000, vol=(720, 240)),
    "12h": dict(label="12 horas", minutes=720,   yf=("1h", PERIOD), agg=60,
                taker=("1h", 60),   atr=0.75, rr=2.0, hold=12, refit_h=336, min_bars=1000,
                wf_min=500, vol=(360, 120)),
    "24h": dict(label="24 horas", minutes=1440,  yf=("1d", "7y"),   agg=None,
                taker=("1d", 1440), atr=0.75, rr=2.0, hold=12, refit_h=672, min_bars=1000,
                wf_min=500, vol=(365, 120)),
}
DEFAULT_INTERVAL = "1h"
TF = TIMEFRAMES[INTERVAL]
BAR_MIN = TF["minutes"]
BAR_HOURS = BAR_MIN / 60.0
BAR_FREQ = f"{BAR_MIN}min"
VOL_WIN, VOL_MIN = TF["vol"]
MIN_BARS = TF["min_bars"]

ATR_MULTIPLIER = TF["atr"]     # distancia del stop (depende de la temporalidad)
RISK_REWARD = TF["rr"]        # take profit = stop * RISK_REWARD (depende de la temporalidad)
MAX_HOLD_BARS = TF["hold"]   # salida por tiempo (depende de la temporalidad)
LABEL_HORIZON = MAX_HOLD_BARS

DEFAULT_CAPITAL = 2_000.0
DEFAULT_RISK_PCT = 0.5     # 0,5 % de 2.000 $ = 10 $ por operación
DEFAULT_MAX_POSITION_PCT = 100.0
DEFAULT_THRESHOLD = 0.50     # sólo se usa con el umbral manual
AUTO_MARGIN = 0.05           # umbral automático = equilibrio + margen
BIAS_DEADZONE = 0.02         # sesgo < 2 pp respecto a la base = NEUTRO
HYSTERESIS = 0.03            # para avisar de "ESPERAR" tras LONG/SHORT

WF_START_FRAC = 0.40
WF_MIN_START = TF["wf_min"]
WF_REFIT_BARS = max(1, int(round(TF["refit_h"] / BAR_HOURS)))  # se recalibra al cambiar temporalidad
MAX_TRAIN_SAMPLES = 6000
MIN_TRAIN_SAMPLES = 400
N_RANDOM_SIMS = 200
PLOT_BARS = 500

REFRESH_SECONDS = 300
DAILY_CLOSE_DELAY_MIN = 5    # el cierre diario (00:00 UTC) se consulta pasados 5 min

ANNUAL_PERIODS = int(24 * 365 / BAR_HOURS)

# Orígenes de datos (data-api.binance.vision funciona también desde EE. UU.;
# api.binance.com / fapi devuelven 451 en algunas regiones -> hay respaldo).
BINANCE_SPOT = ["https://data-api.binance.vision", "https://api.binance.com"]
BINANCE_FUT = "https://fapi.binance.com"

# ---------------- SISTEMA DE DISEÑO ----------------
# Base: azul pizarra profundo. Verde azulado = LONG, coral = SHORT,
# ámbar = espera / equilibrio. Un solo tipo de letra.

BG = "#0b1119"
PANEL = "#111a26"
PANEL2 = "#162131"
TEXT = "#E8EEF6"
MUTED = "#93A3B8"
GREEN = "#34D3A6"
RED = "#FF7088"
BLUE = "#6FB4F5"
PURPLE = "#B3A1F2"
GOLD = "#F2C46D"

GRID_COLOR = "#1C2837"
BORDER_COLOR = "#243246"
HOVER_BG = "#1D2A3D"

FONT = "'IBM Plex Sans','Segoe UI',Roboto,'Helvetica Neue',Arial,sans-serif"
PLOT_FONT = "IBM Plex Sans, Segoe UI, Roboto, Arial, sans-serif"

SIGNAL_COLORS = {"LONG": GREEN, "SHORT": RED, "ESPERAR": GOLD}
SIGNAL_ICONS = {"LONG": "▲", "SHORT": "▼", "ESPERAR": "●"}
TREND_COLORS = {"ALCISTA": GREEN, "BAJISTA": RED, "MIXTA": GOLD, "NEUTRO": GOLD}
GAUGE_MAX = 0.80          # fondo de escala de los medidores de P(TP)

PRICE_FEATURES = [          # las 18 originales (sólo precio/volumen)
    "Return", "Ret_3", "Ret_6", "Ret_12", "Ret_24",
    "EMA9_Rel", "EMA21_Rel", "EMA50_Rel",
    "RSI", "ATR_Rel", "Volatility",
    "MACD_Hist", "BB_Pct", "Vol_Rel",
    "Hour_Sin", "Hour_Cos", "Dow_Sin", "Dow_Cos",
]
REGIME_FEATURES = ["ADX", "Vol_Regime"]   # régimen: tendencia vs rango
USE_TA = True    # False = el modelo ignora las variables de AT ampliado (para comparar)
TA_FEATURES = [                           # análisis técnico ampliado (NaN al inicio)
    "Stoch_K", "Stoch_D", "StochRSI", "CCI", "MFI", "CMF",     # osciladores / flujo
    "OBV_Slope", "VWAP_Dist",                                  # volumen
    "ST_Dir", "ST_Dist",                                       # Supertrend
    "Ichi_Cloud", "Ichi_TK", "Ichi_Pos",                       # Ichimoku
    "Donchian_Pos", "Aroon_Osc", "DI_Diff",                    # estructura / tendencia
    "Squeeze", "Body_Ratio", "Upper_Wick", "Lower_Wick",       # volatilidad / velas
    "Engulf", "RSI_Div",                                       # patrones
]
USE_SENTIMENT = True   # False = el modelo ignora las variables de sentimiento nuevas
SENT_FEATURES = ["Wiki_Z", "Wiki_Chg_7",           # atención del público (Wikipedia)
                 "Press_Tone", "Press_Tone_Z"]     # tono de la prensa (GDELT, ~90 d)
EXTRA_FEATURES = [                        # datos externos (pueden ser NaN)
    "Taker_Imb", "Taker_Imb_24",                          # flujo agresor
    "Funding", "Funding_Z", "Funding_Avg24",              # derivados
    "D_EMA200_Rel", "D_Dist_High50", "D_Dist_Low50",      # contexto diario
    "D_Ret_7", "D_Ret_30", "D_RSI",
    "SPX_Ret_1d", "SPX_Ret_5d", "DXY_Ret_5d", "VIX",      # macro
    "VIX_Chg_5d", "TNX_Chg_5d", "Corr_SPX_30",
    "FNG", "FNG_Chg_7",                                   # sentimiento
    "Peer_Ret_6", "Peer_Ret_24", "Rel_24",                # activo de referencia
] + (SENT_FEATURES if USE_SENTIMENT else [])
FEATURES = (PRICE_FEATURES + REGIME_FEATURES
            + (TA_FEATURES if USE_TA else []) + EXTRA_FEATURES)

# ---------------- ESTADO GLOBAL ----------------

def make_config_version():
    return (VERSION, PERIOD, INTERVAL, ATR_MULTIPLIER, RISK_REWARD,
            MAX_HOLD_BARS, WF_REFIT_BARS, MAX_TRAIN_SAMPLES,
            tuple(FEATURES))


CONFIG_VERSION = make_config_version()
_prev = globals().get("S")

S = SimpleNamespace(
    alerts={},        # última señal notificada por activo
    bundles={},       # datos + modelos por activo
    sim_key=None,
    sim=None,         # resultado del backtest del activo mostrado
    view=None,        # datos para reconstruir los gráficos del export
    signal="ESPERAR",
    signal_color=GOLD,
    metrics=None,
    validation=None,
    trades=pd.DataFrame(),
    exported=None,    # símbolo del último export
    auto=False,
    task=None,
    ext={},           # caché de datos externos
)

if _prev is not None:
    if hasattr(_prev, "ext"):
        S.ext = _prev.ext
    if hasattr(_prev, "alerts"):
        S.alerts = _prev.alerts
    _old = getattr(_prev, "bundles", None)
    if isinstance(_old, dict):
        S.bundles = {k: v for k, v in _old.items()
                     if v.get("config") == CONFIG_VERSION}

_LOCK = threading.RLock()

# Ajustes de interfaz (se actualizan en cada render)
UI = {"mobile": False, "size": 23, "min": 10_000}


@contextlib.contextmanager
def ui_mode(mobile):
    old = dict(UI)
    UI.update(mobile=mobile, size=18 if mobile else 23,
              min=135 if mobile else 10_000)
    try:
        yield
    finally:
        UI.clear()
        UI.update(old)


def rgba(hex_color, alpha):
    h = hex_color.lstrip("#")
    r, g, b = (int(h[i:i + 2], 16) for i in (0, 2, 4))
    return f"rgba({r},{g},{b},{alpha})"


def set_status(text, color=BLUE, bold=False):
    weight = ";font-weight:600" if bold else ""
    try:
        status.value = (
            f"<span class='qd-status' style='color:{color};"
            f"background:{rgba(color, 0.12)};border:1px solid "
            f"{rgba(color, 0.32)};padding:5px 12px;border-radius:99px;"
            f"font-size:12.5px;display:inline-block{weight}'>{text}</span>")
    except Exception:
        pass


def fname(kind, symbol):
    base = symbol.split("-")[0].lower()
    return {
        "backtest": f"quant_{base}_backtest.csv",
        "trades": f"quant_{base}_operaciones.csv",
        "candles": f"quant_{base}_velas.csv",
        "metrics": f"quant_{base}_metricas.csv",
        "html": f"quant_{base}_dashboard.html",
    }[kind]


# ============================================================
# 1. DATOS E INDICADORES
# ============================================================

def _fmt_dur(minutes):
    """Duración legible: 45 min, 1 h, 24 h."""
    m = float(minutes)
    return f"{m:g} min" if m < 60 else f"{m / 60:g} h"


def download_data(symbol):
    yf_interval, yf_period = TF["yf"]
    data = yf.download(
        symbol, period=yf_period, interval=yf_interval,
        auto_adjust=True, progress=False, threads=False,
    )

    if data is None or data.empty:
        raise RuntimeError(
            "Yahoo Finance no devolvió datos. Inténtalo de nuevo.")

    if isinstance(data.columns, pd.MultiIndex):
        data.columns = data.columns.get_level_values(0)

    required = ["Open", "High", "Low", "Close", "Volume"]
    missing = [c for c in required if c not in data.columns]
    if missing:
        raise ValueError(f"Faltan columnas de mercado: {missing}")

    data = data[required].copy()
    data = data.apply(pd.to_numeric, errors="coerce")
    data = data.replace([np.inf, -np.inf], np.nan).dropna()

    if data.index.tz is None:
        data.index = data.index.tz_localize("UTC")
    else:
        data.index = data.index.tz_convert("UTC")

    # Yahoo no ofrece velas nativas de 45m, 4h, 6h ni 12h. Descargamos la
    # vela base (15m o 1h) y agregamos OHLCV en bloques UTC alineados con
    # la época (00:00 UTC) cuando la vela pedida es más larga que la base.
    if TF["agg"] is not None:
        first_raw = data.index[0]
        data = data.resample(
            BAR_FREQ, origin="epoch", label="left", closed="left"
        ).agg({
            "Open": "first", "High": "max", "Low": "min",
            "Close": "last", "Volume": "sum"
        }).dropna(subset=["Open", "High", "Low", "Close"])
        # La primera vela puede quedar a medias (el histórico empieza a mitad
        # de bloque): se descarta para no distorsionar OHLC ni volumen.
        if len(data) and data.index[0] < first_raw:
            data = data.iloc[1:]

    data.index = data.index.floor(BAR_FREQ)
    data = data[~data.index.duplicated(keep="last")].sort_index()

    # Descartar la vela en formación (aún no ha cerrado).
    if len(data):
        now = pd.Timestamp.now(tz="UTC")
        if data.index[-1] + pd.Timedelta(minutes=BAR_MIN) > now:
            data = data.iloc[:-1]

    # Huecos: reindexar a rejilla completa. Los huecos cortos (<=3 velas)
    # se rellenan con el último cierre y volumen 0; el resto se descarta.
    gaps = 0
    if len(data) > 1:
        full = pd.date_range(data.index[0], data.index[-1],
                             freq=BAR_FREQ, tz=data.index.tz)
        gaps = len(full) - len(data)
        if gaps > 0:
            data = data.reindex(full)
            data["Close"] = data["Close"].ffill(limit=3)
            for col in ("Open", "High", "Low"):
                data[col] = data[col].fillna(data["Close"])
            data["Volume"] = data["Volume"].fillna(0.0)
            data = data.dropna()
            data.index.name = "Datetime"

    if len(data) < MIN_BARS:
        raise RuntimeError(
            f"Solo hay {len(data)} velas cerradas de {INTERVAL} "
            f"(mínimo {MIN_BARS}). Yahoo limita el histórico de esta "
            "temporalidad; prueba otra o inténtalo de nuevo.")

    return data, gaps


def calculate_rsi(series, period=14):
    delta = series.diff()
    gain = delta.clip(lower=0)
    loss = -delta.clip(upper=0)

    avg_gain = gain.ewm(alpha=1 / period, min_periods=period, adjust=False).mean()
    avg_loss = loss.ewm(alpha=1 / period, min_periods=period, adjust=False).mean()

    rs = avg_gain / avg_loss.replace(0, np.nan)
    rsi = 100 - 100 / (1 + rs)
    rsi = rsi.mask((avg_loss == 0) & (avg_gain > 0), 100)
    rsi = rsi.mask((avg_loss == 0) & (avg_gain == 0), 50)
    return rsi


def _supertrend(high, low, close, atr, mult=3.0):
    """Supertrend (ATR 14, factor 3). Devuelve (nivel, dirección +1/-1)."""
    n = len(close)
    hl2 = (high + low) / 2.0
    ub, lb = hl2 + mult * atr, hl2 - mult * atr
    fub, flb = ub.copy(), lb.copy()
    st = np.full(n, np.nan)
    dr = np.full(n, np.nan)
    valid = np.where(np.isfinite(atr))[0]
    if len(valid) == 0:
        return st, dr
    s = valid[0]
    dr[s], st[s] = 1.0, lb[s]
    for i in range(s + 1, n):
        fub[i] = ub[i] if (ub[i] < fub[i - 1] or close[i - 1] > fub[i - 1]) else fub[i - 1]
        flb[i] = lb[i] if (lb[i] > flb[i - 1] or close[i - 1] < flb[i - 1]) else flb[i - 1]
        if dr[i - 1] == 1.0:
            dr[i] = -1.0 if close[i] < flb[i] else 1.0
        else:
            dr[i] = 1.0 if close[i] > fub[i] else -1.0
        st[i] = flb[i] if dr[i] == 1.0 else fub[i]
    return st, dr


def add_ta_indicators(data):
    """
    Análisis técnico ampliado. Todo se calcula SÓLO con velas pasadas (sin
    fugas): las nubes de Ichimoku se desplazan hacia delante con shift(+26),
    nunca hacia atrás. Las variables son relativas (sin precio absoluto) para
    que el modelo sea estable entre épocas y activos.
    """
    c, h, l, o, v = (data[k] for k in ("Close", "High", "Low", "Open", "Volume"))
    v = v.replace(0, np.nan)
    rng = (h - l).replace(0, np.nan)

    # --- Osciladores ---
    ll14, hh14 = l.rolling(14).min(), h.rolling(14).max()
    data["Stoch_K"] = 100 * (c - ll14) / (hh14 - ll14).replace(0, np.nan)
    data["Stoch_D"] = data["Stoch_K"].rolling(3).mean()
    rsi = data["RSI"]
    rmin, rmax = rsi.rolling(14).min(), rsi.rolling(14).max()
    data["StochRSI"] = (rsi - rmin) / (rmax - rmin).replace(0, np.nan)

    tp = (h + l + c) / 3.0
    mad = tp.rolling(20).apply(lambda x: np.mean(np.abs(x - x.mean())), raw=True)
    data["CCI"] = ((tp - tp.rolling(20).mean()) / (0.015 * mad.replace(0, np.nan))).clip(-300, 300)

    # --- Volumen / flujo ---
    rmf = tp * v
    pos = rmf.where(tp > tp.shift(1), 0.0).rolling(14).sum()
    neg = rmf.where(tp < tp.shift(1), 0.0).rolling(14).sum()
    data["MFI"] = (100 - 100 / (1 + pos / neg.replace(0, np.nan))).where(neg > 0, 100.0)
    data["MFI"] = data["MFI"].where(pos + neg > 0)
    mfm = ((c - l) - (h - c)) / rng
    data["CMF"] = (mfm * v).rolling(20).sum() / v.rolling(20).sum()
    obv = (np.sign(c.diff()).fillna(0) * v.fillna(0)).cumsum()
    data["OBV"] = obv
    data["OBV_Slope"] = (obv - obv.shift(24)) / v.rolling(24).sum()
    vwap = (tp * v).rolling(24).sum() / v.rolling(24).sum()
    data["VWAP24"] = vwap
    data["VWAP_Dist"] = c / vwap - 1

    # --- Supertrend ---
    st, dr = _supertrend(h.to_numpy(float), l.to_numpy(float), c.to_numpy(float),
                         data["ATR"].to_numpy(float))
    data["Supertrend"] = st
    data["ST_Dir"] = dr
    data["ST_Dist"] = (c - data["Supertrend"]) / c

    # --- Ichimoku (9, 26, 52) ---
    tenkan = (h.rolling(9).max() + l.rolling(9).min()) / 2
    kijun = (h.rolling(26).max() + l.rolling(26).min()) / 2
    sen_a = ((tenkan + kijun) / 2).shift(26)
    sen_b = ((h.rolling(52).max() + l.rolling(52).min()) / 2).shift(26)
    top, bot = np.maximum(sen_a, sen_b), np.minimum(sen_a, sen_b)
    data["Ichi_Top"], data["Ichi_Bot"] = top, bot
    data["Ichi_Cloud"] = (c - (sen_a + sen_b) / 2) / c
    data["Ichi_TK"] = (tenkan - kijun) / c
    pos_c = pd.Series(np.where(c > top, 1.0, np.where(c < bot, -1.0, 0.0)), index=data.index)
    data["Ichi_Pos"] = pos_c.where(top.notna() & bot.notna())

    # --- Estructura y tendencia ---
    hi20, lo20 = h.rolling(20).max(), l.rolling(20).min()
    data["Donchian_Pos"] = (c - lo20) / (hi20 - lo20).replace(0, np.nan)
    up_i = h.rolling(26).apply(np.argmax, raw=True)
    dn_i = l.rolling(26).apply(np.argmin, raw=True)
    data["Aroon_Osc"] = (up_i - dn_i) / 25 * 100
    # DI_Diff se calcula dentro de add_indicators (junto al ADX)

    # --- Volatilidad: Bollinger dentro de Keltner = compresión ---
    data["Squeeze"] = (4 * c.rolling(20).std()) / (3 * data["ATR"]).replace(0, np.nan)

    # --- Velas japonesas ---
    body_hi, body_lo = np.maximum(o, c), np.minimum(o, c)
    data["Body_Ratio"] = (c - o) / rng
    data["Upper_Wick"] = (h - body_hi) / rng
    data["Lower_Wick"] = (body_lo - l) / rng
    bull = (c > o) & (c.shift(1) < o.shift(1)) & (c >= o.shift(1)) & (o <= c.shift(1))
    bear = (c < o) & (c.shift(1) > o.shift(1)) & (c <= o.shift(1)) & (o >= c.shift(1))
    data["Engulf"] = bull.astype(float) - bear.astype(float)

    # --- Divergencia RSI (precio y RSI discrepan entre dos ventanas de 14) ---
    ph, pl = c.rolling(14).max(), c.rolling(14).min()
    rh, rl = rsi.rolling(14).max(), rsi.rolling(14).min()
    bear_div = (ph > ph.shift(14)) & (rh < rh.shift(14))
    bull_div = (pl < pl.shift(14)) & (rl > rl.shift(14))
    data["RSI_Div"] = bull_div.astype(float) - bear_div.astype(float)
    return data


def add_indicators(data):
    data = data.copy()
    close = data["Close"]

    data["Return"] = close.pct_change()
    for n in (3, 6, 12, 24):
        data[f"Ret_{n}"] = close.pct_change(n)

    for period in (9, 21, 50):
        data[f"EMA_{period}"] = close.ewm(span=period, adjust=False).mean()
        data[f"EMA{period}_Rel"] = data[f"EMA_{period}"] / close - 1

    data["RSI"] = calculate_rsi(close)

    prev_close = close.shift(1)
    true_range = pd.concat(
        [
            data["High"] - data["Low"],
            (data["High"] - prev_close).abs(),
            (data["Low"] - prev_close).abs(),
        ],
        axis=1,
    ).max(axis=1)

    data["ATR"] = true_range.ewm(alpha=1 / 14, min_periods=14, adjust=False).mean()
    data["ATR_Rel"] = data["ATR"] / close
    data["Volatility"] = data["Return"].rolling(24).std()
    data["Vol_Regime"] = (data["Volatility"]
                          / data["Volatility"].rolling(VOL_WIN, min_periods=VOL_MIN)
                          .median().replace(0, np.nan)).clip(0, 10)

    up = data["High"].diff()
    dn = -data["Low"].diff()
    plus_dm = pd.Series(np.where((up > dn) & (up > 0), up, 0.0), index=data.index)
    minus_dm = pd.Series(np.where((dn > up) & (dn > 0), dn, 0.0), index=data.index)
    pdi = 100 * plus_dm.ewm(alpha=1 / 14, min_periods=14, adjust=False).mean() / data["ATR"]
    mdi = 100 * minus_dm.ewm(alpha=1 / 14, min_periods=14, adjust=False).mean() / data["ATR"]
    dx = 100 * (pdi - mdi).abs() / (pdi + mdi).replace(0, np.nan)
    data["ADX"] = dx.ewm(alpha=1 / 14, min_periods=14, adjust=False).mean()
    data["DI_Diff"] = pdi - mdi

    macd = close.ewm(span=12, adjust=False).mean() - close.ewm(span=26, adjust=False).mean()
    macd_signal = macd.ewm(span=9, adjust=False).mean()
    data["MACD_Hist"] = (macd - macd_signal) / close

    ma20 = close.rolling(20).mean()
    sd20 = close.rolling(20).std().replace(0, np.nan)
    data["BB_Pct"] = ((close - (ma20 - 2 * sd20)) / (4 * sd20)).clip(-0.5, 1.5)

    vol_mean = data["Volume"].rolling(24).mean().replace(0, np.nan)
    data["Vol_Rel"] = (data["Volume"] / vol_mean).clip(0, 10)

    hour = data.index.hour.to_numpy()
    dow = data.index.dayofweek.to_numpy()
    data["Hour_Sin"] = np.sin(2 * np.pi * hour / 24)
    data["Hour_Cos"] = np.cos(2 * np.pi * hour / 24)
    data["Dow_Sin"] = np.sin(2 * np.pi * dow / 7)
    data["Dow_Cos"] = np.cos(2 * np.pi * dow / 7)

    data = add_ta_indicators(data)
    return data.replace([np.inf, -np.inf], np.nan)


def barrier_labels(open_, high, low, atr, side, tp_mult, sl_mult, horizon):
    """
    Etiqueta por lado con las MISMAS reglas que `simulate`:
      - la señal nace al cierre de j y se entra en la apertura de j+1;
      - TP a tp_mult*ATR[j], SL a sl_mult*ATR[j] de la entrada;
      - la vela de entrada ya cuenta para stop/TP;
      - si stop y TP caen en la misma vela, gana el stop;
      - si en `horizon` velas no se toca el TP ni el stop, la etiqueta
        es 0 (no se alcanzó el TP) y se resuelve en j+horizon;
      - NaN sólo en las últimas velas aún sin resolver.
    Devuelve también la posición en que cada etiqueta queda conocida
    (inf si aún no lo está), necesaria para evitar fugas de datos.
    """
    n = len(open_)
    y = np.full(n, np.nan)
    end = np.full(n, np.inf)

    for j in range(n - 1):
        entry = open_[j + 1]
        tp = entry + side * tp_mult * atr[j]
        sl = entry - side * sl_mult * atr[j]
        last = j + horizon
        resolved = False

        for k in range(j + 1, min(n - 1, last) + 1):
            hit_sl = low[k] <= sl if side == 1 else high[k] >= sl
            hit_tp = high[k] >= tp if side == 1 else low[k] <= tp
            if hit_sl:
                y[j], end[j], resolved = 0.0, k, True
                break
            if hit_tp:
                y[j], end[j], resolved = 1.0, k, True
                break

        if not resolved and last <= n - 1:
            y[j], end[j] = 0.0, last

    return y, end


def prepare_data(raw, symbol):
    df = add_indicators(raw)
    extra, sources = build_external(symbol, df)
    df = df.join(extra)
    # Sólo las variables de precio son obligatorias; las externas pueden
    # ser NaN (HistGradientBoosting las maneja de forma nativa).
    df = df.dropna(subset=PRICE_FEATURES + REGIME_FEATURES + ["ATR"]).copy()

    o = df["Open"].to_numpy(float)
    h = df["High"].to_numpy(float)
    l = df["Low"].to_numpy(float)
    a = df["ATR"].to_numpy(float)
    tp_mult = ATR_MULTIPLIER * RISK_REWARD

    for side, tag in ((1, "L"), (-1, "S")):
        y, end = barrier_labels(o, h, l, a, side, tp_mult,
                                ATR_MULTIPLIER, LABEL_HORIZON)
        df[f"Target_{tag}"] = y
        df[f"Label_End_{tag}"] = end
    return df, sources


def breakeven_probability():
    """
    P(TP) mínima para que la esperanza sea cero (en unidades R):
    p*RR - (1-p) = 0  ->  p = 1 / (RR + 1).
    Ignora las salidas por tiempo (aproximación).
    """
    return 1.0 / (RISK_REWARD + 1)


# ============================================================
# 1b. DATOS EXTERNOS (flujo, derivados, macro, sentimiento)
# Cada fuente es opcional: si una falla, sus columnas quedan en NaN
# (el modelo las ignora) y el panel indica qué fuentes no cargaron.
# ============================================================

def _get_json(url, params=None, timeout=20, retries=2):
    if params:
        url = url + "?" + urllib.parse.urlencode(params)
    last = None
    for k in range(retries + 1):
        try:
            req = urllib.request.Request(
                url, headers={"User-Agent": "Mozilla/5.0"})
            with urllib.request.urlopen(req, timeout=timeout) as r:
                return json.loads(r.read().decode("utf-8"))
        except Exception as e:           # red, 451 por región, timeout…
            last = e
            _time.sleep(0.5 * (k + 1))
    raise last


def _daily_cutoff():
    """Instante (epoch, s) del último cierre diario ya consultable: 00:00 UTC + margen."""
    now = pd.Timestamp.now(tz="UTC")
    cut = now.normalize() + pd.Timedelta(minutes=DAILY_CLOSE_DELAY_MIN)
    if now < cut:
        cut -= pd.Timedelta(days=1)
    return cut.timestamp()


def _daily_stamp():
    """Fecha (UTC) del último cierre diario consultable; cambia una vez al día."""
    return pd.Timestamp(_daily_cutoff(), unit="s", tz="UTC").strftime("%Y-%m-%d")


def _cache_get(key, ttl, fn, daily=False):
    """Caché con caducidad; si la descarga falla, usa la última copia.
    daily=True: además caduca al pasar el cierre diario (00:05 UTC)."""
    now = _time.time()
    hit = S.ext.get(key)
    if hit and now - hit[0] < ttl and (not daily or hit[0] >= _daily_cutoff()):
        return hit[1]
    try:
        val = fn()
    except Exception:
        if hit:
            return hit[1]
        raise
    S.ext[key] = (now, val)
    return val


def _now_ms():
    return int(pd.Timestamp.now(tz="UTC").timestamp() * 1000)


def fetch_taker(pair, start_ms, interval="1h", step_min=60):
    """Velas de Binance spot con volumen de compra agresora (taker)."""
    step_ms = int(step_min * 60_000)
    last_err, rows = None, []
    for base in BINANCE_SPOT:
        rows, cur = [], start_ms
        try:
            while cur < _now_ms():
                chunk = _get_json(f"{base}/api/v3/klines", {
                    "symbol": pair, "interval": interval,
                    "startTime": cur, "limit": 1000})
                if not chunk:
                    break
                rows += chunk
                nxt = int(chunk[-1][0]) + step_ms
                if nxt <= cur or len(chunk) < 1000:
                    break
                cur = nxt
            if rows:
                break
        except Exception as e:
            last_err, rows = e, []
    if not rows:
        raise RuntimeError(f"sin klines ({last_err})")

    df = pd.DataFrame(rows).iloc[:, [0, 5, 9]]
    df.columns = ["t", "Vol", "TakerBuy"]
    df.index = pd.to_datetime(df.pop("t").astype("int64"), unit="ms", utc=True)
    df = df.apply(pd.to_numeric, errors="coerce")
    return df[~df.index.duplicated(keep="last")].sort_index()


def _funding_binance(pair, start_ms):
    rows, cur = [], start_ms
    while cur < _now_ms():
        chunk = _get_json(f"{BINANCE_FUT}/fapi/v1/fundingRate", {
            "symbol": pair, "startTime": cur, "limit": 1000})
        if not chunk:
            break
        rows += [(int(x["fundingTime"]), float(x["fundingRate"]))
                 for x in chunk]
        nxt = rows[-1][0] + 1
        if nxt <= cur or len(chunk) < 1000:
            break
        cur = nxt
    return rows


def _funding_bybit(pair, start_ms):
    rows, end = [], _now_ms()
    while end > start_ms:
        j = _get_json("https://api.bybit.com/v5/market/funding/history", {
            "category": "linear", "symbol": pair,
            "startTime": start_ms, "endTime": end, "limit": 200})
        lst = (j.get("result") or {}).get("list") or []
        if not lst:
            break
        part = [(int(x["fundingRateTimestamp"]), float(x["fundingRate"]))
                for x in lst]
        rows += part
        oldest = min(t for t, _ in part)
        if oldest <= start_ms or len(lst) < 200:
            break
        end = oldest - 1
    return rows


def fetch_funding(pair, start_ms):
    """Funding rate real (en %). Binance Futures y, si falla, Bybit."""
    errs = []
    for name, fn in (("Binance", _funding_binance), ("Bybit", _funding_bybit)):
        try:
            rows = fn(pair, start_ms)
            if rows:
                s = pd.Series(
                    [r for _, r in rows],
                    index=pd.to_datetime([t for t, _ in rows],
                                         unit="ms", utc=True)) * 100.0
                s.index = s.index.floor("1h")
                s = s[~s.index.duplicated(keep="last")].sort_index()
                return s, name
        except Exception as e:
            errs.append(f"{name}: {e}")
    raise RuntimeError("; ".join(errs) or "sin datos")


def fetch_fng():
    """Fear & Greed Index diario (alternative.me)."""
    j = _get_json("https://api.alternative.me/fng/",
                  {"limit": 0, "format": "json"})
    data = j["data"]
    s = pd.Series(
        [float(x["value"]) for x in data],
        index=pd.to_datetime([int(x["timestamp"]) for x in data],
                             unit="s", utc=True).tz_localize(None).normalize())
    return s[~s.index.duplicated(keep="last")].sort_index()


def _yf_close(ticker, period="5y"):
    d = yf.download(ticker, period=period, interval="1d",
                    auto_adjust=True, progress=False, threads=False)
    if d is None or d.empty:
        raise RuntimeError("sin datos")
    if isinstance(d.columns, pd.MultiIndex):
        d.columns = d.columns.get_level_values(0)
    s = pd.to_numeric(d["Close"], errors="coerce").dropna()
    idx = pd.to_datetime(s.index)
    if idx.tz is not None:
        idx = idx.tz_convert("UTC").tz_localize(None)
    s.index = idx.normalize()
    return s[~s.index.duplicated(keep="last")].sort_index()


def _daily_to_hourly(daily, index):
    """
    Pasa series/columnas diarias a la rejilla horaria SIN mirar al futuro:
    el dato del día D sólo está disponible desde el día D+1.
    """
    d = daily.copy()
    d.index = pd.DatetimeIndex(d.index).normalize() + pd.Timedelta(days=1)
    d = d[~d.index.duplicated(keep="last")].sort_index()
    naive = index.tz_convert("UTC").tz_localize(None)
    out = d.reindex(naive, method="ffill")
    out.index = index
    return out


def _add_pair(out, src, names):
    for n in names:
        if n in src:
            out[n] = src[n].to_numpy(float)


def build_external(symbol, df):
    """
    Devuelve (DataFrame con EXTRA_FEATURES alineado a df.index, estado).
    estado = {fuente: (ok, detalle)}
    """
    idx = df.index
    out = pd.DataFrame(np.nan, index=idx, columns=EXTRA_FEATURES)
    status = {}
    pair = symbol.replace("-USD", "USDT")
    start_ms = int((idx[0] - pd.Timedelta(hours=72)).timestamp() * 1000)

    # 1) Flujo agresor (taker buy / sell)
    try:
        tk_int, tk_min = TF["taker"]
        # El inicio depende del histórico de la temporalidad: se incluye en
        # la clave de caché para no mezclar descargas de distinto tamaño.
        tk = _cache_get(("taker", pair, tk_int, start_ms // 3_600_000), 1800,
                        lambda: fetch_taker(pair, start_ms, tk_int, tk_min))
        # Si la vela del modelo es más larga que la de Binance, agregar los
        # volúmenes antes de alinearlos; no reindexar directamente (daría
        # casi todo NaN).
        if BAR_MIN > tk_min:
            tk = tk.resample(
                BAR_FREQ, origin="epoch", label="left", closed="left"
            ).sum(min_count=1)
        vol, buy = tk["Vol"], tk["TakerBuy"]
        net = 2 * buy - vol
        imb = (net / vol.replace(0, np.nan)).clip(-1, 1)
        imb24 = net.rolling(24, min_periods=1).sum() / vol.rolling(24, min_periods=1).sum().replace(0, np.nan)
        f = pd.DataFrame({"Taker_Imb": imb, "Taker_Imb_24": imb24})
        f = f.reindex(idx)
        _add_pair(out, f, ["Taker_Imb", "Taker_Imb_24"])
        status["Flujo taker"] = (True, f"Binance spot {tk_int}")
    except Exception as e:
        status["Flujo taker"] = (False, str(e)[:80])

    # 2) Funding rate real
    try:
        fs, src = _cache_get(("funding", pair), 1800,
                             lambda: fetch_funding(pair, start_ms))
        # disponible desde la liquidación: se alinea con el CIERRE de la vela
        h = fs.reindex(idx + pd.Timedelta(minutes=BAR_MIN), method="ffill")
        h.index = idx
        # El funding cambia cada 8 h: en velas de menos de 1 h las ventanas
        # se escalan para cubrir el mismo tiempo real (30 días / 24 h).
        k = 1.0 / BAR_HOURS if BAR_HOURS < 1 else 1.0
        fw, fmin, fav = int(720 * k), int(240 * k), max(24, int(24 * k))
        z = ((h - h.rolling(fw, min_periods=fmin).mean())
             / h.rolling(fw, min_periods=fmin).std().replace(0, np.nan))
        out["Funding"] = h.to_numpy(float)
        out["Funding_Z"] = z.to_numpy(float)
        out["Funding_Avg24"] = h.rolling(fav, min_periods=1).mean().to_numpy(float)
        status["Funding"] = (True, src)
    except Exception as e:
        status["Funding"] = (False, str(e)[:80])

    # 3) Contexto diario del propio activo (tendencia de fondo)
    dclose = None
    try:
        per = "8y" if BAR_MIN >= 1440 else "5y"
        dclose = _cache_get(("yf", symbol, per), 6 * 3600,
                            lambda: _yf_close(symbol, per), daily=True)
        d = pd.DataFrame(index=dclose.index)
        d["D_EMA200_Rel"] = dclose / dclose.ewm(span=200, adjust=False).mean() - 1
        d["D_Dist_High50"] = dclose / dclose.rolling(50).max() - 1
        d["D_Dist_Low50"] = dclose / dclose.rolling(50).min() - 1
        d["D_Ret_7"] = dclose.pct_change(7)
        d["D_Ret_30"] = dclose.pct_change(30)
        d["D_RSI"] = calculate_rsi(dclose)
        _add_pair(out, _daily_to_hourly(d, idx), list(d.columns))
        status["Contexto diario"] = (True, f"yfinance {per[:-1]} años")
    except Exception as e:
        status["Contexto diario"] = (False, str(e)[:80])

    # 4) Macro: S&P500, dólar, VIX, bono 10 años
    macro = {"SPX": "^GSPC", "DXY": "DX-Y.NYB", "VIX": "^VIX", "TNX": "^TNX"}
    series, bad = {}, []
    for name, tk_ in macro.items():
        try:
            series[name] = _cache_get(("yf", tk_), 6 * 3600,
                                      lambda t=tk_: _yf_close(t), daily=True)
        except Exception:
            bad.append(name)
    if series:
        m = pd.DataFrame(index=pd.DatetimeIndex([]))
        if "SPX" in series:
            m = m.join(pd.DataFrame({
                "SPX_Ret_1d": series["SPX"].pct_change(),
                "SPX_Ret_5d": series["SPX"].pct_change(5)}), how="outer")
            if dclose is not None:
                both = pd.concat([dclose.pct_change(), series["SPX"].pct_change()],
                                 axis=1, join="inner").dropna()
                both.columns = ["a", "b"]
                m = m.join(pd.DataFrame(
                    {"Corr_SPX_30": both["a"].rolling(30).corr(both["b"])}),
                    how="outer")
        if "DXY" in series:
            m = m.join(pd.DataFrame({
                "DXY_Ret_5d": series["DXY"].pct_change(5)}), how="outer")
        if "VIX" in series:
            m = m.join(pd.DataFrame({
                "VIX": series["VIX"],
                "VIX_Chg_5d": series["VIX"].pct_change(5)}), how="outer")
        if "TNX" in series:
            m = m.join(pd.DataFrame({
                "TNX_Chg_5d": series["TNX"].diff(5)}), how="outer")
        m = m.sort_index().ffill(limit=5)
        _add_pair(out, _daily_to_hourly(m, idx), list(m.columns))
    status["Macro"] = (
        (len(series) > 0),
        ("yfinance: " + ", ".join(series)
         + (f" · fallan {', '.join(bad)}" if bad else "")) if series
        else "ninguno disponible")

    # 5) Fear & Greed
    try:
        fng = _cache_get(("fng",), 6 * 3600, fetch_fng, daily=True)
        g = pd.DataFrame({"FNG": fng, "FNG_Chg_7": fng.diff(7)})
        _add_pair(out, _daily_to_hourly(g, idx), ["FNG", "FNG_Chg_7"])
        status["Fear & Greed"] = (True, "alternative.me")
    except Exception as e:
        status["Fear & Greed"] = (False, str(e)[:80])

    # 6) Activo de referencia (BTC para ETH/SOL; ETH para BTC)
    peer = "ETH-USD" if symbol == "BTC-USD" else "BTC-USD"
    try:
        pc = _cache_get(("peer", peer, INTERVAL), 1800,
                        lambda: download_data(peer)[0]["Close"])
        pc = pc.reindex(idx)
        out["Peer_Ret_6"] = pc.pct_change(6).to_numpy(float)
        out["Peer_Ret_24"] = pc.pct_change(24).to_numpy(float)
        out["Rel_24"] = (df["Ret_24"] - pc.pct_change(24)).to_numpy(float)
        status["Activo ref."] = (True, peer)
    except Exception as e:
        status["Activo ref."] = (False, str(e)[:80])

    # 7) Sentimiento nuevo: atención (Wikipedia) y tono de prensa (GDELT)
    if USE_SENTIMENT:
        try:
            add_sentiment_features(out, symbol, idx, status)
        except Exception as e:
            status["Sentimiento"] = (False, str(e)[:80])

    out = out.replace([np.inf, -np.inf], np.nan)
    return out, status


# ============================================================
# 1c. ANÁLISIS DE SENTIMIENTO (noticias, atención, miedo/codicia, derivados)
# ============================================================
# Dos usos combinados:
#  a) Variables para el MODELO (históricas, sin mirar al futuro): atención en
#     Wikipedia y tono de la prensa (GDELT) -> entran junto al resto de datos externos.
#  b) Panel «Análisis de sentimiento»: titulares recientes puntuados + índice
#     compuesto (-100 … +100) que mezcla noticias, prensa, Fear & Greed y
#     posicionamiento en derivados, y lo contrasta con la señal del modelo.

try:
    from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer
    _VADER = SentimentIntensityAnalyzer()
except Exception:
    _VADER = None

import re
import html as _html
import xml.etree.ElementTree as ET
from concurrent.futures import ThreadPoolExecutor

# Léxico específico de cripto (puntuación VADER aprox. -4 … +4)
_CRYPTO_LEX = {
    "surge": 2.5, "surges": 2.5, "soar": 2.8, "soars": 2.8, "rally": 2.3, "rallies": 2.3,
    "breakout": 2.0, "bullish": 2.6, "bull": 1.5, "moon": 1.8, "adoption": 1.6,
    "inflow": 1.8, "inflows": 1.8, "accumulate": 1.6, "accumulation": 1.6,
    "approval": 1.8, "approves": 1.8, "approved": 1.8, "halving": 1.0, "ath": 2.4,
    "record": 1.5, "upgrade": 1.2, "partnership": 1.2, "etf": 0.8, "gains": 1.8,
    "rebound": 1.8, "recovers": 1.6, "recovery": 1.5, "jumps": 1.8, "climbs": 1.6,
    "crash": -3.0, "crashes": -3.0, "plunge": -2.8, "plunges": -2.8, "dump": -2.4,
    "dumps": -2.4, "bearish": -2.6, "bear": -1.5, "hack": -2.8, "hacked": -2.8,
    "exploit": -2.4, "ban": -2.2, "bans": -2.2, "lawsuit": -2.0, "sues": -2.0,
    "fraud": -3.0, "scam": -2.8, "liquidation": -2.2, "liquidations": -2.2,
    "outflow": -1.8, "outflows": -1.8, "selloff": -2.5, "sell-off": -2.5,
    "tumbles": -2.5, "slumps": -2.3, "drops": -1.8, "falls": -1.6, "fear": -1.6,
    "bankrupt": -3.0, "bankruptcy": -3.0, "insolvency": -2.8, "rejects": -1.6,
    "delay": -1.0, "delays": -1.0, "crackdown": -2.2, "investigation": -1.5,
    "warns": -1.2, "warning": -1.2, "risk": -0.8, "volatility": -0.4,
}
# Lista de respaldo si VADER no está disponible: mismas palabras, sin matices
_BASIC_NEG = {"not", "no", "never", "without", "fails", "failed"}

_ASSET_KEYS = {
    "BTC-USD": (r"bitcoin|btc", "bitcoin"),
    "ETH-USD": (r"ethereum|\bether\b|\beth\b", "ethereum"),
    "SOL-USD": (r"solana|\bsol\b", "solana"),
}

_FEEDS = [
    ("CoinDesk", "https://www.coindesk.com/arc/outboundfeeds/rss/"),
    ("Cointelegraph", "https://cointelegraph.com/rss"),
    ("Decrypt", "https://decrypt.co/feed"),
    ("Bitcoin Magazine", "https://bitcoinmagazine.com/.rss/full/"),
]


def score_headline(text):
    """Puntuación de sentimiento de un titular en [-1, 1]."""
    t = (text or "").strip()
    if not t:
        return 0.0
    if _VADER is not None:
        if not getattr(_VADER, "_qd_lex", False):
            _VADER.lexicon.update(_CRYPTO_LEX)
            _VADER._qd_lex = True
        return float(_VADER.polarity_scores(t)["compound"])
    words = re.findall(r"[a-z\-]+", t.lower())
    tot, neg = 0.0, False
    for i, w in enumerate(words):
        v = _CRYPTO_LEX.get(w, 0.0)
        if v and i > 0 and words[i - 1] in _BASIC_NEG:
            v = -v
        tot += v
    return float(np.tanh(tot / 4.0))


def _parse_rss(xml_text, source):
    root = ET.fromstring(xml_text)
    items = []
    for it in root.iter("item"):
        title = _html.unescape((it.findtext("title") or "").strip())
        link = (it.findtext("link") or "").strip()
        pub = it.findtext("pubDate") or ""
        try:
            ts = pd.to_datetime(pub, utc=True)
        except Exception:
            ts = pd.Timestamp.now(tz="UTC")
        if title:
            items.append({"title": title, "link": link, "source": source, "time": ts})
    return items


def _fetch_feed(name, url):
    req = urllib.request.Request(url, headers={"User-Agent": "Mozilla/5.0"})
    with urllib.request.urlopen(req, timeout=7) as r:
        return _parse_rss(r.read().decode("utf-8", "ignore"), name)


def fetch_news(symbol):
    """
    Titulares recientes del activo (RSS de prensa cripto + Google News), cada uno con
    su puntuación. Devuelve (DataFrame, [fuentes ok], [fuentes caídas]).
    """
    pat, query = _ASSET_KEYS.get(symbol, (symbol.split("-")[0].lower(), symbol.split("-")[0]))
    feeds = list(_FEEDS) + [("Google News", "https://news.google.com/rss/search?q="
                             + urllib.parse.quote(query + " crypto when:2d")
                             + "&hl=en-US&gl=US&ceid=US:en")]
    items, ok, bad = [], [], []

    def one(f):
        try:
            return f[0], _fetch_feed(*f), None
        except Exception as e:
            return f[0], [], str(e)[:60]

    with ThreadPoolExecutor(max_workers=len(feeds)) as ex:
        for name, res, err in ex.map(one, feeds):
            (bad if err else ok).append(name)
            items += res
    if not items:
        raise RuntimeError("sin titulares (" + ", ".join(bad) + ")")
    d = pd.DataFrame(items).drop_duplicates("title")
    rx = re.compile(pat, re.I)
    d = d[d["title"].str.contains(rx) | (d["source"] == "Google News")].copy()
    now = pd.Timestamp.now(tz="UTC")
    d = d[d["time"] >= now - pd.Timedelta(days=3)]
    if d.empty:
        raise RuntimeError("ningún titular reciente sobre el activo")
    d["score"] = d["title"].map(score_headline)
    d["age_h"] = (now - d["time"]).dt.total_seconds() / 3600
    d["w"] = 0.5 ** (d["age_h"].clip(lower=0) / 24.0)         # vida media 24 h
    return d.sort_values("time", ascending=False).reset_index(drop=True), ok, bad


def news_sentiment(symbol):
    """Media ponderada por recencia (-1…+1) y el detalle; None si no hay red."""
    d, ok, bad = _cache_get(("news", symbol), 900, lambda: fetch_news(symbol))
    score = float((d["score"] * d["w"]).sum() / d["w"].sum())
    return score, d, ok, bad


def fetch_wiki_views(symbol, days=900):
    """Visitas diarias a la página de Wikipedia del activo (atención del público)."""
    art = {"BTC-USD": "Bitcoin", "ETH-USD": "Ethereum", "SOL-USD": "Solana_(blockchain_platform)"}
    art = art.get(symbol, symbol.split("-")[0])
    end = pd.Timestamp.now(tz="UTC").normalize() - pd.Timedelta(days=1)
    start = end - pd.Timedelta(days=days)
    url = ("https://wikimedia.org/api/rest_v1/metrics/pageviews/per-article/"
           f"en.wikipedia/all-access/user/{art}/daily/{start:%Y%m%d}/{end:%Y%m%d}")
    j = _get_json(url, timeout=15)
    items = j["items"]
    s = pd.Series([float(x["views"]) for x in items],
                  index=pd.to_datetime([x["timestamp"][:8] for x in items]))
    return s[~s.index.duplicated(keep="last")].sort_index()


def fetch_gdelt_tone(symbol):
    """Tono medio diario de la prensa mundial sobre el activo (GDELT; ~últimos 90 días)."""
    q = {"BTC-USD": "bitcoin", "ETH-USD": "ethereum", "SOL-USD": "solana"}.get(
        symbol, symbol.split("-")[0].lower())
    base = "https://api.gdeltproject.org/api/v2/doc/doc"
    j = _get_json(base, {"query": q, "mode": "timelinetone", "timespan": "90d",
                         "format": "json"}, timeout=25, retries=1)
    data = j["timeline"][0]["data"]
    s = pd.Series([float(x["value"]) for x in data],
                  index=pd.to_datetime([x["date"] for x in data], utc=True)
                  .tz_localize(None))
    s = s.resample("1D").mean().dropna()
    if len(s) < 20:
        raise RuntimeError("serie demasiado corta")
    return s


def add_sentiment_features(out, symbol, idx, status):
    """Rellena las SENT_FEATURES (históricas y sin mirar al futuro) y anota el estado."""
    # Atención en Wikipedia
    try:
        w = _cache_get(("wiki", symbol), 6 * 3600, lambda: fetch_wiki_views(symbol),
                       daily=True)
        lw = np.log1p(w)
        d = pd.DataFrame({
            "Wiki_Z": (lw - lw.rolling(30, min_periods=15).mean())
                      / lw.rolling(30, min_periods=15).std().replace(0, np.nan),
            "Wiki_Chg_7": lw.diff(7)})
        _add_pair(out, _daily_to_hourly(d, idx), list(d.columns))
        status["Atención (Wikipedia)"] = (True, "visitas diarias")
    except Exception as e:
        status["Atención (Wikipedia)"] = (False, str(e)[:80])
    # Tono de la prensa (GDELT)
    try:
        t = _cache_get(("gdelt", symbol), 6 * 3600, lambda: fetch_gdelt_tone(symbol),
                       daily=True)
        d = pd.DataFrame({
            "Press_Tone": t,
            "Press_Tone_Z": (t - t.rolling(30, min_periods=10).mean())
                            / t.rolling(30, min_periods=10).std().replace(0, np.nan)})
        _add_pair(out, _daily_to_hourly(d, idx), list(d.columns))
        status["Tono de prensa (GDELT)"] = (True, f"{len(t)} días")
    except Exception as e:
        status["Tono de prensa (GDELT)"] = (False, str(e)[:80])


def sentiment_snapshot(symbol, df, st):
    """
    Componentes del índice compuesto. Cada uno: (nombre, valor -1…+1, texto, detalle).
    Sólo se incluyen los que tienen dato.
    """
    r = df.iloc[-1]
    comps, news_df, news_ok, news_bad, news_err = [], None, [], [], None

    def g(n):
        v = r.get(n, np.nan)
        return float(v) if v is not None and np.isfinite(v) else np.nan

    # 1) Titulares
    try:
        sc, news_df, news_ok, news_bad = news_sentiment(symbol)
        comps.append(("Titulares recientes", sc, f"{sc * 100:+.0f}",
                      f"{len(news_df)} titulares · media ponderada por recencia"))
    except Exception as e:
        news_err = str(e)[:90]

    # 2) Tono de prensa (GDELT)
    z = g("Press_Tone_Z")
    if np.isfinite(z):
        comps.append(("Tono de la prensa mundial", float(np.tanh(z / 2)), f"{z:+.1f} σ",
                      f"tono medio {g('Press_Tone'):+.2f} frente a sus últimos 30 días"))

    # 3) Fear & Greed
    fng = g("FNG")
    if np.isfinite(fng):
        lab = ("Miedo extremo" if fng < 25 else "Miedo" if fng < 45 else
               "Neutral" if fng <= 55 else "Codicia" if fng <= 75 else "Codicia extrema")
        comps.append(("Fear & Greed", (fng - 50) / 50, f"{fng:.0f}/100", lab))

    # 4) Posicionamiento en derivados (funding)
    fz = g("Funding_Z")
    if np.isfinite(fz):
        comps.append(("Funding (apalancamiento)", float(np.tanh(fz / 2)), f"{fz:+.1f} σ",
                      "positivo = longs pagan a shorts (apuesta alcista)"))

    # 5) Presión agresora (taker)
    ti = g("Taker_Imb_24")
    if np.isfinite(ti):
        comps.append(("Flujo comprador/vendedor 24 h", float(np.clip(ti * 5, -1, 1)),
                      f"{ti * 100:+.1f} %", "compras agresoras frente a ventas agresoras"))

    # Atención (informativa, no direccional)
    wz = g("Wiki_Z")
    attention = None
    if np.isfinite(wz):
        attention = (wz, "Muy alta" if wz > 1.5 else "Alta" if wz > 0.5 else
                     "Baja" if wz < -0.5 else "Normal")
    return comps, attention, news_df, (news_ok, news_bad, news_err)


def _sent_label(x):
    if x >= 0.45:
        return "EUFORIA", GREEN
    if x >= 0.15:
        return "OPTIMISMO", GREEN
    if x <= -0.45:
        return "PÁNICO", RED
    if x <= -0.15:
        return "PESIMISMO", RED
    return "NEUTRAL", GOLD


def _bar(v):
    """Barra centrada: izquierda = pesimismo, derecha = optimismo."""
    col = GREEN if v > 0.05 else RED if v < -0.05 else GOLD
    w = abs(v) * 50
    left = 50 if v >= 0 else 50 - w
    return ("<div style='position:relative;height:8px;border-radius:4px;background:"
            f"{GRID_COLOR};min-width:90px'>"
            f"<div style='position:absolute;left:{left:.1f}%;width:{w:.1f}%;top:0;bottom:0;"
            f"border-radius:4px;background:{col}'></div>"
            f"<div style='position:absolute;left:50%;top:-2px;bottom:-2px;width:1px;"
            f"background:{MUTED}'></div></div>")


def sentiment_composite(comps):
    return float(np.mean([c[1] for c in comps])) if comps else float("nan")


def sentiment_section_html(df, st, symbol=None, sources=None):
    symbol = symbol or DEFAULT_SYMBOL
    try:
        comps, attention, news_df, (n_ok, n_bad, n_err) = sentiment_snapshot(symbol, df, st)
    except Exception as exc:
        return section("Análisis de sentimiento", f"No se pudo calcular: {exc}")
    if not comps:
        return section("Análisis de sentimiento",
                       "Sin datos de sentimiento disponibles ahora mismo (¿sin conexión a "
                       "las fuentes?). El resto del dashboard no se ve afectado.")

    comp = sentiment_composite(comps)
    label, lcol = _sent_label(comp)
    sig = st.get("signal", "ESPERAR")
    sig_side = 1 if sig == "LONG" else -1 if sig == "SHORT" else 0
    s_side = 1 if comp >= 0.15 else -1 if comp <= -0.15 else 0
    if sig_side == 0:
        conf, ccol, cnote = "Sin señal", MUTED, "el modelo espera; el sentimiento no abre operaciones"
    elif s_side == 0:
        conf, ccol, cnote = "Neutral", GOLD, "sentimiento sin dirección clara"
    elif s_side == sig_side:
        conf, ccol, cnote = "Coincide", GREEN, f"el sentimiento respalda la señal {sig}"
    else:
        conf, ccol, cnote = "Diverge", RED, f"el sentimiento va contra la señal {sig}: más cautela"
    extreme = abs(comp) >= 0.6
    n_pos = sum(1 for c in comps if c[1] >= 0.15)
    n_neg = sum(1 for c in comps if c[1] <= -0.15)

    cards = [
        card("Sentimiento compuesto", f"{label.capitalize()}", f"{comp * 100:+.0f} / 100 · "
             f"{n_pos} optimistas · {n_neg} pesimistas · {len(comps) - n_pos - n_neg} neutrales",
             lcol),
        card("Frente a la señal del modelo", conf, cnote, ccol),
    ]
    if attention:
        wz, wl = attention
        cards.append(card("Atención del público", wl, f"visitas a Wikipedia {wz:+.1f} σ vs. 30 d "
                          "(mide interés, no dirección)", BLUE))
    if extreme:
        cards.append(card("Extremo de sentimiento", "Riesgo contrario",
                          "las lecturas extremas suelen preceder giros: no perseguir el precio",
                          GOLD))

    rows = "".join(
        f"<tr><td>{n}</td><td style='color:{GREEN if v >= .15 else RED if v <= -.15 else GOLD}'>"
        f"<b>{t}</b></td><td style='width:34%'>{_bar(v)}</td><td>{note}</td></tr>"
        for n, v, t, note in comps)
    comp_table = ("<div class='qd-tw'><table class='qd-table'><tr><th>Fuente</th><th>Lectura</th>"
                  "<th>Pesimismo ◄ ► Optimismo</th><th>Detalle</th></tr>" + rows + "</table></div>")

    if news_df is not None and len(news_df):
        nr = []
        for r in news_df.head(8).itertuples():
            col = GREEN if r.score >= 0.15 else RED if r.score <= -0.15 else GOLD
            tag = "POSITIVO" if r.score >= 0.15 else "NEGATIVO" if r.score <= -0.15 else "NEUTRAL"
            age = (f"{r.age_h:.0f} h" if r.age_h < 48 else f"{r.age_h / 24:.0f} d")
            title = _html.escape(r.title)
            if r.link.startswith("http"):
                title = (f"<a href='{_html.escape(r.link)}' target='_blank' rel='noopener' "
                         f"style='color:{TEXT};text-decoration:none'>{title}</a>")
            nr.append(f"<tr><td>{badge(tag, col)}</td><td style='color:{col}'>{r.score:+.2f}</td>"
                      f"<td>{title}</td><td>{_html.escape(r.source)}</td><td>{age}</td></tr>")
        news_table = ("<div class='qd-tw'><table class='qd-table'><tr><th>Lectura</th>"
                      "<th>Puntos</th><th>Titular</th><th>Fuente</th><th>Hace</th></tr>"
                      + "".join(nr) + "</table></div>")
    else:
        news_table = (f"<div class='qd-empty'>Sin titulares: {n_err or 'fuentes no disponibles'}."
                      "</div>")

    engine = "VADER + léxico cripto" if _VADER is not None else "léxico cripto propio (sin VADER)"
    feeds = (f"RSS ok: {', '.join(n_ok)}" if n_ok else "RSS: ninguno disponible") + (
        f" · caídos: {', '.join(n_bad)}" if n_bad else "")
    desc = (f"Índice de -100 (pánico) a +100 (euforia) que promedia las fuentes con dato: "
            f"titulares puntuados con {engine}, tono de la prensa (GDELT), Fear & Greed, "
            "funding y flujo taker. Es contexto: el modelo ya recibe como variables la "
            "atención del público, el tono de la prensa, Fear & Greed y el funding, y la señal "
            "LONG/SHORT sigue decidiéndola él. El sentimiento extremo suele ser contrario. "
            f"{feeds}.")
    return (section("Análisis de sentimiento", desc, justify=True) + grid(cards, 200)
            + comp_table + "<div style='height:8px'></div>" + news_table)


# ============================================================
# 2. MODELO Y VALIDACIÓN
# ============================================================

def new_model():
    # Sin class_weight (probabilidades calibradas) y sin early stopping
    # (mismo modelo en validación, walk-forward y modelo en vivo).
    return HistGradientBoostingClassifier(
        max_depth=4, learning_rate=0.08, max_iter=70,
        min_samples_leaf=40, l2_regularization=1.0,
        early_stopping=False, random_state=42,
    )


def proba_pos(model, X):
    classes = list(model.classes_)
    if 1 not in classes:
        return np.zeros(len(X))
    return model.predict_proba(X)[:, classes.index(1)]


def walk_forward_probs(X, labels):
    """P(TP) por lado para cada vela usando sólo etiquetas ya resueltas."""
    n = len(X)
    start = max(WF_MIN_START, int(n * WF_START_FRAC))
    if n - start < 200:
        raise RuntimeError("Histórico insuficiente para el backtest.")

    p = {side: np.full(n, np.nan) for side in labels}
    total = max(1, (n - 1 - start) // WF_REFIT_BARS + 1)

    for step, b in enumerate(range(start, n - 1, WF_REFIT_BARS), 1):
        set_status(f"Walk-forward: bloque {step}/{total}…")
        e = min(b + WF_REFIT_BARS, n - 1)

        for side, (y, end) in labels.items():
            idx = np.where(np.isfinite(y) & (end <= b))[0][-MAX_TRAIN_SAMPLES:]
            if len(idx) < MIN_TRAIN_SAMPLES or len(np.unique(y[idx])) < 2:
                continue
            model = new_model().fit(X[idx], y[idx].astype(int))
            p[side][b:e] = proba_pos(model, X[b:e])

    return p, start


def validate_side(X, y):
    """Validación cronológica (con hueco = horizonte) frente a la base."""
    idx = np.where(np.isfinite(y))[0]
    Xl, yl = X[idx], y[idx].astype(int)

    rows = []
    for tr, te in TimeSeriesSplit(n_splits=4, gap=LABEL_HORIZON).split(Xl):
        if len(np.unique(yl[tr])) < 2:
            continue
        model = new_model().fit(Xl[tr], yl[tr])
        p = np.clip(proba_pos(model, Xl[te]), 1e-6, 1 - 1e-6)
        base = float(np.clip(yl[tr].mean(), 1e-6, 1 - 1e-6))
        yt = yl[te]

        try:
            auc = roc_auc_score(yt, p)
        except ValueError:
            auc = np.nan

        rows.append({
            "rate": float(yt.mean()),
            "auc": auc,
            "logloss": log_loss(yt, p, labels=[0, 1]),
            "ll_base": log_loss(yt, np.full(len(yt), base), labels=[0, 1]),
            "brier": brier_score_loss(yt, p),
            "br_base": float(np.mean((base - yt) ** 2)),
        })

    if not rows:
        raise RuntimeError("No se pudo validar el modelo.")

    out = pd.DataFrame(rows).mean(numeric_only=True).to_dict()
    out["skill"] = 1 - out["logloss"] / out["ll_base"]
    out["folds"] = len(rows)
    out["samples"] = int(len(idx))
    return out


def _bp(local, msg=None, creep=None):
    """Avance dentro de get_bundle (0-100) -> tramo 5-65 % de la barra de carga."""
    load_progress(5 + 0.60 * local, msg,
                  None if creep is None else 5 + 0.60 * creep)


def get_bundle(symbol):
    """Descarga y entrena. Sólo reentrena si hay una vela cerrada nueva."""
    with _LOCK:
        _bp(0, f"Descargando datos de {symbol}…", creep=18)
        raw, gaps = download_data(symbol)
        key = (symbol, raw.index[-1], CONFIG_VERSION, _daily_stamp())
        _bp(20, "Datos descargados")

        cached = S.bundles.get(symbol)
        if cached is not None and cached["key"] == key:
            return cached

        set_status(f"{symbol}: vela nueva, preparando etiquetas…")
        _bp(25, "Preparando variables y etiquetas…", creep=38)
        df, sources = prepare_data(raw, symbol)

        # Descarta las columnas externas sin datos (fuente caída): un
        # modelo con columnas 100 % NaN falla en scikit-learn.
        active = [f for f in FEATURES if df[f].notna().mean() > 0.5]
        X = df[active].to_numpy(float)
        labels = {
            "LONG": (df["Target_L"].to_numpy(float),
                     df["Label_End_L"].to_numpy(float)),
            "SHORT": (df["Target_S"].to_numpy(float),
                      df["Label_End_S"].to_numpy(float)),
        }

        set_status(f"{symbol}: entrenando y validando (20-60 s)…")
        _bp(40, "Entrenando y validando el modelo…", creep=75)
        p_wf, start = walk_forward_probs(X, labels)

        validation = {side: validate_side(X, y)
                      for side, (y, _) in labels.items()}
        validation["samples"] = validation["LONG"]["samples"]

        # Control: mismo modelo con SÓLO las 18 variables de precio (sin AT ampliado ni datos externos).
        # Si los datos extra no mejoran esto, son ruido.
        set_status(f"{symbol}: comparando con el modelo sólo-precio…")
        _bp(78, "Comparando con el modelo sólo-precio…", creep=90)
        Xp = df[PRICE_FEATURES].to_numpy(float)
        validation["ablation"] = {side: validate_side(Xp, y)
                                  for side, (y, _) in labels.items()}

        _bp(92, "Calculando probabilidades actuales…")
        p_live, base = {}, {}
        for side, (y, _) in labels.items():
            idx = np.where(np.isfinite(y))[0][-MAX_TRAIN_SAMPLES:]
            base[side] = float(y[idx].mean())   # tasa base de TP por lado
            model = new_model().fit(X[idx], y[idx].astype(int))
            p_live[side] = float(proba_pos(model, X[-1:])[0])

        arrays = {
            "open": df["Open"].to_numpy(float),
            "high": df["High"].to_numpy(float),
            "low": df["Low"].to_numpy(float),
            "close": df["Close"].to_numpy(float),
            "atr": df["ATR"].to_numpy(float),
        }

        bundle = {
            "key": key, "config": CONFIG_VERSION, "symbol": symbol,
            "df": df, "arrays": arrays, "gaps": gaps, "sources": sources,
            "p_wf": p_wf, "start": start, "validation": validation,
            "p_live": p_live, "base": base, "breakeven": breakeven_probability(),
            "built_at": _time.time(),
        }
        S.bundles[symbol] = bundle
        S.sim_key = None
        _bp(100)
        return bundle


# ============================================================
# 3. RIESGO, POSICIÓN Y SEÑAL
# ============================================================

def calculate_position(capital, entry, stop, risk_pct, max_position_pct):
    zero = {"qty": 0.0, "notional": 0.0, "risk": 0.0, "capped": False}
    risk_cash = capital * risk_pct / 100.0
    distance = abs(entry - stop)

    if (not np.isfinite(distance) or distance <= 0
            or entry <= 0 or capital <= 0):
        return zero

    qty_by_risk = risk_cash / distance
    qty_by_notional = capital * max_position_pct / 100.0 / entry
    qty = max(0.0, min(qty_by_risk, qty_by_notional))

    return {"qty": qty, "notional": qty * entry, "risk": qty * distance,
            "capped": qty_by_notional < qty_by_risk}


def decide_signal(p_long, p_short, threshold):
    if not (np.isfinite(p_long) and np.isfinite(p_short)):
        return "ESPERAR"
    if max(p_long, p_short) < threshold or p_long == p_short:
        return "ESPERAR"
    return "LONG" if p_long > p_short else "SHORT"


def signals_from_probs(p_long, p_short, threshold):
    sig = np.zeros(len(p_long), dtype=int)
    ok = np.isfinite(p_long) & np.isfinite(p_short)
    sig[ok & (p_long >= threshold) & (p_long > p_short)] = 1
    sig[ok & (p_short >= threshold) & (p_short > p_long)] = -1
    return sig


def effective_threshold(bundle, params):
    if params["auto"]:
        return float(min(0.90, bundle["breakeven"] + AUTO_MARGIN))
    return float(params["threshold"])


def live_state(bundle, params):
    df = bundle["df"]
    last = df.iloc[-1]
    price, atr, rsi = float(last["Close"]), float(last["ATR"]), float(last["RSI"])

    if not np.isfinite(atr) or atr <= 0:
        raise RuntimeError("El ATR no es válido para calcular stops.")

    dist = ATR_MULTIPLIER * atr
    levels = {
        "LONG": (price - dist, price + dist * RISK_REWARD),
        "SHORT": (price + dist, price - dist * RISK_REWARD),
    }
    sizes = {
        side: calculate_position(params["capital"], price, levels[side][0],
                                 params["risk"], params["maxpos"])
        for side in levels
    }

    thr = effective_threshold(bundle, params)
    p_long, p_short = bundle["p_live"]["LONG"], bundle["p_live"]["SHORT"]

    if last["EMA_9"] > last["EMA_21"] > last["EMA_50"]:
        trend = "ALCISTA"
    elif last["EMA_9"] < last["EMA_21"] < last["EMA_50"]:
        trend = "BAJISTA"
    else:
        trend = "MIXTA"

    # Sesgo direccional: cuánto se desvía cada P(TP) de SU tasa base.
    # Comparar P(LONG) con P(SHORT) en bruto casi siempre da "iguales"
    # (ambas rondan la tasa base ~33 %); lo informativo es la desviación.
    base = bundle["base"]
    edge_long, edge_short = p_long - base["LONG"], p_short - base["SHORT"]
    bias = edge_long - edge_short
    bias_label = ("ALCISTA" if bias > BIAS_DEADZONE
                  else "BAJISTA" if bias < -BIAS_DEADZONE else "NEUTRO")

    return {
        "price": price, "atr": atr, "rsi": rsi, "trend": trend,
        "levels": levels, "sizes": sizes, "threshold": thr,
        "base_long": base["LONG"], "base_short": base["SHORT"],
        "edge_long": edge_long, "edge_short": edge_short,
        "bias": bias, "bias_label": bias_label,
        "p_long": p_long, "p_short": p_short,
        "signal": decide_signal(p_long, p_short, thr),
        "ts": df.index[-1],
    }


# ============================================================
# 4. BACKTEST VELA A VELA (High/Low, stop, TP)
# ============================================================

def simulate(A, sig, start, capital, risk_pct, max_pos_pct, record=True):
    """
    La señal se calcula al cierre de la vela t y se ejecuta en la
    apertura de t+1. Stop y take profit se evalúan con High/Low; si
    ambos caen en la misma vela se asume que se toca primero el stop.
    Una sola posición a la vez. Tamaño según riesgo y exposición máxima.
    """
    o = A["open"].tolist()
    h = A["high"].tolist()
    l = A["low"].tolist()
    c = A["close"].tolist()
    a = A["atr"].tolist()
    sig = np.asarray(sig).tolist()
    n = len(c)

    equity = float(capital)
    curve = np.empty(n - start)
    trades = []
    pos = None
    pending = 0

    for t in range(start, n):
        # --- Entrada (apertura de la vela t) ---
        if pending != 0 and pos is None:
            d = pending
            entry = o[t]
            dist = ATR_MULTIPLIER * a[t - 1]
            stop = entry - d * dist
            target = entry + d * dist * RISK_REWARD
            size = calculate_position(equity, entry, stop, risk_pct, max_pos_pct)
            qty = size["qty"]

            if qty > 0:
                pos = {
                    "d": d, "entry": entry, "stop": stop, "target": target,
                    "qty": qty, "t0": t, "eq0": equity,
                    "risk_eff": size["risk"] / equity if equity > 0 else 0.0,
                }
        pending = 0

        # --- Salida ---
        if pos is not None:
            d = pos["d"]
            bars = t - pos["t0"] + 1
            exit_price, reason = None, None
            gap_ok = t > pos["t0"]

            if d == 1:
                if l[t] <= pos["stop"]:
                    exit_price = min(pos["stop"], o[t]) if gap_ok else pos["stop"]
                    reason = "STOP"
                elif h[t] >= pos["target"]:
                    exit_price, reason = pos["target"], "TP"
            else:
                if h[t] >= pos["stop"]:
                    exit_price = max(pos["stop"], o[t]) if gap_ok else pos["stop"]
                    reason = "STOP"
                elif l[t] <= pos["target"]:
                    exit_price, reason = pos["target"], "TP"

            if exit_price is None and (bars >= MAX_HOLD_BARS or t == n - 1):
                exit_price = c[t]
                reason = "TIEMPO" if bars >= MAX_HOLD_BARS else "FIN"

            if exit_price is not None:
                qty = pos["qty"]
                pnl = d * qty * (exit_price - pos["entry"])
                equity += pnl

                if record:
                    trades.append({
                        "t0": pos["t0"], "t1": t, "Direction": d,
                        "Entry": pos["entry"], "Exit": exit_price,
                        "Qty": qty, "Bars": bars, "Reason": reason,
                        "PnL": pnl,
                        "Trade_Return": pnl / pos["eq0"] if pos["eq0"] > 0 else 0.0,
                        "Risk_Eff": pos["risk_eff"],
                    })
                pos = None

        # --- Valoración a mercado ---
        mtm = equity
        if pos is not None:
            mtm += pos["d"] * pos["qty"] * (c[t] - pos["entry"])
        curve[t - start] = mtm

        # --- Señal al cierre de t (se ejecutará en t+1) ---
        if pos is None and sig[t] != 0 and t < n - 1:
            pending = int(sig[t])

    return curve, trades


def calculate_metrics(bt, trades, capital, random_returns, risk_target):
    equity = bt["Equity"]
    final_equity = float(equity.iloc[-1])
    total_return = final_equity / capital - 1
    buy_hold = float(bt["Buy_Hold"].iloc[-1]) / capital - 1

    dd = bt["Drawdown"]
    max_dd = float(dd.min())

    under = (dd < 0)
    groups = (~under).cumsum()
    dd_duration = int(under.groupby(groups).sum().max()) if under.any() else 0

    r = equity.pct_change().dropna()
    std = r.std(ddof=1)
    sharpe = np.sqrt(ANNUAL_PERIODS) * r.mean() / std if std > 0 else np.nan

    downside = np.sqrt(np.mean(np.minimum(r.to_numpy(), 0) ** 2))
    sortino = np.sqrt(ANNUAL_PERIODS) * r.mean() / downside if downside > 0 else np.nan

    years = len(bt) / ANNUAL_PERIODS
    ann_return = ((final_equity / capital) ** (1 / years) - 1
                  if years > 0 and final_equity > 0 else np.nan)
    calmar = (ann_return / abs(max_dd)
              if max_dd < 0 and np.isfinite(ann_return) else np.nan)

    if trades is not None and not trades.empty:
        pnl = trades["PnL"]
        wins = pnl[pnl > 0].sum()
        losses = -pnl[pnl < 0].sum()
        win_rate = float((pnl > 0).mean())
        profit_factor = float(wins / losses) if losses > 0 else np.nan
        n_trades = int(len(trades))
        avg_trade = float(trades["Trade_Return"].mean())
        expectancy = float(pnl.mean())
        best = float(trades["Trade_Return"].max())
        worst = float(trades["Trade_Return"].min())
        exposure = float(trades["Bars"].sum() / len(bt))
        risk_eff = float(trades["Risk_Eff"].mean())
    else:
        win_rate = profit_factor = avg_trade = expectancy = np.nan
        best = worst = risk_eff = np.nan
        n_trades, exposure = 0, 0.0

    rr = np.asarray(random_returns, dtype=float)
    if n_trades > 0 and len(rr):
        percentile = float((rr < total_return).mean())
        random_median = float(np.median(rr))
    else:
        percentile = random_median = np.nan

    return {
        "Final_Equity": final_equity, "Return": total_return,
        "Buy_Hold_Return": buy_hold, "Max_Drawdown": max_dd,
        "DD_Duration_Bars": dd_duration,
        "Sharpe": sharpe, "Sortino": sortino, "Calmar": calmar,
        "Win_Rate": win_rate, "Profit_Factor": profit_factor,
        "Trades": n_trades, "Avg_Trade_Return": avg_trade,
        "Expectancy_USD": expectancy,
        "Best_Trade_Return": best, "Worst_Trade_Return": worst,
        "Exposure": exposure,
        "Bars": len(bt), "Risk_Effective": risk_eff,
        "Risk_Target": risk_target / 100.0,
        "Random_Percentile": percentile, "Random_Median_Return": random_median,
    }


def get_simulation(bundle, params, threshold):
    capital, risk_pct, max_pos = params["capital"], params["risk"], params["maxpos"]
    key = (bundle["key"], capital, risk_pct, max_pos, round(threshold, 4))

    with _LOCK:
        if S.sim is not None and S.sim_key == key:
            return S.sim

        set_status("Simulando backtest y señales aleatorias…")

        df, A = bundle["df"], bundle["arrays"]
        start = bundle["start"]
        p_long, p_short = bundle["p_wf"]["LONG"], bundle["p_wf"]["SHORT"]
        n = len(df)

        sig = signals_from_probs(p_long, p_short, threshold)
        curve, raw_trades = simulate(
            A, sig, start, capital, risk_pct, max_pos, record=True)

        bt = pd.DataFrame({
            "Timestamp": df.index[start:],
            "Close": A["close"][start:],
            "P_TP_Long": p_long[start:],
            "P_TP_Short": p_short[start:],
            "Signal": sig[start:],
            "Equity": curve,
            "Buy_Hold": capital * A["close"][start:] / A["close"][start - 1],
        })
        bt["Drawdown"] = bt["Equity"] / bt["Equity"].cummax() - 1

        trades = pd.DataFrame(raw_trades)
        if not trades.empty:
            trades["Entry_Time"] = df.index[trades["t0"].to_numpy()]
            trades["Exit_Time"] = df.index[trades["t1"].to_numpy()]
            trades = trades.drop(columns=["t0", "t1"])

        # Contraste: mismas señales, pero barajadas al azar en el tiempo
        random_returns = []
        if len(raw_trades) > 0:
            rng = np.random.default_rng(42)
            window = sig[start:n - 1].copy()
            for _ in range(N_RANDOM_SIMS):
                shuffled = sig.copy()
                shuffled[start:n - 1] = rng.permutation(window)
                rc, _ = simulate(
                    A, shuffled, start, capital, risk_pct, max_pos, record=False)
                random_returns.append(rc[-1] / capital - 1)

        metrics = calculate_metrics(
            bt, trades, capital, random_returns, risk_pct)

        S.sim = {"bt": bt, "trades": trades, "metrics": metrics}
        S.sim_key = key
        return S.sim


# ============================================================
# 5. WIDGETS
# ============================================================

style_w = {"description_width": "initial"}
_full = widgets.Layout(width="100%", max_width="440px")


def _secret(name):
    """Lee un secreto de Colab (icono de la llave) o variable de entorno."""
    try:
        from google.colab import userdata
        return userdata.get(name) or ""
    except Exception:
        return os.environ.get(name, "")


symbol_widget = widgets.Dropdown(
    options=SYMBOLS, value=DEFAULT_SYMBOL, description="Activo",
    style=style_w, layout=_full)
timeframe_widget = widgets.Dropdown(
    options=[(t["label"], k) for k, t in TIMEFRAMES.items()],
    value=INTERVAL,
    description="Temporalidad", style=style_w, layout=_full)

# --- Reloj junto a la temporalidad: hora actual de España + vela seleccionada ---
_DIAS_ES = ["lunes", "martes", "miércoles", "jueves", "viernes", "sábado", "domingo"]
timeframe_clock = widgets.HTML(
    layout=widgets.Layout(margin="0 0 0 6px", align_self="center"))


def _fmt_span(minutes):
    """5 -> '5 min', 60 -> '1 h', 90 -> '1 h 30 min', 1440 -> '24 h'."""
    h, m = divmod(int(minutes), 60)
    if h and m:
        return f"{h} h {m} min"
    return f"{h} h" if h else f"{m} min"


def timeframe_clock_html(nowrap=True):
    """Hora actual en España + duración de la temporalidad = hora resultante,
    con el día de la semana y la fecha a la que corresponde."""
    tz = ZoneInfo("Europe/Madrid")
    now = datetime.now(tz)
    target = now + timedelta(minutes=BAR_MIN)
    diff_days = (target.date() - now.date()).days
    rel = {0: "hoy", 1: "mañana"}.get(diff_days, f"+{diff_days} días")
    dia = _DIAS_ES[target.weekday()]
    return (
        f"<div style='font-size:13px;line-height:1.35;color:{MUTED};"
        f"white-space:{'nowrap' if nowrap else 'normal'}'>🕒 Ahora (España): "
        f"<b style='color:{TEXT}'>{now:%H:%M}</b> {now.tzname()} "
        f"<span style='color:{GOLD}'>+ {_fmt_span(BAR_MIN)}</span> → "
        f"<b style='color:{TEXT}'>{target:%H:%M}</b> · "
        f"<b style='color:{BLUE}'>{dia} {target:%d/%m/%Y}</b> ({rel})</div>")


def update_timeframe_clock():
    try:
        timeframe_clock.value = timeframe_clock_html()
    except Exception:
        pass


async def timeframe_clock_loop():
    """Refresca el reloj cada 20 s para que la hora siempre sea la actual."""
    while True:
        update_timeframe_clock()
        await asyncio.sleep(20)
capital_widget = widgets.BoundedFloatText(
    value=DEFAULT_CAPITAL, min=100, max=1_000_000_000, step=500,
    description="", style=style_w, layout=_full)
risk_widget = widgets.BoundedFloatText(
    value=DEFAULT_RISK_PCT, min=0.1, max=10.0, step=0.1,
    description="Riesgo por operación (%)", style=style_w, layout=_full)
max_position_widget = widgets.BoundedFloatText(
    value=DEFAULT_MAX_POSITION_PCT, min=1, max=1000, step=25,
    description="Exposición máxima (%)", style=style_w, layout=_full)
auto_thr_widget = widgets.Checkbox(
    value=True, description="Umbral automático (equilibrio + margen)",
    indent=False)
threshold_widget = widgets.FloatSlider(
    value=DEFAULT_THRESHOLD, min=0.30, max=0.80, step=0.01,
    description="Umbral manual P(TP)", readout_format=".2f",
    style=style_w, layout=_full, disabled=True)
auto_thr_widget.observe(
    lambda ch: setattr(threshold_widget, "disabled", bool(ch["new"])),
    names="value")

refresh_button = widgets.Button(
    description="Actualizar ahora", button_style="primary", icon="refresh")
daily_button = widgets.Button(
    description="Forzar consulta diaria", button_style="info", icon="calendar",
    tooltip="Vuelve a descargar los cierres diarios y recalcula los modelos")
daily_html = widgets.HTML(layout=widgets.Layout(margin="6px 0 0 0"))
start_button = widgets.Button(
    description="Iniciar actualización automática",
    button_style="success", icon="play")
stop_button = widgets.Button(
    description="Detener actualización", button_style="warning", icon="stop")

_DL = [("backtest", "Backtest CSV"), ("trades", "Operaciones CSV"),
       ("candles", "Velas CSV"), ("metrics", "Métricas CSV"),
       ("html", "VER DASHBOARD COMPLETO")]
download_buttons = {
    kind: widgets.Button(description=label, button_style="", icon="download")
    for kind, label in _DL
}
# El botón del dashboard abre el HTML en el navegador (no lo descarga)
download_buttons["html"].icon = ""   # la mano animada la pone el CSS (.qd-html-btn)
download_buttons["html"].add_class("qd-html-btn")
download_buttons["html"].tooltip = "Abrir el dashboard en una pestaña nueva"

# Texto "VER DASHBOARD COMPLETO" para el panel de Parámetros (sustituye al botón).
# Al pulsarlo intenta activar el botón del panel Exportar, que es el que abre el dashboard.
params_dashboard_text = widgets.HTML(
    "<div style='text-align:right'><span class='qd-nav-link' "
    "onclick=\"try{var b=Array.from(document.querySelectorAll('button')).find("
    "function(x){return x.textContent.indexOf('VER DASHBOARD COMPLETO')>-1});"
    "if(b){b.click();}}catch(e){}\">"
    "Ver la previsión en el navegador</span></div>",
    layout=widgets.Layout(width="100%", margin="0"))

status = widgets.HTML()
diag_button = widgets.Button(
    description="Forzar autodiagnóstico", button_style="warning", icon="heartbeat",
    tooltip="Vuelve a revisar datos, modelos y previsión y consulta al proveedor "
            "si hay velas más nuevas (no reentrena los modelos)",
    layout=widgets.Layout(display="none"))   # sólo se ve con avisos o errores
health_output = widgets.Output(layout=widgets.Layout(margin="0 0 6px 0"))   # panel de autodiagnóstico
alert_html = widgets.HTML()
output = widgets.Output()
forecast_output = widgets.Output(layout=widgets.Layout(margin="0 0 6px 0"))   # Previsión de tendencia (entre Umbral y Acciones)

mobile_widget = widgets.Checkbox(
    value=True, description="Modo móvil (compacto)", indent=False)
telegram_enable_widget = widgets.Checkbox(
    value=False, description="Enviar alertas a Telegram", indent=False)
telegram_token_widget = widgets.Password(
    value=_secret("TELEGRAM_TOKEN"), description="Token de Telegram",
    style=style_w, layout=_full)
telegram_chat_widget = widgets.Text(
    value=_secret("TELEGRAM_CHAT_ID"), description="ID del chat",
    style=style_w, layout=_full)
telegram_test_button = widgets.Button(
    description="Enviar prueba", button_style="info", icon="send")


def read_params():
    return {
        "capital": float(capital_widget.value),
        "risk": float(risk_widget.value),
        "maxpos": float(max_position_widget.value),
        "threshold": float(threshold_widget.value),
        "auto": bool(auto_thr_widget.value),
    }


def set_output(items, target=None):
    """Sustituye la salida de golpe, sin depender del contexto de la celda."""
    fmt = get_ipython().display_formatter.format
    outs = []
    for item in items:
        data, meta = fmt(item)
        outs.append({"output_type": "display_data",
                     "data": data, "metadata": meta})
    (target if target is not None else output).outputs = tuple(outs)


def panel(title, children):
    box = widgets.VBox(
        [widgets.HTML(f"<div class='qd-ptitle'>{title}</div>")] + children)
    box.add_class("qd-panel")
    return box


# ============================================================
# 6. DESCARGAS Y APERTURA DEL DASHBOARD EN EL NAVEGADOR
# ============================================================

def open_html_in_browser(filename):
    """
    Abre el HTML en una pestaña nueva del navegador usando un Blob.
    Los pop-ups deben estar permitidos para colab.research.google.com.
    """
    with open(filename, "r", encoding="utf-8") as handle:
        html_text = handle.read()

    # json.dumps genera un literal JS válido; se escapa "</" para que
    # ningún "</script>" del HTML interrumpa el script que lo contiene.
    payload = json.dumps(html_text).replace("</", "<\\/")

    js = f"""
    (function() {{
      const html = {payload};
      const blob = new Blob([html], {{type: 'text/html'}});
      const url = URL.createObjectURL(blob);
      const w = window.open(url, '_blank');
      if (!w) {{
        alert('El navegador bloqueó la ventana emergente. '
              + 'Permite los pop-ups para este sitio y vuelve a pulsar.');
      }}
    }})();
    """
    # Se muestra dentro de la salida de estado para no ensuciar el notebook
    with status_js:
        status_js.clear_output()
        display(Javascript(js))


# Salida oculta que aloja el JavaScript de apertura
status_js = widgets.Output(layout=widgets.Layout(height="0px", overflow="hidden"))
loading_js = widgets.Output(layout=widgets.Layout(height="0px", overflow="hidden"))

# ---------- Indicador de carga en el centro de la pantalla visible ----------
_LOAD_DEPTH = 0
_LOAD_LOCK = threading.Lock()
_LOAD_MAP = (0.0, 100.0)    # (base, ancho): traduce el avance local al porcentaje mostrado
_LOAD_LAST = -1.0


def _run_loading_js(call):
    try:
        with loading_js:
            loading_js.clear_output()
            display(Javascript(_LOADING_INSTALL_JS + "\n" + call))
    except Exception:
        pass


def show_loading(msg="Cargando datos…"):
    """Muestra (o actualiza) el indicador de carga. Admite llamadas anidadas."""
    global _LOAD_DEPTH, _LOAD_LAST
    with _LOAD_LOCK:
        _LOAD_DEPTH += 1
        if _LOAD_DEPTH == 1:
            _LOAD_LAST = -1.0
    _run_loading_js(f"window.__qdLoad.show({json.dumps(msg)});")


def load_progress(pct, msg=None, creep=None):
    """Porcentaje de carga (0-100, según _LOAD_MAP). Sólo desde el hilo
    principal; en los hilos de trabajo el número avanza solo, despacio."""
    global _LOAD_LAST
    if _LOAD_DEPTH <= 0:
        return
    if threading.current_thread() is not threading.main_thread():
        return
    base, span = _LOAD_MAP
    value = base + float(pct) * span / 100.0
    cap = None if creep is None else base + float(creep) * span / 100.0
    if msg is None and cap is None and abs(value - _LOAD_LAST) < 1.0:
        return
    _LOAD_LAST = value
    _run_loading_js(
        f"window.__qdLoad.progress({value:.1f}, {json.dumps(msg)}, "
        f"{json.dumps(None if cap is None else round(cap, 1))});")


def hide_loading():
    """Oculta el indicador (tras mostrar 100 %) cuando termina la última tarea."""
    global _LOAD_DEPTH
    with _LOAD_LOCK:
        _LOAD_DEPTH = max(0, _LOAD_DEPTH - 1)
        done = _LOAD_DEPTH == 0
    if done:
        _run_loading_js("window.__qdLoad.hide();")


def download_file(kind, user_click=False):
    symbol = S.exported
    if not symbol:
        set_status("Actualiza el dashboard primero.", RED)
        return
    filename = fname(kind, symbol)
    if not os.path.exists(filename):
        set_status(f"No existe {filename}. Actualiza el dashboard primero.", RED)
        return

    # El dashboard HTML se abre directamente en una pestaña nueva
    if kind == "html":
        if not user_click:
            return   # nunca se abre el navegador de forma automática
        try:
            open_html_in_browser(filename)
            set_status("Dashboard abierto en una pestaña nueva.", GREEN)
            return
        except Exception as exc:
            set_status("No se pudo abrir en el navegador "
                       f"({type(exc).__name__}); se descargará.", GOLD)

    # Resto de archivos (CSV) y respaldo: descarga normal
    try:
        from google.colab import files
        files.download(filename)
    except Exception:
        from IPython.display import FileLink
        display(FileLink(filename))


# ============================================================
# 7. SISTEMA DE DISEÑO (CSS COMPARTIDO) Y COMPONENTES HTML
# ============================================================

_CSS_RAW = """
@import url('https://fonts.googleapis.com/css2?family=IBM+Plex+Sans:wght@400;500;600;700&display=swap');

.qd{font-family:@FONT@;color:@TEXT@;background:@BG@;line-height:1.5;
  font-feature-settings:"tnum" 1,"lnum" 1;-webkit-font-smoothing:antialiased}
.qd *{box-sizing:border-box}
.qd h1,.qd h2,.qd h3,.qd p{font-family:inherit;margin:0;padding:0;border:0}
.qd-wrap{padding:8px 4px 6px}

/* Cabecera y señal */
.qd-top{display:grid;grid-template-columns:minmax(0,1.1fr) minmax(0,1fr);gap:14px}
.qd-box{background:@PANEL@;border:1px solid @BORDER@;border-radius:18px;padding:16px 18px}
.qd-brand{display:flex;align-items:center;gap:9px;color:@MUTED@;font-size:12.5px}
.qd-live{width:7px;height:7px;border-radius:50%;background:@GREEN@;
  animation:qdp 2.4s ease-out infinite}
@keyframes qdp{0%{box-shadow:0 0 0 0 rgba(52,211,166,.45)}
  80%,100%{box-shadow:0 0 0 8px rgba(52,211,166,0)}}
@media (prefers-reduced-motion:reduce){.qd-live{animation:none}}
.qd-sym{font-size:15px;font-weight:500;color:@MUTED@;margin-top:18px}
.qd-price{font-size:40px;font-weight:600;letter-spacing:-.025em;line-height:1.15;margin-top:2px}
.qd-meta{display:flex;flex-wrap:wrap;gap:10px 0;margin-top:18px;
  padding-top:14px;border-top:1px solid @BORDER@}
.qd-meta div{padding-right:18px;margin-right:18px;border-right:1px solid @BORDER@;font-size:13px}
.qd-meta div:last-child{border:0;margin:0;padding:0}
.qd-meta small{display:block;color:@MUTED@;font-size:11.5px;margin-bottom:1px}

.qd-sig{border-left:3px solid var(--c)}
.qd-sigk{color:@MUTED@;font-size:12.5px}
.qd-pill{display:flex;align-items:center;gap:10px;font-size:26px;font-weight:600;
  letter-spacing:-.01em;color:var(--c);margin-top:4px}
.qd-pill span{font-size:15px}
.qd-blink{animation:qdblink 1.2s ease-in-out infinite}
@keyframes qdblink{0%,100%{opacity:1}50%{opacity:.15}}
@media (prefers-reduced-motion:reduce){.qd-blink{animation:none}}
.qd-note{color:@MUTED@;font-size:12.5px;margin-top:6px;line-height:1.55}
.qd-g{margin-top:18px}
.qd-gh{display:flex;justify-content:space-between;align-items:baseline;font-size:13px}
.qd-gh span{color:@MUTED@}
.qd-gh b{font-size:17px;font-weight:600}
.qd-bar{position:relative;height:6px;background:@GRID@;border-radius:99px;margin:11px 0 3px}
.qd-bar i{position:absolute;left:0;top:0;bottom:0;border-radius:99px}
.qd-bar u{position:absolute;top:-5px;bottom:-5px;width:2px;border-radius:2px;text-decoration:none}
.qd-leg{display:flex;flex-wrap:wrap;gap:6px 18px;color:@MUTED@;font-size:11.5px;margin-top:16px}
.qd-leg em{display:inline-block;width:10px;height:3px;border-radius:2px;margin-right:6px;vertical-align:middle}

/* Banda de métricas */
.qd-stripw{border:1px solid @BORDER@;border-radius:16px;overflow:hidden;background:@PANEL@}
.qd-strip{display:flex;flex-wrap:wrap;margin:0 -1px -1px 0}
.qd-cell{flex:1 1 150px;min-width:0;padding:14px 16px;
  border-right:1px solid @BORDER@;border-bottom:1px solid @BORDER@}
.qd-k{color:@MUTED@;font-size:12px;display:flex;align-items:center;gap:7px}
.qd-k:before{content:"";width:6px;height:6px;border-radius:50%;background:var(--c);flex:none}
.qd-v{font-weight:600;letter-spacing:-.015em;margin-top:6px;color:@TEXT@;line-height:1.2}
.qd-s{color:@MUTED@;font-size:11.5px;margin-top:5px;line-height:1.45}

/* Secciones */
.qd-sec{margin:20px 0 8px;padding-bottom:6px;border-bottom:1px solid @BORDER@}
.qd-sec h2{font-size:19px;font-weight:600;letter-spacing:-.01em;color:@TEXT@}
.qd-desc{color:@MUTED@;font-size:13px;line-height:1.7;max-width:80ch;margin-top:6px}
.qd-gap{height:14px}

/* Paneles de operación */
.qd-pair{display:grid;grid-template-columns:repeat(auto-fit,minmax(280px,1fr));gap:8px}
.qd-trade{background:@PANEL@;border:1px solid @BORDER@;border-top:2px solid var(--c);
  border-radius:16px;padding:12px 14px 10px}
.qd-trade{cursor:zoom-in;transition:transform .15s ease,box-shadow .15s ease}
.qd-trade:hover{transform:translateY(-2px);box-shadow:0 8px 22px rgba(0,0,0,.35)}
.qd-th{display:flex;justify-content:space-between;align-items:baseline;margin-bottom:6px}
.qd-th b{font-size:17px;font-weight:600;color:var(--c)}
.qd-th span{color:@MUTED@;font-size:12px}
.qd-lv{display:grid;grid-template-columns:repeat(3,1fr);gap:1px;background:@BORDER@;
  border:1px solid @BORDER@;border-radius:8px;overflow:hidden;margin:10px 0 8px}
.qd-lv div{background:@PANEL2@;padding:9px 11px;min-width:0}
.qd-lv small{display:block;color:@MUTED@;font-size:11.5px}
.qd-lv b{font-size:14px;font-weight:600}
.qd-row{display:flex;justify-content:space-between;gap:12px;padding:7px 0;
  border-bottom:1px solid #182334;font-size:13px}
.qd-row:last-of-type{border-bottom:0}
.qd-row span{color:@MUTED@}
.qd-warn{color:@GOLD@;font-size:12.5px;padding:8px 0 2px;line-height:1.5}
.qd-fine{color:@MUTED@;font-size:11.5px;margin-top:8px}

/* Tablas */
.qd-tw{overflow-x:auto;border:1px solid @BORDER@;border-radius:16px;background:@PANEL@}
.qd-table{border-collapse:collapse;width:100%;font-size:12.5px}
.qd-table th{background:@PANEL2@;color:@MUTED@;font-size:12px;font-weight:500;
  padding:10px 12px;text-align:right;white-space:nowrap}
.qd-table td{padding:9px 12px;text-align:right;border-top:1px solid #182334;
  color:@TEXT@;white-space:nowrap}
.qd-table tr:nth-child(even) td{background:rgba(255,255,255,.018)}
.qd-table th:first-child,.qd-table td:first-child{text-align:left}
.qd-b{display:inline-block;padding:1px 9px;border-radius:99px;font-size:11.5px;font-weight:600}

/* Botón destacado del dashboard: pastilla verde con brillo que barre, halo
   pulsante, latido suave y una mano que "pulsa" hacia el texto */
.qd-html-btn{position:relative!important;padding-left:34px!important;border-radius:999px!important;
  border:0!important;cursor:pointer!important;font-weight:800!important;
  color:#FFFFFF!important;text-shadow:0 1px 2px rgba(0,60,35,.55)!important;
  background:
    linear-gradient(105deg,rgba(255,255,255,0) 35%,rgba(255,255,255,.55) 50%,rgba(255,255,255,0) 65%) no-repeat,
    linear-gradient(120deg,#047857,#10B981 45%,#34D399 70%,#10B981)!important;
  background-size:250% 100%,100% 100%!important;background-position:-150% 0,0 0!important;
  box-shadow:0 3px 10px rgba(16,185,129,.45),0 0 0 0 rgba(52,211,153,.65);
  animation:qdshine 2.6s ease-in-out infinite,qdring 1.6s ease-out infinite}
.qd-html-btn:hover{filter:brightness(1.1) saturate(1.1);animation-play-state:paused}
.qd-html-btn:active{filter:brightness(.95)}
.qd-html-btn::before{content:"👉";position:absolute;left:10px;top:50%;margin-top:-9px;font-size:15px;
  line-height:18px;animation:qdpoint .8s ease-in-out infinite}
@keyframes qdshine{0%{background-position:-150% 0,0 0}60%,100%{background-position:250% 0,0 0}}
@keyframes qdring{0%{box-shadow:0 3px 10px rgba(16,185,129,.45),0 0 0 0 rgba(52,211,153,.65)}
  100%{box-shadow:0 3px 10px rgba(16,185,129,.45),0 0 0 10px rgba(52,211,153,0)}}
@keyframes qdpoint{0%,100%{transform:translateX(-3px)}50%{transform:translateX(7px)}}
@media (prefers-reduced-motion:reduce){.qd-html-btn,.qd-html-btn::before{animation:none}}

/* Enlace "VER EN EL NAVEGADOR": texto verde con brillo pulsante, subrayado y flecha que se mueve */
.qd-nav-link{display:inline-block;margin:2px 0 0 0;font-size:14px;font-weight:800;
  letter-spacing:.03em;color:@GREEN@;cursor:pointer;text-decoration:none;
  animation:qdnavglow 1.4s ease-in-out infinite}
.qd-nav-link::after{content:" \u279C";display:inline-block;animation:qdnavarrow .8s ease-in-out infinite}
.qd-nav-link:hover{filter:brightness(1.3);animation-play-state:paused}
@keyframes qdnavglow{0%,100%{text-shadow:0 0 0 rgba(52,211,153,0);filter:brightness(1)}
  50%{text-shadow:0 0 12px rgba(52,211,153,.95);filter:brightness(1.3)}}
@keyframes qdnavarrow{0%,100%{transform:translateX(0)}50%{transform:translateX(7px)}}
@media (prefers-reduced-motion:reduce){.qd-nav-link,.qd-nav-link::after{animation:none}}
.qd-empty{color:@MUTED@;font-size:13px;padding:14px 2px}

/* Horario de mercado (España) */
.qd-clock{display:flex;flex-wrap:wrap;gap:8px 16px;align-items:center;margin:0 0 10px;
  font-size:12.5px;color:@MUTED@}
.qd-clock b{color:@TEXT@;font-weight:600}
.qd-sess{font-size:11.5px;font-weight:600;white-space:nowrap}
.qd-table td.qd-ev{text-align:left;white-space:normal;min-width:190px}
.qd-table td.qd-ev small{display:block;color:@MUTED@;font-size:11.5px;margin-top:2px}
.qd-table td.qd-hr{font-weight:600;font-size:13.5px;letter-spacing:-.01em}
.qd-table tr.qd-next td{background:rgba(242,196,109,.09)}
.qd-table tr.qd-next td:first-child{box-shadow:inset 3px 0 0 @GOLD@}
.qd-table th.l,.qd-table td.l{text-align:left}

/* Pie y gráficos exportados */
.qd-foot{color:@MUTED@;font-size:12px;line-height:1.75;border-top:1px solid @BORDER@;
  margin-top:34px;padding-top:14px;max-width:none;
  text-align:justify;text-justify:inter-word;hyphens:auto}
.qd-foot b{color:@TEXT@;font-weight:600}
.qd-chart{border:1px solid @BORDER@;border-radius:16px;overflow:hidden;margin:8px 0;background:@BG@}

/* Controles de ipywidgets */
.qd-panel{background:@PANEL@;border:1px solid @BORDER@;border-radius:16px;
  padding:8px 12px;margin:0 0 6px 0}
.qd-ptitle{color:@TEXT@;font-weight:600;font-size:13.5px;margin:0 0 4px 0;font-family:@FONT@}
.qd-panel label,.qd-panel .widget-label,.qd-panel .widget-readout{color:@TEXT@!important;font-family:@FONT@}
.qd-panel input,.qd-panel select{background:@PANEL2@!important;color:@TEXT@!important;
  border:1px solid @BORDER@!important;border-radius:8px}
.qd-alert{background:@PANEL@;color:@TEXT@;padding:13px 16px;border:1px solid @BORDER@;
  border-left:3px solid var(--c);border-radius:16px;margin:8px 0;font-family:@FONT@;
  font-size:13px;line-height:1.65}
.qd-alert b{color:var(--c);font-size:14px}
.qd-alert small{color:@MUTED@}

@media (max-width:760px){
  .qd-top{grid-template-columns:1fr}
  .qd-box{padding:16px}
  .qd-price{font-size:32px}
  .qd-cell{flex-basis:135px;padding:12px 13px}
  .qd-sec{margin-top:16px}
}
"""

_CSS_TOKENS = {
    "BG": BG, "PANEL": PANEL, "PANEL2": PANEL2, "TEXT": TEXT, "MUTED": MUTED,
    "GREEN": GREEN, "RED": RED, "BLUE": BLUE, "GOLD": GOLD, "PURPLE": PURPLE,
    "BORDER": BORDER_COLOR, "GRID": GRID_COLOR, "HOVER": HOVER_BG,
    "FONT": FONT,
}


def base_css():
    css = _CSS_RAW
    for key, value in _CSS_TOKENS.items():
        css = css.replace(f"@{key}@", value)
    return css


def wrap(*parts):
    return "<div class='qd qd-wrap'>" + "".join(parts) + "</div>"


def fmt_pct(v, d=2):
    return f"{v:.{d}%}" if v is not None and np.isfinite(v) else "N/D"


def fmt_num(v, d=2):
    return f"{v:,.{d}f}" if v is not None and np.isfinite(v) else "N/D"


def card(title, value, subtitle, color, size=None):
    size = size or UI["size"]
    return (
        f"<div class='qd-cell' style='--c:{color}'>"
        f"<div class='qd-k'>{title}</div>"
        f"<div class='qd-v' style='font-size:{size}px'>{value}</div>"
        f"<div class='qd-s'>{subtitle}</div></div>"
    )


def grid(cards, min_width=155, gap=10):
    min_width = min(min_width, UI["min"])
    return ("<div class='qd-stripw'><div class='qd-strip' "
            f"style='--w:{min_width}px'>" + "".join(cards) + "</div></div>")


JUSTIFY_TITLES = {   # apartados cuyo texto descriptivo va justificado
    "Gestión del riesgo y tamaño de posición", "Análisis técnico", "Análisis de sentimiento", "Autodiagnóstico",
    "Calendario de mercado · Horario de España", "Mercado",
    "Validación cronológica del modelo", "Backtest walk-forward vela a vela",
}


def section(title, desc="", justify=False):
    justify = justify or title in JUSTIFY_TITLES
    st = (" style='text-align:justify;text-justify:inter-word;hyphens:auto;"
          "max-width:none'") if justify else ""
    d = f"<p class='qd-desc'{st}>{desc}</p>" if desc else ""
    return f"<div class='qd-sec'><h2>{title}</h2>{d}</div>"


def badge(text, color):
    return (f"<span class='qd-b' style='color:{color};"
            f"background:{rgba(color, 0.13)};border:1px solid "
            f"{rgba(color, 0.32)}'>{text}</span>")


def row(label, value, color=None):
    style = f" style='color:{color}'" if color else ""
    return f"<div class='qd-row'><span>{label}</span><b{style}>{value}</b></div>"


def _gpos(p):
    return max(0.0, min(1.0, p / GAUGE_MAX)) * 100


def gauge(label, p, color, thr, be):
    value = f"{p:.1%}" if np.isfinite(p) else "N/D"
    width = _gpos(p) if np.isfinite(p) else 0
    return (
        f"<div class='qd-g'><div class='qd-gh'><span>{label}</span>"
        f"<b style='color:{color}'>{value}</b></div>"
        f"<div class='qd-bar'><i style='width:{width:.1f}%;"
        f"background:{color}'></i>"
        f"<u style='left:{_gpos(be):.1f}%;background:{GOLD}'></u>"
        f"<u style='left:{_gpos(thr):.1f}%;background:{TEXT}'></u></div></div>"
    )


BAR_REST_COLOR = "#5B7FA8"   # color del tramo restante de la barra (hasta 100 %)


def make_trade_panel(title, color, entry, stop, target, position, asset,
                     capital, risk_pct, prob=None, edge=None, leader=False):
    reward = position["qty"] * abs(target - entry)
    rr = reward / position["risk"] if position["risk"] > 0 else 0
    eff = position["risk"] / capital * 100 if capital > 0 else 0.0

    if position["capped"]:
        risk_line = (
            "<div class='qd-warn'>Limitado por la exposición máxima: el "
            f"riesgo real es {eff:.2f}% (objetivo {risk_pct:.2f}%).</div>")
    else:
        risk_line = row("Riesgo efectivo", f"{eff:.2f}% del capital")

    if prob is not None and np.isfinite(prob):
        edge_txt = ""
        if edge is not None and np.isfinite(edge):
            ec = GREEN if edge > 0 else RED if edge < 0 else MUTED
            edge_txt = (f" · <b style=\"color:{ec};font-size:12px\">"
                        f"{edge*100:+.1f} pp vs base</b>")
        trend_span = (f"<span><b style=\"color:{color};font-size:12px\">"
                      f"P(TP) {prob:.1%}</b>{edge_txt}</span>")
    else:
        trend_span = ""

    # Texto del título: LONG = COMPRAR, SHORT = VENDER
    _acciones = {"LONG": "COMPRAR", "SHORT": "VENDER"}
    title_txt = (f"{title} = {_acciones[title]}" if title in _acciones
                 else title)

    # El lado con mayor porcentaje P(TP) parpadea
    title_cls = ' class="qd-blink"' if leader else ""

    # Barra fina: parte rellena = P(TP) en el color del lado; el resto hasta
    # el 100 % en otro color (BAR_REST_COLOR).
    if prob is not None and np.isfinite(prob):
        _w = max(0.0, min(100.0, prob * 100))
        prob_bar = (
            f"<div title='P(TP) {prob:.1%} · resto hasta 100 %: {100 - _w:.1f} %' "
            "style='display:flex;height:5px;border-radius:3px;overflow:hidden;"
            f"margin:0 0 10px;background:{BAR_REST_COLOR}'>"
            f"<i style='width:{_w:.2f}%;background:{color}'></i></div>")
    else:
        prob_bar = ""

    return f"""
    <div class="qd-trade" style="--c:{color}" title="Pulsa para ampliar en el centro de la pantalla">
      <div class="qd-th"><b{title_cls}>{title_txt}</b>
        {trend_span}
        <span>Beneficio/riesgo {rr:.2f}</span></div>
      {prob_bar}
      <div class="qd-lv">
        <div><small>Entrada</small><b style=\"color:{TEXT}\">{entry:,.2f}</b></div>
        <div><small>Take profit</small><b style="color:{GREEN}">{target:,.2f}</b></div>
        <div><small>Stop loss</small><b style="color:{RED}">{stop:,.2f}</b></div>
      </div>
      <div style="margin:2px 0 10px">{timeframe_clock_html(nowrap=False)}</div>
      {row("Tamaño estimado", f"{position['qty']:.6f} {asset}")}
      {row("Valor nocional", f"{position['notional']:,.2f} USD")}
      {row("Riesgo al stop", f"{position['risk']:,.2f} USD")}
      {row("Beneficio al objetivo", f"{reward:,.2f} USD")}
      {risk_line}
      <div class="qd-fine">Precios en USD, entrada al cierre.
      No se envían órdenes.</div>
    </div>
    """


def trades_table_html(trades, n=12):
    if trades is None or trades.empty:
        return "<div class='qd-empty'>Sin operaciones en el periodo.</div>"

    rows = []
    for r in trades.tail(n).iloc[::-1].itertuples():
        color = GREEN if r.PnL >= 0 else RED
        side = "LONG" if r.Direction > 0 else "SHORT"
        side_color = GREEN if r.Direction > 0 else RED
        reason_color = {"TP": GREEN, "STOP": RED}.get(r.Reason, MUTED)
        rows.append(
            f"<tr><td>{r.Entry_Time:%Y-%m-%d %H:%M}</td>"
            f"<td>{r.Exit_Time:%Y-%m-%d %H:%M}</td>"
            f"<td>{badge(side, side_color)}</td>"
            f"<td>{badge(r.Reason, reason_color)}</td><td>{int(r.Bars)}</td>"
            f"<td style='color:{color}'>{r.Trade_Return:+.2%}</td>"
            f"<td style='color:{color}'>{'+' if r.PnL >= 0 else '−'}"
            f"${abs(r.PnL):,.2f}</td></tr>"
        )

    return (
        "<div class='qd-tw'><table class='qd-table'>"
        "<tr><th>Entrada</th><th>Salida</th><th>Dirección</th><th>Motivo</th>"
        "<th>Barras</th><th>Retorno</th><th>Resultado</th></tr>"
        + "".join(rows) + "</table></div>"
    )


def candles_table_html(recent):
    rows = []
    for ts, r in recent.iloc[::-1].iterrows():
        up = r["Close"] >= r["Open"]
        color = GREEN if up else RED
        rows.append(
            f"<tr><td>{ts:%Y-%m-%d %H:%M}</td>"
            f"<td>${r['Open']:,.2f}</td><td>${r['High']:,.2f}</td>"
            f"<td>${r['Low']:,.2f}</td>"
            f"<td style='color:{color}'>${r['Close']:,.2f}</td>"
            f"<td>{r['Volume']:,.0f}</td><td>{r['RSI']:.1f}</td>"
            f"<td>${r['ATR']:,.2f}</td></tr>"
        )
    return (
        "<div class='qd-tw'><table class='qd-table'>"
        "<tr><th>Vela (UTC)</th><th>Apertura</th><th>Máximo</th><th>Mínimo</th>"
        "<th>Cierre</th><th>Volumen</th><th>RSI</th><th>ATR</th></tr>"
        + "".join(rows) + "</table></div>"
    )


def validation_cards(v, breakeven):
    out = []
    for side in ("LONG", "SHORT"):
        s = v[side]
        auc_color = GREEN if s["auc"] > 0.53 else GOLD
        skill_color = GREEN if s["skill"] > 0.005 else GOLD
        out += [
            card(f"Tasa de TP · {side}", fmt_pct(s["rate"], 1),
                 f"Equilibrio ≈ {fmt_pct(breakeven, 1)}", BLUE),
            card(f"AUC · {side}", fmt_num(s["auc"], 3), "0,50 = azar", auc_color),
            card(f"Log-loss · {side}", fmt_num(s["logloss"], 4),
                 f"Base {fmt_num(s['ll_base'], 4)} · mejora "
                 f"{s['skill'] * 100:+.2f}%", skill_color),
            card(f"Brier · {side}", fmt_num(s["brier"], 4),
                 f"Base {fmt_num(s['br_base'], 4)}", PURPLE),
        ]
    ab = v.get("ablation")
    if ab:
        for side in ("LONG", "SHORT"):
            d_auc = v[side]["auc"] - ab[side]["auc"]
            col = GREEN if d_auc > 0.005 else (RED if d_auc < -0.005 else GOLD)
            out.append(card(
                f"AUC solo precio · {side}", fmt_num(ab[side]["auc"], 3),
                f"Con datos extra {fmt_num(v[side]['auc'], 3)} ({d_auc:+.3f})",
                col))
    return out


def backtest_cards(m, size=None):
    pct = m["Random_Percentile"]
    rand_color = GREEN if np.isfinite(pct) and pct >= 0.95 else GOLD
    risk_color = (GREEN if np.isfinite(m["Risk_Effective"])
                  and m["Risk_Effective"] >= 0.9 * m["Risk_Target"] else GOLD)
    return [
        card("Rentabilidad", fmt_pct(m["Return"]),
             f"Capital final ${m['Final_Equity']:,.2f}",
             GREEN if m["Return"] >= 0 else RED, size),
        card("Buy & hold", fmt_pct(m["Buy_Hold_Return"]),
             "Mismo tramo", BLUE, size),
        card("Drawdown máx.", fmt_pct(m["Max_Drawdown"]),
             f"Duración máx. {m['DD_Duration_Bars']} barras", RED, size),
        card("Operaciones", str(m["Trades"]),
             f"Exposición {fmt_pct(m['Exposure'], 1)}", PURPLE, size),
        card("Win rate", fmt_pct(m["Win_Rate"], 1), "Por operación", GREEN, size),
        card("Profit factor", fmt_num(m["Profit_Factor"]),
             "Ganancia bruta / pérdida bruta", PURPLE, size),
        card("Expectativa", f"${fmt_num(m['Expectancy_USD'])}",
             "Resultado medio por operación", GOLD, size),
        card("Riesgo efectivo", fmt_pct(m["Risk_Effective"]),
             f"Medio por operación · objetivo {fmt_pct(m['Risk_Target'])}",
             risk_color, size),
        card("Sharpe", fmt_num(m["Sharpe"]), "Todas las barras", BLUE, size),
        card("Sortino", fmt_num(m["Sortino"]), "Riesgo a la baja", BLUE, size),
        card("Calmar", fmt_num(m["Calmar"]), "Rent. anual / drawdown", BLUE, size),
        card("Percentil vs azar", fmt_pct(pct, 0),
             f"{N_RANDOM_SIMS} sims. aleatorias · ≥95% = distinguible",
             rand_color, size),
    ]


# ============================================================
# 8. GRÁFICOS
# ============================================================

def apply_dark_style(fig, height, title=None):
    fig.update_layout(
        template="plotly_dark", height=height,
        paper_bgcolor=BG, plot_bgcolor=PANEL,
        font=dict(family=PLOT_FONT, color=TEXT, size=12),
        margin=dict(l=58, r=26, t=72 if title else 62, b=40),
        legend=dict(orientation="h", yanchor="bottom", y=1.02,
                    xanchor="right", x=1,
                    font=dict(color=MUTED, size=11),
                    bgcolor="rgba(0,0,0,0)"),
        hoverlabel=dict(bgcolor=HOVER_BG, bordercolor=BORDER_COLOR,
                        font=dict(color=TEXT, size=12, family=PLOT_FONT)),
        hovermode="x unified",
        xaxis_rangeslider_visible=False,
    )
    if title is not None:
        fig.update_layout(title=dict(
            text=title, x=0.012, xanchor="left", y=0.95,
            font=dict(color=TEXT, size=16, family=PLOT_FONT)))

    if UI["mobile"]:
        fig.update_layout(
            margin=dict(l=46, r=10, t=56, b=30),
            legend=dict(orientation="h", y=-0.12, x=0, xanchor="left",
                        font=dict(color=MUTED, size=10),
                        bgcolor="rgba(0,0,0,0)"),
            font=dict(family=PLOT_FONT, color=TEXT, size=11),
            hovermode="x",
        )

    axis = dict(
        color=MUTED, tickfont=dict(color=MUTED, size=11),
        title_font=dict(color=MUTED),
        gridcolor=GRID_COLOR, gridwidth=1, zeroline=False,
        linecolor=BORDER_COLOR, showline=True, mirror=False,
    )
    fig.update_xaxes(**axis, showspikes=True, spikemode="across",
                     spikecolor=MUTED, spikethickness=1, spikedash="dot")
    fig.update_yaxes(**axis)

    for annotation in fig.layout.annotations:
        annotation.font.color = MUTED
        annotation.font.size = 12
        if annotation.xref == "paper" and annotation.yref == "paper":
            annotation.x = 0
            annotation.xanchor = "left"
    return fig


def trade_traces(trades, start_ts):
    """Marcadores de entradas y salidas dentro de la ventana mostrada."""
    traces = []
    if trades is None or trades.empty:
        return traces

    recent = trades[trades["Entry_Time"] >= start_ts]
    for d, name, color, symbol in (
            (1, "Entrada LONG", GREEN, "triangle-up"),
            (-1, "Entrada SHORT", RED, "triangle-down")):
        s = recent[recent["Direction"] == d]
        if len(s):
            traces.append(go.Scatter(
                x=s["Entry_Time"], y=s["Entry"], mode="markers", name=name,
                marker=dict(symbol=symbol, size=10, color=color,
                            line=dict(color=BG, width=1)),
                hovertemplate="%{y:,.2f}<extra>" + name + "</extra>"))

    ex = trades[trades["Exit_Time"] >= start_ts]
    if len(ex):
        colors = [GREEN if p >= 0 else RED for p in ex["PnL"]]
        traces.append(go.Scatter(
            x=ex["Exit_Time"], y=ex["Exit"], mode="markers", name="Salida",
            marker=dict(symbol="x", size=8, color=colors,
                        line=dict(color=colors, width=2)),
            hovertemplate="%{y:,.2f}<extra>Salida</extra>"))
    return traces


def build_figures(df, bt, symbol, asset, levels, mobile, trades=None,
                  capital=None):
    plot_df = df.tail(200 if mobile else PLOT_BARS)

    price_fig = make_subplots(
        rows=3, cols=1, shared_xaxes=True, vertical_spacing=0.035,
        row_heights=[0.62, 0.18, 0.20],
        subplot_titles=(f"{symbol} · últimas {len(plot_df)} velas",
                        "Volumen", "RSI (14)"),
    )
    price_fig.add_trace(go.Candlestick(
        x=plot_df.index, open=plot_df["Open"], high=plot_df["High"],
        low=plot_df["Low"], close=plot_df["Close"], name=asset,
        increasing=dict(line=dict(color=GREEN, width=1),
                        fillcolor=rgba(GREEN, 0.75)),
        decreasing=dict(line=dict(color=RED, width=1),
                        fillcolor=rgba(RED, 0.75)),
    ), row=1, col=1)

    for column, color in [("EMA_9", BLUE), ("EMA_21", PURPLE), ("EMA_50", GOLD)]:
        price_fig.add_trace(go.Scatter(
            x=plot_df.index, y=plot_df[column], name=column.replace("_", " "),
            line=dict(color=color, width=1.2)), row=1, col=1)

    if not mobile:
        for tr in trade_traces(trades, plot_df.index[0]):
            price_fig.add_trace(tr, row=1, col=1)

    for text, level, color in levels:
        price_fig.add_hline(
            y=level, line_dash="dot", line_color=color, line_width=1,
            annotation_text=text, annotation_font_color=color,
            annotation_font_size=11, annotation_position="right",
            row=1, col=1)

    volume_colors = [rgba(GREEN, 0.55) if c >= o else rgba(RED, 0.55)
                     for c, o in zip(plot_df["Close"], plot_df["Open"])]
    price_fig.add_trace(go.Bar(
        x=plot_df.index, y=plot_df["Volume"], name="Volumen",
        marker_color=volume_colors, marker_line_width=0,
        showlegend=False), row=2, col=1)

    price_fig.add_hrect(y0=30, y1=70, fillcolor=rgba(BLUE, 0.06),
                        line_width=0, row=3, col=1)
    price_fig.add_trace(go.Scatter(
        x=plot_df.index, y=plot_df["RSI"], name="RSI",
        line=dict(color=BLUE, width=1.4), showlegend=False), row=3, col=1)
    price_fig.add_hline(y=70, line_dash="dot", line_color=RED,
                        line_width=1, row=3, col=1)
    price_fig.add_hline(y=30, line_dash="dot", line_color=GREEN,
                        line_width=1, row=3, col=1)
    price_fig.update_yaxes(range=[0, 100], row=3, col=1)
    apply_dark_style(price_fig, height=560 if mobile else 800)

    equity_fig = go.Figure()
    base = float(capital) if capital else float(bt["Equity"].iloc[0])
    equity_fig.add_trace(go.Scatter(
        x=bt["Timestamp"], y=np.full(len(bt), base), name="Capital inicial",
        line=dict(color=BORDER_COLOR, width=1, dash="dot"),
        hoverinfo="skip", showlegend=False))
    equity_fig.add_trace(go.Scatter(
        x=bt["Timestamp"], y=bt["Equity"], name="Estrategia",
        line=dict(color=GREEN, width=2),
        fill="tonexty", fillcolor=rgba(GREEN, 0.07)))
    equity_fig.add_trace(go.Scatter(
        x=bt["Timestamp"], y=bt["Buy_Hold"], name="Buy & hold",
        line=dict(color=BLUE, width=1.5, dash="dash")))
    equity_fig.update_yaxes(tickprefix="$", tickformat=",.0f")
    apply_dark_style(equity_fig, height=300 if mobile else 380,
                     title="Curva de capital")

    drawdown_fig = go.Figure()
    drawdown_fig.add_trace(go.Scatter(
        x=bt["Timestamp"], y=bt["Drawdown"] * 100, name="Drawdown",
        fill="tozeroy", fillcolor=rgba(RED, 0.20),
        line=dict(color=RED, width=1.2),
        hovertemplate="%{y:.2f}%<extra>Drawdown</extra>"))
    drawdown_fig.update_yaxes(ticksuffix="%")
    apply_dark_style(drawdown_fig, height=220 if mobile else 270,
                     title="Drawdown")

    return price_fig, equity_fig, drawdown_fig


def chart_levels(st):
    sig = st["signal"]
    lv = st["levels"]
    if sig in ("LONG", "SHORT"):
        stop, target = lv[sig]
        return [(f"{sig} SL", stop, RED), (f"{sig} TP", target, GREEN)]
    return [("LONG SL", lv["LONG"][0], MUTED), ("LONG TP", lv["LONG"][1], MUTED),
            ("SHORT SL", lv["SHORT"][0], MUTED), ("SHORT TP", lv["SHORT"][1], MUTED)]


# ---------- Previsión de tendencia (gráfico de líneas al inicio) ----------

def _fmt_horizon(minutes):
    """90 -> '1.5 h', 1440 -> '1 días' (se usa para el horizonte del gráfico)."""
    m = float(minutes)
    if m < 60:
        return f"{m:g} min"
    if m < 1440:
        return f"{m / 60:g} h"
    d = m / 1440
    return f"{d:g} día" if d == 1 else f"{d:g} días"


def forecast_path(df, st):
    """Proyección ilustrativa a MAX_HOLD_BARS velas de la temporalidad activa.

    - Línea central: el precio actual se inclina hacia el lado que favorece el
      modelo. Desplazamiento final = sesgo (P(TP) LONG - P(TP) SHORT, ambos
      frente a su tasa base) x distancia del take profit (ATR x múltiplo x R:R),
      repartido linealmente entre las velas del horizonte.
    - Bandas: volatilidad reciente por vela (desv. típica de los log-retornos)
      escalada con la raíz del tiempo (+-1 sigma y +-2 sigma).
    No es una predicción garantizada: es una guía visual coherente con la señal.
    """
    horizon = int(MAX_HOLD_BARS)
    price = float(st["price"])
    atr = float(st["atr"])

    rets = np.log(df["Close"].astype(float)).diff().dropna().tail(300)
    sigma = float(rets.std()) if len(rets) >= 30 else np.nan
    if not np.isfinite(sigma) or sigma <= 0:
        sigma = atr / price

    bias = float(st["bias"]) if np.isfinite(st["bias"]) else 0.0
    shift = bias * ATR_MULTIPLIER * atr * RISK_REWARD

    k = np.arange(0, horizon + 1)
    centre = price + shift * k / horizon
    spread = price * sigma * np.sqrt(k)
    return {"H": horizon, "k": k, "centre": centre, "spread": spread,
            "shift": shift, "sigma": sigma, "bias": bias}


def _hist_bars(df, mobile):
    """Velas de histórico que muestra el gráfico (escala con la temporalidad)."""
    n = int(min(len(df), max(60, 4 * int(MAX_HOLD_BARS))))
    return min(n, 120) if mobile else n


def forecast_levels(df, n_hist):
    """Techo y suelo máximos de la ventana mostrada + zona de consolidación.

    - Techo = máximo de los High; suelo = mínimo de los Low de las n_hist velas.
    - Consolidación = tramo de W velas (W depende de la temporalidad) cuyo rango
      High-Low está entre el 30 % más estrecho de las últimas ~300 velas y cuyo
      cierre apenas deriva (<= 50 % del rango): el precio oscila en una banda.
      Se toma el tramo más reciente (se unen los tramos consecutivos).
      'active' = termina en las últimas velas; si no, es la consolidación previa.
    """
    win = df.tail(n_hist)
    out = {"ceil": float(win["High"].max()), "floor": float(win["Low"].min()),
           "ceil_ts": win["High"].idxmax(), "floor_ts": win["Low"].idxmin(),
           "n": len(win), "W": 0, "consol": None}

    W = int(np.clip(round(MAX_HOLD_BARS * 0.75), 12, 36))
    out["W"] = W
    ref = df.tail(max(300, n_hist + W))
    if len(ref) < W + 5:
        return out

    rng = ref["High"].rolling(W).max() - ref["Low"].rolling(W).min()
    drift = (ref["Close"] - ref["Close"].shift(W - 1)).abs()
    cutoff = rng.dropna().quantile(0.30)
    ok = ((rng <= cutoff) & (drift <= 0.5 * rng)).to_numpy()

    first_end = (len(ref) - len(win)) + W - 1     # ventana completa dentro de lo mostrado
    cand = np.flatnonzero(ok)
    cand = cand[cand >= first_end]
    if len(cand) == 0:
        return out

    last = int(cand[-1])
    run = last
    while run - 1 >= first_end and ok[run - 1]:
        run -= 1
    a = run - W + 1
    seg = ref.iloc[a:last + 1]
    top, bottom = float(seg["High"].max()), float(seg["Low"].min())
    bars_ago = len(ref) - 1 - last
    out["consol"] = {
        "start": ref.index[a], "end": ref.index[last],
        "top": top, "bottom": bottom, "bars": last - a + 1,
        "bars_ago": bars_ago, "active": bars_ago <= max(2, W // 4),
        "width": (top - bottom) / ((top + bottom) / 2),
    }
    return out


def coincident_levels(df, n_hist, atr, price, max_levels=3, k=3):
    """Techos y suelos donde COINCIDEN varias velas de la temporalidad elegida.

    - Se toman los máximos (High) y mínimos (Low) locales de las n_hist velas
      mostradas: una vela es máximo/mínimo local si supera/no supera a las k
      velas de cada lado.
    - Se agrupan los que están a menos de ±tol entre sí (tol = 0,25·ATR actual,
      con un mínimo del 0,05 % del precio; el ATR ya depende de la temporalidad).
    - Cuentan como nivel los grupos con >= 2 toques; se muestran los de más
      toques (a igualdad, el más cercano al precio actual).
    """
    win = df.tail(n_hist)
    tol = max(0.25 * float(atr), 0.0005 * float(price))
    out = {"tol": tol, "res": [], "sup": []}

    def levels(series, is_high):
        roll = series.rolling(2 * k + 1, center=True)
        swing = (series == (roll.max() if is_high else roll.min())).to_numpy()
        pts = series[swing].dropna()
        if len(pts) < 2:
            return []
        vals, times = pts.to_numpy(float), pts.index
        order = np.argsort(vals)
        groups, cur = [], [order[0]]
        for j in order[1:]:
            if vals[j] - vals[cur[0]] <= 2 * tol:
                cur.append(j)
            else:
                groups.append(cur)
                cur = [j]
        groups.append(cur)
        found = [{"price": float(vals[g].mean()), "count": len(g),
                  "first": times[g].min(), "last": times[g].max()}
                 for g in groups if len(g) >= 2]
        found.sort(key=lambda d: (-d["count"], abs(d["price"] - price)))
        return sorted(found[:max_levels], key=lambda d: d["price"])

    out["res"] = levels(win["High"], True)
    out["sup"] = levels(win["Low"], False)
    return out


def build_forecast_figure(df, st, symbol, asset, mobile):
    fp = forecast_path(df, st)
    horizon, centre, spread = fp["H"], fp["centre"], fp["spread"]
    price = float(st["price"])

    n_hist = _hist_bars(df, mobile)
    hist = df["Close"].tail(n_hist)
    lv = forecast_levels(df, n_hist)
    cz = lv["consol"]
    co = coincident_levels(df, n_hist, float(st["atr"]), price,
                           2 if mobile else 3)

    t0 = df.index[-1]
    fut = pd.date_range(t0, periods=horizon + 1, freq=BAR_FREQ)
    color = TREND_COLORS.get(st["bias_label"], GOLD)

    fig = make_subplots(rows=2, cols=1, shared_xaxes=True,
                        vertical_spacing=0.03, row_heights=[0.78, 0.22])

    def band(mult, alpha, name):
        fig.add_trace(go.Scatter(
            x=fut, y=np.maximum(centre + mult * spread, 0), mode="lines",
            line=dict(width=0), hoverinfo="skip", showlegend=False))
        fig.add_trace(go.Scatter(
            x=fut, y=np.maximum(centre - mult * spread, 0), mode="lines",
            line=dict(width=0), fill="tonexty", fillcolor=rgba(color, alpha),
            name=name, hoverinfo="skip"))

    band(2, 0.07, "Rango ±2σ (≈95 %)")
    fig.data[-1].showlegend = False
    band(1, 0.14, "Rango ±1σ (≈68 %)")

    # Capa opcional de velas japonesas: oculta al inicio ("legendonly"); se activa
    # pulsando "Velas japonesas" en la leyenda. Va antes de la línea de cierre
    # para que la línea quede dibujada por encima.
    cwin = df.tail(n_hist)
    fig.add_trace(go.Candlestick(
        x=cwin.index, open=cwin["Open"], high=cwin["High"],
        low=cwin["Low"], close=cwin["Close"], name="Velas japonesas",
        visible="legendonly",
        increasing=dict(line=dict(color=GREEN, width=1),
                        fillcolor=rgba(GREEN, 0.75)),
        decreasing=dict(line=dict(color=RED, width=1),
                        fillcolor=rgba(RED, 0.75))), row=1, col=1)

    fig.add_trace(go.Scatter(
        x=hist.index, y=hist.values, mode="lines", name=f"{asset} · cierre",
        line=dict(color=BLUE, width=2),
        hovertemplate="$%{y:,.2f}<extra>Cierre</extra>"))
    fig.add_trace(go.Scatter(
        x=fut, y=centre, mode="lines", name="Proyección central",
        line=dict(color=color, width=2.6, dash="dash"),
        hovertemplate="$%{y:,.2f}<extra>Proyección</extra>"))

    # Histograma de volumen (parte inferior): verde = vela alcista, rojo = bajista
    vwin = df.tail(n_hist)
    vcolors = [rgba(GREEN, 0.60) if cl >= op else rgba(RED, 0.60)
               for cl, op in zip(vwin["Close"], vwin["Open"])]
    fig.add_trace(go.Bar(
        x=vwin.index, y=vwin["Volume"], name="Volumen", showlegend=False,
        marker_color=vcolors, marker_line_width=0,
        hovertemplate="%{y:,.0f}<extra>Volumen</extra>"), row=2, col=1)

    # Techo y suelo máximos de la temporalidad: líneas finas horizontales
    x_line = [hist.index[0], fut[-1]]
    fig.add_trace(go.Scatter(
        x=x_line, y=[lv["ceil"]] * 2, mode="lines", name="Techo máx.",
        line=dict(color=RED, width=1, dash="dot"), hoverinfo="skip"))
    fig.add_trace(go.Scatter(
        x=x_line, y=[lv["floor"]] * 2, mode="lines", name="Suelo mín.",
        line=dict(color=GREEN, width=1, dash="dot"), hoverinfo="skip"))
    for _kind, _col, _name in (("res", RED, "Techos coincidentes"),
                               ("sup", GREEN, "Suelos coincidentes")):
        for _i, _lvl in enumerate(co[_kind]):           # líneas finas a todo el gráfico
            fig.add_trace(go.Scatter(
                x=x_line, y=[_lvl["price"]] * 2, mode="lines", name=_name,
                legendgroup=_kind, showlegend=(_i == 0),
                line=dict(color=_col, width=1, dash="longdash"),
                hoverinfo="skip"))
    if cz:
        for _y in (cz["top"], cz["bottom"]):            # límites proyectados a todo el gráfico
            fig.add_trace(go.Scatter(
                x=x_line, y=[_y] * 2, mode="lines", showlegend=False,
                line=dict(color=PURPLE, width=1, dash="dash"), hoverinfo="skip"))
        fig.add_trace(go.Scatter(                       # entrada de leyenda
            x=[None], y=[None], mode="markers",
            name="Consolidación" if cz["active"] else "Consolidación previa",
            marker=dict(symbol="square", size=11, color=rgba(PURPLE, 0.45),
                        line=dict(color=PURPLE, width=1))))

    end = float(centre[-1])
    fig.add_trace(go.Scatter(
        x=[fut[-1]], y=[end], mode="markers", showlegend=False,
        marker=dict(color=color, size=9, line=dict(color=BG, width=2)),
        hoverinfo="skip"))

    span = _fmt_horizon(horizon * BAR_MIN)
    if mobile:
        title = f"Previsión · {TF['label']} · {horizon} velas"
    else:
        title = (f"Previsión de tendencia · {symbol} · {TF['label']} · "
                 f"próximas {horizon} velas (≈ {span})")
    apply_dark_style(fig, height=400 if mobile else 540, title=title)
    fig.update_yaxes(tickprefix="$", tickformat=",.0f", row=1, col=1)
    fig.update_yaxes(tickformat=".2s", title_text="Volumen", row=2, col=1)
    # La leyenda ahora tiene más entradas: más margen para que no pise el título
    fig.update_layout(margin=dict(b=70) if mobile else dict(t=104))

    # Etiquetas de techo/suelo y zona de consolidación
    x_left = hist.index[0].to_pydatetime()
    fig.add_annotation(x=x_left, y=lv["ceil"], text=f"Techo máx. ${lv['ceil']:,.0f}",
                       showarrow=False, xanchor="left", yanchor="bottom",
                       font=dict(color=RED, size=11))
    fig.add_annotation(x=x_left, y=lv["floor"], text=f"Suelo mín. ${lv['floor']:,.0f}",
                       showarrow=False, xanchor="left", yanchor="top",
                       font=dict(color=GREEN, size=11))
    _span = fut[-1].to_pydatetime() - x_left
    for _kind, _col, _txt, _anchor in (("res", RED, "Techo", "bottom"),
                                       ("sup", GREEN, "Suelo", "top")):
        for _i, _lvl in enumerate(co[_kind]):
            fig.add_annotation(
                x=x_left + _span * (0.46 + 0.11 * _i), y=_lvl["price"],
                text=f"{_txt} ×{_lvl['count']} ${_lvl['price']:,.0f}",
                showarrow=False, xanchor="left", yanchor=_anchor,
                font=dict(color=_col, size=10))
    if cz:
        x0, x1 = cz["start"].to_pydatetime(), cz["end"].to_pydatetime()
        fig.add_shape(type="rect", x0=x0, x1=x1, y0=cz["bottom"], y1=cz["top"],
                      fillcolor=rgba(PURPLE, 0.13), layer="below",
                      line=dict(color=PURPLE, width=1, dash="dot"))
        fig.add_annotation(x=x0 + (x1 - x0) / 2, y=cz["top"],
                           text="Consolidación" if cz["active"] else "Consolidación previa",
                           showarrow=False, xanchor="center", yanchor="top",
                           font=dict(color=PURPLE, size=11))
        x_lab = x_left + (fut[-1].to_pydatetime() - x_left) * 0.22
        fig.add_annotation(x=x_lab, y=cz["top"], text=f"Consol. máx. ${cz['top']:,.0f}",
                           showarrow=False, xanchor="left", yanchor="bottom",
                           font=dict(color=PURPLE, size=10))
        fig.add_annotation(x=x_lab, y=cz["bottom"], text=f"Consol. mín. ${cz['bottom']:,.0f}",
                           showarrow=False, xanchor="left", yanchor="top",
                           font=dict(color=PURPLE, size=10))

    # Línea "Ahora" y etiqueta del objetivo central (después del estilo, que
    # recolorea las anotaciones existentes)
    now_x = t0.to_pydatetime()
    fig.add_shape(type="line", x0=now_x, x1=now_x, y0=0, y1=1, yref="paper",
                  line=dict(color=MUTED, width=1, dash="dot"))
    fig.add_annotation(x=now_x, y=1, yref="paper", text="Ahora",
                       showarrow=False, yanchor="bottom", xanchor="center",
                       font=dict(color=MUTED, size=11))
    pct = end / price - 1
    fig.add_annotation(
        x=fut[-1], y=end, text=f"${end:,.0f} ({pct:+.2%})", showarrow=False,
        xanchor="right", yanchor="bottom" if fp["shift"] >= 0 else "top",
        yshift=12 if fp["shift"] >= 0 else -12,
        font=dict(color=color, size=12))
    return fig


def forecast_section_html(df, st, mobile=False):
    fp = forecast_path(df, st)
    horizon, price = fp["H"], float(st["price"])
    end = float(fp["centre"][-1])
    pct = end / price - 1
    lo, hi = max(end - fp["spread"][-1], 0.0), end + fp["spread"][-1]
    label = st["bias_label"]
    col = TREND_COLORS.get(label, GOLD)

    lv = forecast_levels(df, _hist_bars(df, mobile))
    cz = lv["consol"]
    if cz is None:
        cons_txt = "sin consolidación clara en esa ventana"
    elif cz["active"]:
        cons_txt = (f"consolidación <b style='color:{PURPLE}'>activa</b> entre "
                    f"${cz['bottom']:,.0f} y ${cz['top']:,.0f} ({cz['bars']} velas, "
                    f"ancho {cz['width']:.2%})")
    else:
        cons_txt = (f"última consolidación entre ${cz['bottom']:,.0f} y "
                    f"${cz['top']:,.0f}, terminó hace {cz['bars_ago']} velas")
    co = coincident_levels(df, _hist_bars(df, mobile), float(st["atr"]), price,
                           2 if mobile else 3)

    def _lv_txt(items, name):
        if not items:
            return f"sin {name} coincidentes"
        return (f"{name} coincidentes " + ", ".join(
            f"${d['price']:,.0f} (×{d['count']})" for d in items))
    coin_txt = (f"{_lv_txt(co['res'], 'techos')} · {_lv_txt(co['sup'], 'suelos')} "
                f"(máximos/mínimos locales a ±${co['tol']:,.0f}, al menos 2 toques). ")
    coin_txt = coin_txt[:1].upper() + coin_txt[1:]
    levels_txt = (
        f"Techo máximo <b style='color:{RED}'>${lv['ceil']:,.0f}</b> y suelo "
        f"mínimo <b style='color:{GREEN}'>${lv['floor']:,.0f}</b> de las últimas "
        f"{lv['n']} velas; {cons_txt} (tramo de {lv['W']} velas con el rango entre "
        "el 30 % más estrecho y sin deriva). " + coin_txt)
    desc = (
        f"Proyección a {horizon} velas de {TF['label'].lower()} "
        f"(≈ {_fmt_horizon(horizon * BAR_MIN)}), el mismo horizonte máximo de "
        f"las operaciones. Sesgo del modelo: <b style='color:{col}'>{label}</b> "
        f"({fp['bias'] * 100:+.1f} pp) · objetivo central "
        f"<b style='color:{TEXT}'>${end:,.0f}</b> ({pct:+.2%}) · rango ±1σ "
        f"${lo:,.0f} – ${hi:,.0f}. " + levels_txt +
        "Es una proyección ilustrativa, no una "
        "predicción: la línea central inclina el precio actual según el sesgo "
        "del modelo × la distancia del take profit, y las bandas ensanchan la "
        "volatilidad reciente con la raíz del tiempo. Cambia con la "
        "temporalidad elegida.")
    return section("Previsión de tendencia", desc, justify=True)


# ---------- Autodiagnóstico: ¿datos, modelo y previsión están al día? ----------

HEALTH_ICONS = {"ok": ("✔", GREEN), "warn": ("⚠", GOLD),
                "error": ("✖", RED), "info": ("•", MUTED)}


# ============================================================
# 8b. PANEL DE ANÁLISIS TÉCNICO (lectura clásica de cada indicador)
# ============================================================

def ta_snapshot(df, st):
    """Lista de (grupo, nombre, valor, puntuación -1/0/+1, lectura) en la última vela."""
    r = df.iloc[-1]
    price = float(r["Close"])
    out = []

    def g(name):
        v = r.get(name, np.nan)
        return float(v) if v is not None and np.isfinite(v) else np.nan

    def add(grp, name, val, score, note):
        out.append((grp, name, val, int(score), note))

    # Tendencia
    t = st["trend"]
    add("Tendencia", "EMA 9/21/50", t.capitalize(),
        1 if t == "ALCISTA" else -1 if t == "BAJISTA" else 0, "Orden de las medias")
    d = g("ST_Dir")
    if np.isfinite(d):
        add("Tendencia", "Supertrend (14, 3)", f"{g('Supertrend'):,.0f}",
            d, "Precio por encima: alcista" if d > 0 else "Precio por debajo: bajista")
    ip = g("Ichi_Pos")
    if np.isfinite(ip):
        add("Tendencia", "Ichimoku", "Sobre nube" if ip > 0 else "Bajo nube" if ip < 0 else "Dentro",
            ip, "Nube 9/26/52 (sin adelantar el futuro)")
    adx, di = g("ADX"), g("DI_Diff")
    if np.isfinite(adx) and np.isfinite(di):
        sc = (1 if di > 0 else -1) if adx >= 20 else 0
        add("Tendencia", "ADX / DI±", f"{adx:.0f}", sc,
            "Tendencia con fuerza" if adx >= 20 else "Sin tendencia (rango)")
    ar = g("Aroon_Osc")
    if np.isfinite(ar):
        add("Tendencia", "Aroon (25)", f"{ar:+.0f}", 1 if ar > 50 else -1 if ar < -50 else 0,
            "Máximos recientes" if ar > 50 else "Mínimos recientes" if ar < -50 else "Sin sesgo")

    # Momento
    rsi = g("RSI")
    add("Momento", "RSI (14)", f"{rsi:.0f}", 1 if rsi < 30 else -1 if rsi > 70 else 0,
        "Sobreventa" if rsi < 30 else "Sobrecompra" if rsi > 70 else "Zona neutra")
    mh = g("MACD_Hist")
    if np.isfinite(mh):
        add("Momento", "MACD (12,26,9)", f"{mh*100:+.3f}%", 1 if mh > 0 else -1,
            "Histograma positivo" if mh > 0 else "Histograma negativo")
    k, dd = g("Stoch_K"), g("Stoch_D")
    if np.isfinite(k) and np.isfinite(dd):
        sc = 1 if (k < 20 and k > dd) else -1 if (k > 80 and k < dd) else 0
        add("Momento", "Estocástico", f"{k:.0f}/{dd:.0f}", sc,
            "Sobreventa girando" if sc > 0 else "Sobrecompra girando" if sc < 0 else "Sin extremo")
    sr = g("StochRSI")
    if np.isfinite(sr):
        add("Momento", "StochRSI", f"{sr:.2f}", 1 if sr < 0.2 else -1 if sr > 0.8 else 0,
            "Sobreventa" if sr < 0.2 else "Sobrecompra" if sr > 0.8 else "Neutro")
    cci = g("CCI")
    if np.isfinite(cci):
        add("Momento", "CCI (20)", f"{cci:+.0f}", 1 if cci < -100 else -1 if cci > 100 else 0,
            "Sobreventa" if cci < -100 else "Sobrecompra" if cci > 100 else "Neutro")
    div = g("RSI_Div")
    add("Momento", "Divergencia RSI", "Alcista" if div > 0 else "Bajista" if div < 0 else "No",
        div if np.isfinite(div) else 0, "Precio y RSI discrepan" if div else "Sin divergencia")

    # Volumen
    mfi = g("MFI")
    if np.isfinite(mfi):
        add("Volumen", "MFI (14)", f"{mfi:.0f}", 1 if mfi < 20 else -1 if mfi > 80 else 0,
            "Sobreventa" if mfi < 20 else "Sobrecompra" if mfi > 80 else "Neutro")
    cmf = g("CMF")
    if np.isfinite(cmf):
        add("Volumen", "CMF (20)", f"{cmf:+.2f}", 1 if cmf > 0.05 else -1 if cmf < -0.05 else 0,
            "Acumulación" if cmf > 0.05 else "Distribución" if cmf < -0.05 else "Equilibrado")
    ob = g("OBV_Slope")
    if np.isfinite(ob):
        add("Volumen", "OBV (24 velas)", f"{ob:+.2f}", 1 if ob > 0.05 else -1 if ob < -0.05 else 0,
            "Volumen acompaña la subida" if ob > 0.05 else "Volumen acompaña la caída" if ob < -0.05 else "Plano")
    vd = g("VWAP_Dist")
    if np.isfinite(vd):
        add("Volumen", "VWAP 24 velas", f"{vd*100:+.2f}%", 1 if vd > 0 else -1,
            "Precio sobre el VWAP" if vd > 0 else "Precio bajo el VWAP")

    # Volatilidad / estructura
    bb = g("BB_Pct")
    add("Estructura", "Bollinger %B", f"{bb:.2f}", 1 if bb < 0 else -1 if bb > 1 else 0,
        "Bajo la banda inferior" if bb < 0 else "Sobre la banda superior" if bb > 1 else "Dentro de bandas")
    dp = g("Donchian_Pos")
    if np.isfinite(dp):
        add("Estructura", "Donchian (20)", f"{dp:.2f}", 1 if dp >= 0.95 else -1 if dp <= 0.05 else 0,
            "Ruptura de máximos" if dp >= 0.95 else "Ruptura de mínimos" if dp <= 0.05 else "Dentro del canal")
    sq = g("Squeeze")
    if np.isfinite(sq):
        add("Estructura", "Compresión BB/KC", f"{sq:.2f}", 0,
            "Compresión: ruptura posible" if sq < 1 else "Volatilidad normal")

    # Patrón de la última vela
    br, uw, lw, en = g("Body_Ratio"), g("Upper_Wick"), g("Lower_Wick"), g("Engulf")
    if en > 0:
        add("Velas", "Última vela", "Envolvente", 1, "Envolvente alcista")
    elif en < 0:
        add("Velas", "Última vela", "Envolvente", -1, "Envolvente bajista")
    elif np.isfinite(br) and abs(br) < 0.1:
        add("Velas", "Última vela", "Doji", 0, "Indecisión")
    elif np.isfinite(lw) and lw > 0.6 and abs(br) < 0.3:
        add("Velas", "Última vela", "Martillo", 1, "Mecha inferior larga")
    elif np.isfinite(uw) and uw > 0.6 and abs(br) < 0.3:
        add("Velas", "Última vela", "Estrella fugaz", -1, "Mecha superior larga")
    else:
        add("Velas", "Última vela", "Sin patrón", 0, "Sin figura clásica")
    return out


def ta_section_html(df, st):
    try:
        items = ta_snapshot(df, st)
    except Exception as exc:
        return section("Análisis técnico", f"No se pudo calcular: {exc}")
    n_up = sum(1 for it in items if it[3] > 0)
    n_dn = sum(1 for it in items if it[3] < 0)
    n_ne = len(items) - n_up - n_dn
    net = (n_up - n_dn) / max(1, n_up + n_dn)
    label = "ALCISTA" if net > 0.25 else "BAJISTA" if net < -0.25 else "NEUTRO"
    lcol = GREEN if label == "ALCISTA" else RED if label == "BAJISTA" else GOLD

    cards = [card("Consenso AT", label.capitalize(),
                  f"{n_up} alcistas · {n_dn} bajistas · {n_ne} neutros", lcol)]
    for grp, name, val, sc, note in items:
        col = GREEN if sc > 0 else RED if sc < 0 else MUTED
        cards.append(card(f"{grp} · {name}", val, note, col))
    return (section("Análisis técnico",
                    "Lectura clásica de cada indicador en la última vela cerrada "
                    f"de {TF['label'].lower()}. Los osciladores se leen por extremos "
                    "(sobreventa = alcista, sobrecompra = bajista) y los de tendencia por "
                    "su dirección. Es una lectura descriptiva: la señal LONG/SHORT sigue "
                    "decidiéndola el modelo, que usa estas mismas variables como entrada.")
            + grid(cards, 170))


def health_check(c):
    """Revisa que los datos y el estudio de tendencia estén actualizados y sean
    coherentes. No consulta la red (usa lo ya descargado) y nunca lanza
    excepciones: si una comprobación falla, se informa como aviso.

    Devuelve una lista de (nivel, comprobación, detalle) con nivel en
    'ok' | 'warn' | 'error' | 'info'.
    """
    out = []

    def add(level, name, detail):
        out.append((level, name, detail))

    def guard(fn):
        try:
            fn()
        except Exception as exc:          # una comprobación rota no tumba el panel
            add("warn", fn.__name__.replace("_chk_", "").replace("_", " ").capitalize(),
                f"no se pudo comprobar ({type(exc).__name__}: {exc})")

    df, st, symbol = c["df"], c["st"], c["symbol"]
    now = pd.Timestamp.now(tz="UTC")

    # 1) Frescura de la última vela cerrada
    def _chk_ultima_vela():
        last = df.index[-1]
        last = last.tz_localize("UTC") if last.tzinfo is None else last.tz_convert("UTC")
        expected = now.floor(BAR_FREQ) - pd.Timedelta(minutes=BAR_MIN)
        lag = int(round((expected - last).total_seconds() / 60.0 / BAR_MIN))
        txt = f"última vela {last:%Y-%m-%d %H:%M} UTC"
        if lag <= 0:
            add("ok", "Última vela", f"{txt} · es la última cerrada")
        elif lag == 1:
            add("ok", "Última vela",
                f"{txt} · 1 vela de retraso (normal: el proveedor tarda unos minutos)")
        elif lag <= 3:
            add("warn", "Última vela",
                f"{txt} · {lag} velas de retraso ({_fmt_span(lag * BAR_MIN)}); "
                "el proveedor va con retraso, vuelve a actualizar")
        else:
            add("error", "Última vela",
                f"{txt} · {lag} velas de retraso ({_fmt_span(lag * BAR_MIN)}); "
                "los datos están desactualizados")
    guard(_chk_ultima_vela)

    # 1b) Sólo al forzar el autodiagnóstico: ¿el proveedor tiene velas más nuevas?
    def _chk_proveedor():
        p = c.get("probe")
        if p is None:
            return
        if p.get("error"):
            add("warn", "Proveedor", f"no se pudo consultar ({p['error']})")
            return
        mine = df.index[-1]
        mine = mine.tz_localize("UTC") if mine.tzinfo is None else mine.tz_convert("UTC")
        newer = int(round((p["last"] - mine).total_seconds() / 60.0 / BAR_MIN))
        if newer <= 0:
            add("ok", "Proveedor",
                "sin velas más nuevas que las del dashboard (consultado ahora)")
        else:
            add("warn", "Proveedor",
                f"ya tiene {newer} vela(s) más nueva(s) (hasta {p['last']:%Y-%m-%d %H:%M} "
                "UTC); pulsa «Actualizar ahora»")
    guard(_chk_proveedor)

    # 2) Datos diarios (cierres del activo, macro y Fear & Greed)
    def _chk_datos_diarios():
        info = daily_trend_info(symbol)
        if info is None:
            add("info", "Datos diarios", "aún sin copia de los cierres diarios")
            return
        today = pd.Timestamp.now(tz="UTC").tz_localize(None).normalize()
        lag_d = (today - pd.Timedelta(days=1) - pd.Timestamp(info["last"]).normalize()).days
        age = _fmt_span(max(0, (_time.time() - info["fetched"]) / 60.0))
        txt = f"último cierre diario {info['last']:%d/%m/%Y} · consultado hace {age}"
        if info["stale"]:
            add("warn", "Datos diarios",
                f"{txt} · pendiente de consultar tras el cierre diario "
                "(pulsa «Forzar consulta diaria»)")
        elif lag_d > 1:
            add("warn", "Datos diarios",
                f"{txt} · falta el cierre de hace {lag_d} días")
        else:
            add("ok", "Datos diarios", f"{txt} · al día")
    guard(_chk_datos_diarios)

    # 3) Modelos: entrenados con el cierre diario vigente
    def _chk_modelos():
        key = c.get("bundle_key")
        built = c.get("built_at")
        age = (f" · entrenados hace {_fmt_span(max(0, (_time.time() - built) / 60.0))}"
               if built else "")
        if key is None:
            add("info", "Modelos", "sin información de entrenamiento" + age)
            return
        if key[3] == _daily_stamp():
            add("ok", "Modelos", f"calculados con el cierre diario de {key[3]}{age}")
        else:
            add("warn", "Modelos",
                f"calculados con el cierre diario de {key[3]}, ya hay uno más "
                f"reciente ({_daily_stamp()}); se recalculan al actualizar{age}")
    guard(_chk_modelos)

    # 4) Calidad de los datos de mercado
    def _chk_calidad():
        problems = []
        if not df.index.is_monotonic_increasing or df.index.has_duplicates:
            problems.append("índice de fechas desordenado o duplicado")
        t = df.tail(200)
        ohlc = t[["Open", "High", "Low", "Close"]].to_numpy(float)
        if not np.isfinite(ohlc).all():
            problems.append("valores vacíos o infinitos en las últimas velas")
        elif (ohlc <= 0).any():
            problems.append("precios a cero o negativos")
        else:
            eps = 1e-9
            bad = ((t["High"] < t["Low"] * (1 - eps))
                   | (t["Close"] > t["High"] * (1 + eps))
                   | (t["Close"] < t["Low"] * (1 - eps))
                   | (t["Open"] > t["High"] * (1 + eps))
                   | (t["Open"] < t["Low"] * (1 - eps))).sum()
            if bad:
                problems.append(f"{int(bad)} velas con OHLC incoherente")
            jump = float(t["Close"].pct_change().abs().max())
            if np.isfinite(jump) and jump > 0.20:
                problems.append(f"salto de {jump:.0%} entre dos cierres consecutivos")
        if BAR_MIN < 1440:
            zero = float((df["Volume"].tail(50) <= 0).mean())
            if zero > 0.30:
                problems.append(f"{zero:.0%} de velas sin volumen en las últimas 50")
        atr = float(st["atr"])
        if not np.isfinite(atr) or atr <= 0:
            problems.append("ATR no válido")
        gaps = int(c.get("gaps") or 0)
        gap_txt = f"{gaps} huecos corregidos de {len(df):,} velas"
        if gaps / max(len(df), 1) > 0.02:
            problems.append(f"demasiados huecos ({gap_txt})")
        if problems:
            add("error" if any(("vacíos" in p or "cero" in p or "ATR" in p
                                or "OHLC" in p or "índice" in p) for p in problems)
                else "warn", "Calidad de datos", "; ".join(problems))
        else:
            add("ok", "Calidad de datos",
                f"OHLC coherente, sin valores vacíos · {gap_txt}")
    guard(_chk_calidad)

    # 5) Fuentes externas y variables extra de la última vela
    def _chk_fuentes():
        src = c.get("sources") or {}
        bad = [k for k, (ok, _) in src.items() if not ok]
        cols = [f for f in EXTRA_FEATURES if f in df.columns]
        nan_last = int(df[cols].iloc[-1].isna().sum()) if cols else 0
        n_ok = sum(1 for ok, _ in src.values() if ok)
        txt = f"{n_ok}/{len(src)} fuentes"
        if cols:
            txt += f" · {nan_last}/{len(cols)} variables extra sin dato en la última vela"
        if bad:
            add("warn", "Fuentes externas",
                f"{txt} · caídas: {', '.join(bad)} (el modelo sigue con las demás)")
        elif cols and nan_last > len(cols) / 2:
            add("warn", "Fuentes externas",
                f"{txt} · más de la mitad sin dato: revisa las fuentes")
        else:
            add("ok", "Fuentes externas", txt)
    guard(_chk_fuentes)

    # 6) Señal del modelo y niveles de operación
    def _chk_senal():
        problems = []
        for k in ("p_long", "p_short", "threshold"):
            v = st[k]
            if not np.isfinite(v) or not (0.0 < v < 1.0):
                problems.append(f"{k} fuera de rango ({v})")
        price = float(st["price"])
        if abs(price - float(df["Close"].iloc[-1])) > 1e-6 * max(price, 1.0):
            problems.append("el precio de la señal no coincide con el último cierre")
        (sl_l, tp_l), (sl_s, tp_s) = st["levels"]["LONG"], st["levels"]["SHORT"]
        if not (sl_l < price < tp_l) or not (tp_s < price < sl_s):
            problems.append("stops y objetivos mal ordenados respecto al precio")
        if problems:
            add("error", "Señal", "; ".join(problems))
        else:
            add("ok", "Señal",
                f"P(TP) LONG {st['p_long']:.1%} · SHORT {st['p_short']:.1%} · "
                f"umbral {st['threshold']:.1%}; niveles coherentes")
    guard(_chk_senal)

    # 7) Previsión de tendencia
    def _chk_prevision():
        fp = forecast_path(df, st)
        price = float(st["price"])
        centre, spread = np.asarray(fp["centre"], float), np.asarray(fp["spread"], float)
        if not (np.isfinite(centre).all() and np.isfinite(spread).all()):
            add("error", "Previsión", "la proyección contiene valores no válidos")
        elif spread[-1] <= 0 or centre[-1] <= 0:
            add("error", "Previsión", "banda de volatilidad o objetivo no válidos")
        elif abs(centre[-1] / price - 1) > 0.15:
            add("warn", "Previsión",
                f"objetivo central {centre[-1] / price - 1:+.1%} respecto al precio: "
                "inusualmente lejano, revisa el sesgo")
        else:
            add("ok", "Previsión",
                f"{fp['H']} velas · objetivo {centre[-1] / price - 1:+.2%} · "
                f"±1σ ${spread[-1]:,.0f} · sesgo {st['bias_label'].lower()}")
    guard(_chk_prevision)

    # 8) ¿Tiene el modelo ventaja detectable? (validación cronológica)
    def _chk_validacion():
        v = c["validation"]
        parts, weak = [], 0
        for side in ("LONG", "SHORT"):
            auc, skill = v[side]["auc"], v[side]["skill"]
            parts.append(f"{side} AUC {auc:.3f} · mejora log-loss {skill:+.2%}")
            if not (np.isfinite(auc) and auc > 0.52 and skill > 0):
                weak += 1
        txt = " · ".join(parts)
        if weak == 2:
            add("warn", "Ventaja del modelo",
                f"{txt} · sin ventaja detectable: no te fíes de las probabilidades")
        else:
            add("ok" if weak == 0 else "info", "Ventaja del modelo", txt)
    guard(_chk_validacion)

    return out


def health_section_html(c):
    items = c["health_items"] = health_check(c)
    errs = sum(1 for lv, _, _ in items if lv == "error")
    warns = sum(1 for lv, _, _ in items if lv == "warn")
    if errs:
        head, col = f"{errs} error{'es' if errs != 1 else ''}" + (
            f" · {warns} aviso{'s' if warns != 1 else ''}" if warns else ""), RED
    elif warns:
        head, col = f"{warns} aviso{'s' if warns != 1 else ''}", GOLD
    else:
        head, col = "Todo en orden", GREEN

    checked = c.get("checked_at")
    when = (pd.Timestamp(checked, unit="s", tz="UTC") if checked
            else pd.Timestamp.now(tz="UTC"))
    rows = []
    for level, name, detail in items:
        icon, color = HEALTH_ICONS[level]
        rows.append(
            f"<tr><td style='color:{color};width:28px'><b>{icon}</b></td>"
            f"<td style='white-space:nowrap'>{name}</td>"
            f"<td class='qd-ev'>{detail}</td></tr>")
    table = ("<div class='qd-tw'><table class='qd-table'>"
             "<tr><th></th><th>Comprobación</th><th>Estado</th></tr>"
             + "".join(rows) + "</table></div>")
    desc = (f"{badge(head, col)} Revisión automática de que los datos, los modelos y "
            "la previsión están actualizados y son coherentes (comprobado a las "
            f"{when:%H:%M:%S} UTC, en cada actualización).")
    return section("Autodiagnóstico", desc) + table


def health_summary(c):
    """(errores, avisos) del autodiagnóstico, para la línea de estado."""
    items = c.get("health_items") or health_check(c)
    return (sum(1 for lv, _, _ in items if lv == "error"),
            sum(1 for lv, _, _ in items if lv == "warn"))


# ---------- Precios pasados de las temporalidades (bajo el precio del encabezado) ----------

def _fetch_past_5m(symbol):
    """Cierres de 5 min de los últimos ~5 días (resolución común a todas las
    temporalidades, hasta 24 h atrás)."""
    data = yf.download(symbol, period="5d", interval="5m", auto_adjust=True,
                       progress=False, threads=False)
    if data is None or data.empty:
        raise RuntimeError("Sin datos de 5 min.")
    if isinstance(data.columns, pd.MultiIndex):
        data.columns = data.columns.get_level_values(0)
    s = pd.to_numeric(data["Close"], errors="coerce").dropna()
    s.index = (s.index.tz_localize("UTC") if s.index.tz is None
               else s.index.tz_convert("UTC"))
    return s[~s.index.duplicated(keep="last")].sort_index()


def past_prices_html(df, st, symbol, mobile=False):
    """Fila con el precio pasado de cada temporalidad y la variación del precio
    actual frente a él, medido desde el cierre de la última vela del dashboard:
    - Hace 5 min, 30 min, 45 min, 1 h, 4 h, 6 h, 12 h, 24 h (las temporalidades
      seleccionables) y 2 días: velas de 5 min (descarga con caché de 2 min); si
      fallan, las velas del dashboard cuando su resolución lo permite.
    - Hace 1 semana, 1 mes, 6 meses y 1 año: cierre diario anterior a esa fecha
      (misma caché diaria que usa el modelo; no añade descargas).
    - Máximo anual y mínimo anual: sobre los cierres diarios de los últimos 365
      días hasta la fecha de esta consulta (con la fecha de cada uno).
    - Mayor subida y mayor caída del año en curso (1 de enero hasta la consulta):
      la mayor variación entre dos días, de un mínimo a un máximo posterior
      (tendencia alcista) y de un máximo a un mínimo posterior (bajista), con
      cierres diarios, las dos fechas y los dos precios.
    Si un dato no está disponible se muestra «N/D»."""
    try:
        price = float(st["price"])
        last = df.index[-1]
        last = last.tz_localize("UTC") if last.tzinfo is None else last.tz_convert("UTC")
        ref = last + pd.Timedelta(minutes=BAR_MIN)           # cierre de la última vela
        try:
            s5 = _cache_get(("past5m", symbol), 120, lambda: _fetch_past_5m(symbol))
        except Exception:
            s5 = None
        try:
            per = "8y" if BAR_MIN >= 1440 else "5y"
            dclose = _cache_get(("yf", symbol, per), 6 * 3600,
                                lambda: _yf_close(symbol, per), daily=True)
            today = pd.Timestamp.now(tz="UTC").tz_localize(None).normalize()
            dclose = dclose[dclose.index < today]            # sólo cierres ya cerrados
            if not len(dclose):
                dclose = None
        except Exception:
            dclose = None
        dec = 0 if price >= 10_000 else 2

        def intraday(m):
            past = None
            if s5 is not None and len(s5):
                ts = ref - pd.Timedelta(minutes=m + 5)        # vela de 5 min que cierra en ref-m
                pos = s5.index.searchsorted(ts, side="right") - 1
                if pos >= 0 and ts - s5.index[pos] <= pd.Timedelta(minutes=15):
                    past = float(s5.iloc[pos])
            if past is None and m >= BAR_MIN:
                tt = ref - pd.Timedelta(minutes=m + BAR_MIN)
                if tt in df.index:
                    past = float(df.loc[tt, "Close"])
            return past

        def daily(offset):
            if dclose is None:
                return None
            t = (ref - offset).tz_convert("UTC").tz_localize(None)
            d = t.normalize() - pd.Timedelta(days=1)         # último cierre diario anterior
            pos = dclose.index.searchsorted(d, side="right") - 1
            if pos >= 0 and d - dclose.index[pos] <= pd.Timedelta(days=5):
                return float(dclose.iloc[pos])
            return None

        items = [(f"Hace {_fmt_dur(tf['minutes'])}", intraday(tf["minutes"]),
                  key == INTERVAL, None)
                 for key, tf in TIMEFRAMES.items()]
        items += [("Hace 2 días", intraday(2880), False, None),
                  ("Hace 1 semana", daily(pd.DateOffset(weeks=1)), False, None),
                  ("Hace 1 mes", daily(pd.DateOffset(months=1)), False, None),
                  ("Hace 6 meses", daily(pd.DateOffset(months=6)), False, None),
                  ("Hace 1 año", daily(pd.DateOffset(years=1)), False, None)]
        nd = (f"<b style='display:block;color:{MUTED}'>N/D</b>"
              f"<span style='display:block'>&nbsp;</span>")
        year = None
        if dclose is not None:
            year = dclose[dclose.index >= today - pd.Timedelta(days=365)]
            year = year if len(year) >= 300 else None
        if year is not None:
            hi, lo = float(year.max()), float(year.min())
            d_hi, d_lo = year.idxmax(), year.idxmin()
            items.append(("Máximo anual", hi, False, RED, f"{d_hi:%d/%m/%Y}"))
            items.append(("Mínimo anual", lo, False, GREEN, f"{d_lo:%d/%m/%Y}"))
        else:
            items += [("Máximo anual", None, False, RED, None),
                      ("Mínimo anual", None, False, GREEN, None)]

        chips = []
        for label, past, sel, accent, *rest in items:
            note = rest[0] if rest else None
            border = BLUE if sel else (accent or BORDER_COLOR)
            if past is None or not np.isfinite(past) or past <= 0:
                body = nd
            else:
                chg = price / past - 1
                col = GREEN if chg >= 0 else RED
                body = (f"<b style='display:block'>${past:,.{dec}f}</b>"
                        f"<span style='display:block;color:{col}'>{chg:+.2%}</span>")
                if note:
                    body += (f"<small style='display:block;color:{MUTED}'>"
                             f"{note}</small>")
            chips.append(
                f"<div style='background:{rgba(BLUE, 0.10) if sel else PANEL2};"
                f"border:1px solid {border};border-radius:10px;"
                f"padding:{'4px 7px' if mobile else '6px 10px'};"
                f"min-width:{'72px' if mobile else '92px'};text-align:left;"
                f"font-size:{'11px' if mobile else '12px'};line-height:1.35'>"
                f"<small style='display:block;color:{accent or MUTED}'>{label}</small>"
                f"{body}</div>")
        # Mayor subida (mínimo → máximo posterior) y mayor caída (máximo → mínimo
        # posterior) entre dos días del año natural en curso, con cierres diarios.
        ytd = None
        if dclose is not None:
            ytd = dclose[dclose.index >= pd.Timestamp(today.year, 1, 1)]
            ytd = ytd if len(ytd) >= 10 else None
        for label, accent, up in ((f"Mayor subida {today.year}", GREEN, True),
                                  (f"Mayor caída {today.year}", RED, False)):
            if ytd is None:
                body = nd
            else:
                v = ytd.to_numpy(float)
                if up:
                    j = int((v / np.minimum.accumulate(v) - 1).argmax())
                    i = int(v[:j + 1].argmin())
                else:
                    j = int((v / np.maximum.accumulate(v) - 1).argmin())
                    i = int(v[:j + 1].argmax())
                pct = v[j] / v[i] - 1
                body = (f"<b style='display:block;color:{accent}'>{pct:+.2%}</b>"
                        f"<small style='display:block;color:{MUTED}'>"
                        f"{ytd.index[i]:%d/%m} → {ytd.index[j]:%d/%m}</small>"
                        f"<small style='display:block;color:{MUTED}'>"
                        f"${v[i]:,.{dec}f} → ${v[j]:,.{dec}f}</small>")
            chips.append(
                f"<div style='background:{PANEL2};border:1px solid {accent};"
                f"border-radius:10px;padding:{'4px 7px' if mobile else '6px 10px'};"
                f"min-width:{'72px' if mobile else '92px'};text-align:left;"
                f"font-size:{'11px' if mobile else '12px'};line-height:1.35'>"
                f"<small style='display:block;color:{accent}'>{label}</small>"
                f"{body}</div>")
        style = ("display:flex;flex-wrap:wrap;gap:6px;margin:10px 0 2px 0;"
                 f"color:{TEXT}")
        return (f"<div style='{style}' title='Precio pasado y variación del precio "
                "actual. Hasta 2 días: velas de 5 min; de 1 semana a 1 año: cierre "
                "diario anterior; máximo y mínimo anual: cierres diarios de los "
                "últimos 365 días; mayor subida/caída: entre dos días del año en "
                "curso (de mínimo a máximo posterior, o de máximo a mínimo posterior). "
                "La temporalidad elegida va resaltada.'>" + "".join(chips) + "</div>")
    except Exception:
        return ""


# ============================================================
# 9. BLOQUES DE PÁGINA (notebook y HTML exportado comparten todo)
# ============================================================


# ============================================================
# HORARIO DE MERCADO EN HORA DE ESPAÑA (PENÍNSULA, CET / CEST)
# ============================================================

MADRID = ZoneInfo("Europe/Madrid")

# (zona origen, hora local, días válidos en esa zona 0=lun…6=dom,
#  evento, detalle, sesión, color)
_ALL = (0, 1, 2, 3, 4, 5, 6)
_WD = (0, 1, 2, 3, 4)
MARKET_EVENTS = [
    ("UTC", dtime(0, 0), _ALL, "Cierre de vela diaria y funding",
     "Cierre diario/semanal/mensual de BTC; liquidación de funding en "
     "perpetuos (00:00 UTC).", "Todas", BLUE),
    ("America/Chicago", dtime(17, 0), (6, 0, 1, 2, 3), "Reapertura CME (futuros BTC)",
     "Tras la pausa diaria; el domingo abre con posible hueco (gap) "
     "respecto al spot.", "EE. UU.", PURPLE),
    ("Asia/Tokyo", dtime(9, 0), _WD, "Apertura de Tokio",
     "Arranca la sesión asiática; suele marcar el primer impulso del día.",
     "Asia", GOLD),
    ("Asia/Hong_Kong", dtime(9, 30), _WD, "Apertura de Hong Kong",
     "Flujo asiático y mercados de derivados regionales.", "Asia", GOLD),
    ("UTC", dtime(8, 0), _ALL, "Funding y vencimiento diario de opciones",
     "Funding de las 08:00 UTC y expiración diaria de opciones en Deribit.",
     "Europa", BLUE),
    ("Europe/Madrid", dtime(9, 0), _WD, "Apertura de Europa",
     "Madrid, Fráncfort y Londres abren en torno a esta hora (Londres "
     "08:00 local); sube el volumen.", "Europa", GREEN),
    ("America/New_York", dtime(8, 30), _WD, "Datos macro de EE. UU.",
     "IPC, empleo y otros datos suelen publicarse a las 08:30 ET: "
     "volatilidad alta.", "EE. UU.", RED),
    ("America/New_York", dtime(9, 30), _WD, "Apertura de Nueva York (NYSE/Nasdaq)",
     "Opera el ETF de BTC al contado; máximo de volumen y volatilidad.",
     "EE. UU.", GREEN),
    ("UTC", dtime(16, 0), _ALL, "Funding (16:00 UTC)",
     "Tercera liquidación diaria de funding en perpetuos.", "EE. UU.", BLUE),
    ("America/New_York", dtime(14, 0), _WD, "Decisión de la Fed (días FOMC)",
     "Sólo los días de reunión del FOMC (14:00 ET); movimientos bruscos.",
     "EE. UU.", RED),
    ("America/New_York", dtime(16, 0), _WD, "Cierre de Nueva York",
     "Cierra el ETF de BTC y la bolsa; baja la liquidez de forma notable.",
     "EE. UU.", GOLD),
    ("America/Chicago", dtime(16, 0), (0, 1, 2, 3), "Pausa diaria CME",
     "Cierre diario de futuros hasta las 17:00 CT (el viernes cierra para "
     "el fin de semana).", "EE. UU.", PURPLE),
]


def _next_occurrence(tz_name, t, days, now_madrid):
    tz = ZoneInfo(tz_name)
    base = now_madrid.astimezone(tz)
    for i in range(0, 9):
        d = (base + timedelta(days=i)).date()
        if d.weekday() not in days:
            continue
        cand = datetime.combine(d, t, tzinfo=tz)
        if cand > now_madrid:
            return cand.astimezone(MADRID)
    return None


def market_schedule_html():
    now = datetime.now(MADRID)
    abbr = now.tzname()                       # CET o CEST
    off = int(now.utcoffset().total_seconds() // 3600)

    rows = []
    for tz_name, t, days, name, detail, sess, color in MARKET_EVENTS:
        nxt = _next_occurrence(tz_name, t, days, now)
        # hora típica en España para la tabla (misma fecha de referencia: hoy)
        ref = datetime.combine(now.astimezone(ZoneInfo(tz_name)).date(), t,
                               tzinfo=ZoneInfo(tz_name)).astimezone(MADRID)
        rows.append({"ref": ref, "next": nxt, "name": name, "detail": detail,
                     "sess": sess, "color": color, "days": days})

    rows.sort(key=lambda r: (r["ref"].hour, r["ref"].minute))
    upcoming = [r["next"] for r in rows if r["next"] is not None]
    soonest = min(upcoming) if upcoming else None

    body = []
    for r in rows:
        is_next = soonest is not None and r["next"] == soonest
        cls = " class='qd-next'" if is_next else ""
        tag = (" " + badge("PRÓXIMO", GOLD)) if is_next else ""
        delta = ""
        if r["next"] is not None:
            mins = int((r["next"] - now).total_seconds() // 60)
            h, m = divmod(mins, 60)
            delta = f"en {h} h {m:02d} min" if h else f"en {m} min"
        wd = "Lun–Vie" if r["days"] == _WD else (
            "Todos los días" if r["days"] == _ALL else "Según calendario")
        body.append(
            f"<tr{cls}><td class='qd-hr'>{r['ref']:%H:%M}{tag}</td>"
            f"<td class='qd-ev'><span style='color:{r['color']}'>●</span> "
            f"{r['name']}<small>{r['detail']}</small></td>"
            f"<td class='l'><span class='qd-sess' style='color:{r['color']}'>"
            f"{r['sess']}</span></td>"
            f"<td class='l' style='color:{MUTED}'>{wd}</td>"
            f"<td style='color:{MUTED}'>{delta}</td></tr>")

    clock = ("<div class='qd-clock'><span>Ahora en España: "
             f"<b>{now:%H:%M}</b> ({abbr}, UTC{off:+d})</span>"
             "<span>Todas las horas están convertidas a horario peninsular "
             "(CET en invierno · CEST en verano).</span></div>")

    return (
        section("Calendario de mercado · Horario de España",
                "Momentos clave para operar Bitcoin, expresados "
                "exclusivamente en hora peninsular española (CET/CEST) y "
                "ordenados a lo largo del día. El cambio de hora de EE. UU. y "
                "Europa no coincide durante unas semanas al año; las horas "
                "se recalculan solas.")
        + clock
        + "<div class='qd-tw'><table class='qd-table'>"
          "<tr><th class='l'>Hora (España)</th><th class='l'>Momento clave</th>"
          "<th class='l'>Sesión</th><th class='l'>Frecuencia</th>"
          "<th>Próxima vez</th></tr>"
        + "".join(body) + "</table></div>"
    )


def build_blocks(c, mobile):
    symbol, asset, st = c["symbol"], c["asset"], c["st"]
    df, trades, metrics = c["df"], c["trades"], c["metrics"]
    validation, breakeven, params = c["validation"], c["breakeven"], c["params"]
    capital, risk_pct, max_pos = c["capital"], c["risk_pct"], c["max_pos"]

    price, atr, rsi, thr = st["price"], st["atr"], st["rsi"], st["threshold"]
    p_long, p_short = st["p_long"], st["p_short"]
    sig = st["signal"]
    sig_color = SIGNAL_COLORS[sig]
    tp_atr = ATR_MULTIPLIER * RISK_REWARD
    trend_color = TREND_COLORS.get(st["trend"], MUTED)

    gaps_txt = (f"{c['gaps']} huecos corregidos" if c["gaps"] else "sin huecos")
    src = c.get("sources") or {}
    n_ok = sum(1 for ok, _ in src.values() if ok)
    bad = [k for k, (ok, _) in src.items() if not ok]
    src_color = GREEN if src and not bad else GOLD
    src_txt = (f"{n_ok}/{len(src)} fuentes"
               + (f" · sin {', '.join(bad)}" if bad else ""))
    thr_mode = (f"automático: equilibrio + {AUTO_MARGIN:.0%}"
                if params["auto"] else "manual")

    past_html = past_prices_html(df, st, symbol, mobile)

    hero = f"""
    <div class="qd-box">
      <div class="qd-brand"><span class="qd-live"></span>
        Quant Desk Pro · {VERSION}</div>
      <div class="qd-sym">{symbol} · velas de {TF['label'].lower()}</div>
      <div class="qd-price">${price:,.2f}</div>
      {past_html}
      <div class="qd-meta">
        <div><small>Última vela cerrada</small>{df.index[-1]:%Y-%m-%d %H:%M} UTC</div>
        <div><small>Histórico</small>{len(df):,} velas · {gaps_txt}</div>
        <div><small>Datos extra</small>
          <b style="color:{src_color}">{src_txt}</b></div>
        <div><small>Tendencia EMA</small>
          <b style="color:{trend_color}">{st['trend'].capitalize()}</b></div>
      </div>
    </div>"""

    signal = f"""
    <div class="qd-box qd-sig" style="--c:{sig_color}">
      <div class="qd-sigk">Alerta del modelo</div>
      <div class="qd-pill{' qd-blink' if sig == 'ESPERAR' else ''}"><span>{SIGNAL_ICONS[sig]}</span>{sig}</div>
      <div class="qd-note">Umbral {thr:.1%} ({thr_mode}). Consulta la
        validación antes de confiar en las probabilidades.</div>
      {gauge(f"P(TP) LONG · +{tp_atr:.1f}·ATR antes que −{ATR_MULTIPLIER}·ATR",
             p_long, GREEN, thr, breakeven)}
      {gauge(f"P(TP) SHORT · −{tp_atr:.1f}·ATR antes que +{ATR_MULTIPLIER}·ATR",
             p_short, RED, thr, breakeven)}
      <div class="qd-note">Sesgo del modelo:
        <b style="color:{TREND_COLORS[st['bias_label']]}">{st['bias_label']}</b>
        ({st['bias']*100:+.1f} pp) · LONG {st['edge_long']*100:+.1f} pp vs base
        {st['base_long']:.1%} · SHORT {st['edge_short']*100:+.1f} pp vs base
        {st['base_short']:.1%}. Ambas P(TP) parten de ~33 % por construcción
        (TP a 2R); lo que importa es la desviación, no el valor bruto.</div>
      <div class="qd-leg">
        <span><em style="background:{GOLD}"></em>Equilibrio {breakeven:.1%}</span>
        <span><em style="background:{TEXT}"></em>Umbral usado {thr:.1%}</span>
        <span>Escala 0–{GAUGE_MAX:.0%}</span>
      </div>
    </div>"""

    kpis = [
        card("P(TP) LONG", fmt_pct(p_long, 1), "Modelo LONG", GREEN),
        card("P(TP) SHORT", fmt_pct(p_short, 1), "Modelo SHORT", RED),
        card("Sesgo modelo", st["bias_label"].capitalize(),
             f"{st['bias']*100:+.1f} pp (LONG−SHORT vs base)",
             TREND_COLORS[st["bias_label"]]),
        card("Equilibrio", fmt_pct(breakeven, 1),
             "P(TP) mínima = 1 / (R:R + 1)", GOLD),
        card("RSI (14)", f"{rsi:.1f}", "Momentum", PURPLE),
        card("ATR (14)", f"${atr:,.2f}", "Rango medio", GOLD),
        card("Tendencia EMA", st["trend"].capitalize(), "EMA 9 / 21 / 50",
             trend_color),
    ]

    # Parpadea el título del lado con mayor P(TP) (sin parpadeo si hay empate)
    _lead_long = bool(np.isfinite(p_long) and np.isfinite(p_short)
                      and p_long > p_short)
    _lead_short = bool(np.isfinite(p_long) and np.isfinite(p_short)
                       and p_short > p_long)

    risk_panels = (
        section("Gestión del riesgo y tamaño de posición",
                f"Capital ${capital:,.2f} · riesgo objetivo "
                f"${capital * risk_pct / 100:,.2f} ({risk_pct:.1f}%) · nocional "
                f"máximo {max_pos:.0f}% del capital. Si el tope de exposición "
                "recorta la posición, el riesgo real es menor que el objetivo.")
        + "<div class='qd-pair'>"
        + make_trade_panel("LONG", GREEN, price, *st["levels"]["LONG"],
                           st["sizes"]["LONG"], asset, capital, risk_pct,
                           prob=st["p_long"], edge=st["edge_long"],
                           leader=_lead_long)
        + make_trade_panel("SHORT", RED, price, *st["levels"]["SHORT"],
                           st["sizes"]["SHORT"], asset, capital, risk_pct,
                           prob=st["p_short"], edge=st["edge_short"],
                           leader=_lead_short)
        + "</div>"
    )

    # Orden: [precio | alerta] → LONG/SHORT → previsión de tendencia →
    # tarjetas KPI → horario de mercado
    top = (f"<div class='qd-top'>{hero}{signal}</div>"
           f"<div class='qd-gap'></div>{risk_panels}")
    kpi_block = f"<div class='qd-gap'></div>{grid(kpis, 150)}"
    risk = market_schedule_html()

    market = section(
        "Mercado",
        "Velas, medias exponenciales, volumen y RSI. Las líneas punteadas "
        "marcan stop y objetivo de la señal actual; los triángulos son "
        "las entradas simuladas.")

    val = (
        section("Validación cronológica del modelo",
                f"{validation['LONG']['folds']} tramos con huecos de "
                f"{LABEL_HORIZON} velas entre entrenamiento y test "
                f"({validation['samples']:,} etiquetas por lado), comparados "
                "con una predicción constante. Una mejora de log-loss ≈ 0 y "
                "un AUC ≈ 0,50 indican que no hay señal explotable. Para ser "
                "rentable, la probabilidad debe superar el equilibrio "
                f"({fmt_pct(breakeven, 1)}).")
        + grid(validation_cards(validation, breakeven), 155)
    )

    bt_block = (
        section("Backtest walk-forward vela a vela",
                "Entrada en la apertura siguiente a la señal, stop y TP con "
                "High/Low (si coinciden en una vela, gana el stop), tamaño "
                "por riesgo. No ajustes "
                "el umbral mirando este resultado: invalida el percentil "
                "frente al azar.")
        + grid(backtest_cards(metrics), 155)
    )

    curve = section("Capital y drawdown",
                    "Curva de capital frente a comprar y "
                    "mantener, y caída desde máximos.")

    last_trades = (section("Últimas operaciones simuladas")
                   + trades_table_html(trades, 6 if mobile else 12))

    recent = df.tail(6 if mobile else 12)[
        ["Open", "High", "Low", "Close", "Volume", "RSI", "ATR"]]
    candles = section("Últimas velas") + candles_table_html(recent)

    footer = """
    <div class="qd-foot"><b>Aviso técnico.</b> Las señales y probabilidades no
    garantizan resultados. El backtest usa datos históricos; no simula
    liquidaciones (relevante si la exposición máxima supera el 100%) ni
    profundidad de mercado. Con indicadores técnicos sobre velas cortas es
    habitual no encontrar ventaja: usa la validación, el equilibrio y el
    percentil frente al azar para juzgarlo. La actualización automática depende de que Colab siga
    conectado; para alertas fiables 24/7 ejecuta esta lógica en un servidor.
    No se envían órdenes de compra o venta.</div>"""

    forecast = forecast_section_html(df, st, mobile)

    ta_block = f"<div class='qd-gap'></div>{ta_section_html(df, st)}"

    sent_block = f"<div class='qd-gap'></div>{sentiment_section_html(df, st, symbol, c.get('sources'))}"

    health = health_section_html(c)

    return {"top": top, "forecast": forecast, "kpis": kpi_block, "ta": ta_block,
            "sent": sent_block,
            "health": health, "risk": risk, "market": market,
            "validation": val,
            "backtest": bt_block, "curve": curve, "trades": last_trades,
            "candles": candles, "footer": footer}


# JS de la ventana emergente: se usa en el notebook y se incrusta en el HTML exportado
PANEL_MODAL_JS = r"""
    (function () {
      if (window.__qdModalInstalled) { return; }
      window.__qdModalInstalled = true;

      var st = document.createElement('style');
      st.textContent =
        '#qd-modal-bg{position:fixed;inset:0;z-index:99999;' +
        'background:rgba(3,8,14,.72);backdrop-filter:blur(3px);' +
        'animation:qdfade .18s ease-out}' +
        '#qd-modal-box{position:absolute;left:0;right:0;margin:0 auto;top:12px;max-width:92vw;' +
        'color:#E8EEF6;font-family:\'IBM Plex Sans\',\'Segoe UI\',Roboto,Arial,sans-serif;line-height:1.5;' +
        'max-height:90vh;overflow:auto;' +
        'border-radius:12px;box-shadow:0 20px 60px rgba(0,0,0,.6);animation:qdpop .22s ease-out}' +
        '#qd-modal-box .qd-trade{cursor:default;margin:0;transform:none!important;box-shadow:none!important}' +
        '#qd-modal-x{position:absolute;top:8px;right:10px;z-index:2;width:28px;height:28px;' +
        'border:0;border-radius:50%;background:rgba(255,255,255,.12);color:#fff;font-size:16px;' +
        'line-height:28px;cursor:pointer}' +
        '#qd-modal-x:hover{background:rgba(255,255,255,.28)}' +
        '@keyframes qdfade{from{opacity:0}to{opacity:1}}' +
        '@keyframes qdpop{from{opacity:0;transform:scale(.92)}to{opacity:1;transform:scale(1)}}';
      document.head.appendChild(st);

      function closeModal() {
        var bg = document.getElementById('qd-modal-bg');
        if (bg) { bg.remove(); }
      }

      function openModal(panel) {
        closeModal();
        var r = panel.getBoundingClientRect();
        var bg = document.createElement('div');
        bg.id = 'qd-modal-bg';
        var box = document.createElement('div');
        box.id = 'qd-modal-box';
        box.style.width = Math.round(r.width) + 'px';       // mismo ancho que la zona pulsada
        box.style.minHeight = Math.round(r.height) + 'px';  // y al menos el mismo alto
        var x = document.createElement('button');
        x.id = 'qd-modal-x';
        x.type = 'button';
        x.setAttribute('aria-label', 'Cerrar');
        x.textContent = '\u2715';
        x.addEventListener('click', closeModal);
        box.appendChild(x);
        var clone = panel.cloneNode(true);
        clone.removeAttribute('title');
        box.appendChild(clone);
        bg.appendChild(box);
        bg.addEventListener('click', function (ev) { if (ev.target === bg) { closeModal(); } });
        box.style.visibility = 'hidden';
        document.body.appendChild(bg);

        // Pegado a la parte SUPERIOR de lo que se ve en pantalla. Si la salida
        // vive en un iframe alto (Colab), IntersectionObserver da el tramo
        // visible del iframe en coordenadas del propio iframe.
        function place(top, height) {
          box.style.top = Math.max(12, Math.round(top) + 12) + 'px';
          if (height && height > 220) {
            box.style.maxHeight = Math.round(height - 24) + 'px';
          }
          box.style.visibility = 'visible';
        }
        var done = false;
        try {
          var io = new IntersectionObserver(function (entries) {
            if (done) { return; }
            done = true;
            io.disconnect();
            var e = entries[0];
            if (e && e.isIntersecting && e.intersectionRect) {
              place(e.intersectionRect.top, e.intersectionRect.height);
            } else {
              place(0, 0);
            }
          });
          io.observe(document.documentElement);
        } catch (err) {
          done = true;
          place(0, 0);
        }
        setTimeout(function () { if (!done) { done = true; place(0, 0); } }, 250);
      }

      document.addEventListener('click', function (ev) {
        var t = ev.target;
        if (!t || !t.closest) { return; }
        if (t.closest('#qd-modal-bg')) { return; }
        var panel = t.closest('.qd-trade');
        if (panel) { openModal(panel); }
      });
      document.addEventListener('keydown', function (ev) {
        if (ev.key === 'Escape') { closeModal(); }
      });
    })();
"""


def write_full_dashboard_html(symbol):
    """Siempre en formato escritorio, aunque el modo móvil esté activo."""
    c = S.view
    st = c["st"]
    asset = c["asset"]

    with ui_mode(False):
        blocks = build_blocks(c, mobile=False)
        figs = (build_forecast_figure(c["df"], st, symbol, asset, mobile=False),
                *build_figures(c["df"], c["bt"], symbol, asset,
                               chart_levels(st), mobile=False,
                               trades=c["trades"], capital=c["capital"]))
        parts = []
        for i, (fig, div_id) in enumerate(zip(
                figs, ("forecast_chart", "price_chart", "equity_chart",
                       "drawdown_chart"))):
            html_fig = fig.to_html(
                full_html=False,
                include_plotlyjs="cdn" if i == 0 else False,
                div_id=div_id,
                config={"displayModeBar": True, "responsive": True},
            )
            parts.append(f"<div class='qd-chart'>{html_fig}</div>")

        body = (blocks["forecast"] + parts[0]
                + blocks["top"] + blocks["kpis"] + blocks["ta"] + blocks["sent"] + blocks["health"]
                + blocks["risk"]
                + blocks["market"] + parts[1]
                + blocks["validation"] + blocks["backtest"] + blocks["curve"]
                + parts[2] + parts[3] + blocks["trades"] + blocks["candles"]
                + blocks["footer"])

        html = f"""<!DOCTYPE html>
<html lang="es"><head><meta charset="utf-8">
<meta name="viewport" content="width=device-width, initial-scale=1">
<title>Quant Desk Pro · {symbol}</title>
<style>
{base_css()}
body {{ background:{BG}; margin:0; padding:10px 14px 30px; }}
.page {{ max-width:1240px; margin:0 auto; }}
</style></head><body>
<div class="page qd">{body}</div>
<script>
{PANEL_MODAL_JS}
</script>
</body></html>"""

    with open(fname("html", symbol), "w", encoding="utf-8") as handle:
        handle.write(html)


# ============================================================
# 10. TELEGRAM Y ALERTAS
# ============================================================

def send_telegram(text):
    token = telegram_token_widget.value.strip()
    chat_id = telegram_chat_widget.value.strip()
    if not token or not chat_id:
        return False, "falta el token o el chat ID."

    url = f"https://api.telegram.org/bot{token}/sendMessage"
    payload = urllib.parse.urlencode(
        {"chat_id": chat_id, "text": text}).encode()
    try:
        request = urllib.request.Request(url, data=payload)
        with urllib.request.urlopen(request, timeout=10) as response:
            ok = bool(json.loads(response.read().decode()).get("ok"))
        return ok, "enviado." if ok else "Telegram rechazó el mensaje."
    except Exception as exc:  # no se muestra el token en el error
        return False, f"error de envío ({type(exc).__name__})."


def send_test(_):
    ok, msg = send_telegram("✅ Quant Desk: prueba de conexión correcta.")
    set_status(f"Telegram: {msg}", GREEN if ok else RED)


def notify_signal(symbol, st, bundle):
    """Avisa por Telegram sólo cuando cambia la señal (con histéresis)."""
    if not telegram_enable_widget.value:
        return ""

    sig, prev = st["signal"], S.alerts.get(symbol)
    if sig == prev or (prev is None and sig == "ESPERAR"):
        return ""

    thr = st["threshold"]
    if sig == "ESPERAR" and prev in ("LONG", "SHORT"):
        p_prev = st["p_long"] if prev == "LONG" else st["p_short"]
        if p_prev >= thr - HYSTERESIS:   # aún cerca del umbral: no avisar
            return ""

    asset = symbol.split("-")[0]
    ts, price = st["ts"], st["price"]

    if sig == "ESPERAR":
        text = (f"⚪ {symbol} {INTERVAL}: sin señal (antes {prev}).\n"
                f"Precio ${price:,.2f} · {ts:%Y-%m-%d %H:%M} UTC")
    else:
        stop, target = st["levels"][sig]
        size = st["sizes"][sig]
        eff = size["risk"] / st["capital"] * 100
        icon = "🟢" if sig == "LONG" else "🔴"
        auc = bundle["validation"][sig]["auc"]
        text = (
            f"{icon} {symbol} {INTERVAL} → {sig}\n"
            f"Vela cerrada: {ts:%Y-%m-%d %H:%M} UTC\n"
            f"P(TP) LONG {st['p_long']:.1%} · SHORT {st['p_short']:.1%} · "
            f"sesgo {st['bias_label']} ({st['bias']*100:+.1f} pp) · "
            f"umbral {thr:.1%} (equilibrio {bundle['breakeven']:.1%})\n\n"
            f"Entrada ≈ ${price:,.2f} (cierre; se ejecuta en la apertura)\n"
            f"Stop Loss: ${stop:,.2f}\n"
            f"Take Profit: ${target:,.2f}\n"
            f"Tamaño: {size['qty']:.6f} {asset} "
            f"(≈ ${size['notional']:,.2f})\n"
            f"Riesgo al stop: ${size['risk']:,.2f} ({eff:.2f}% del capital)\n"
            f"Salida por tiempo: {MAX_HOLD_BARS} velas ({_fmt_dur(MAX_HOLD_BARS * BAR_MIN)})\n\n"
            f"Validación del modelo {sig}: AUC {auc:.3f} "
            "(0.50 = azar). Señal orientativa, sin ventaja garantizada."
        )

    ok, msg = send_telegram(text)
    if ok:
        S.alerts[symbol] = sig
    return f" Telegram {symbol}: {msg}"


def monitor_all(params):
    """Entrena (si hay vela nueva) y avisa para TODOS los activos."""
    notes = []
    for symbol in SYMBOLS:
        try:
            bundle = get_bundle(symbol)
            st = live_state(bundle, params)
            st["capital"] = params["capital"]
            note = notify_signal(symbol, st, bundle)
            if note:
                notes.append(note)
        except Exception as exc:
            notes.append(f" {symbol}: {type(exc).__name__}: {exc}")
    return notes


# ============================================================
# 11. RENDER DEL DASHBOARD
# ============================================================

# ============================================================
# 11b. CONSULTA DIARIA DE LA TENDENCIA
# ============================================================

_DAILY_NOTE = ""     # resultado de la última consulta forzada (se conserva)


def daily_trend_info(symbol):
    """Tendencia diaria (EMA 20/50/200 de los cierres diarios) a partir de la última
    copia descargada. No consulta la red. Excluye el día en curso (vela abierta)."""
    per = "8y" if BAR_MIN >= 1440 else "5y"
    hit = S.ext.get(("yf", symbol, per))
    if not hit:
        return None
    fetched, dclose = hit
    today = pd.Timestamp.now(tz="UTC").tz_localize(None).normalize()
    closed = dclose[dclose.index < today]
    if len(closed) < 210:
        return None
    e20, e50, e200 = (float(closed.ewm(span=n, adjust=False).mean().iloc[-1])
                      for n in (20, 50, 200))
    price = float(closed.iloc[-1])
    trend = ("ALCISTA" if e20 > e50 > e200
             else "BAJISTA" if e20 < e50 < e200 else "MIXTA")
    return {"last": closed.index[-1], "price": price, "trend": trend,
            "vs200": price / e200 - 1, "fetched": fetched,
            "stale": fetched < _daily_cutoff()}


def update_daily_html(note=None):
    """Indicador de la tendencia diaria y de si hay que actualizar los datos diarios."""
    global _DAILY_NOTE
    if note is not None:
        _DAILY_NOTE = note
    try:
        info = daily_trend_info(symbol_widget.value)
        if info is None:
            body = ("<b>Tendencia diaria:</b> aún sin datos diarios "
                    "(se consultan al actualizar).")
            color = MUTED
        else:
            age = _fmt_span(max(0, (_time.time() - info["fetched"]) / 60.0))
            color = GOLD if info["stale"] else GREEN
            estado = ("⚠ pendiente de consultar tras el cierre diario"
                      if info["stale"] else "✔ al día")
            body = ("<b>Tendencia diaria:</b> "
                    f"<b style='color:{TREND_COLORS[info['trend']]}'>{info['trend']}</b>"
                    f" · {info['vs200'] * 100:+.1f} % vs EMA200 · último cierre "
                    f"{info['last']:%d/%m/%Y} · consultado hace {age} · "
                    f"<b style='color:{color}'>{estado}</b>")
        extra = (f"<br><span style='color:{MUTED}'>{_DAILY_NOTE}</span>"
                 if _DAILY_NOTE else "")
        daily_html.value = ("<div style='font-size:12.5px;line-height:1.5;"
                            f"color:{TEXT}'>{body}{extra}</div>")
    except Exception:
        pass


def _daily_change_note(before, after, t_start):
    hora = pd.Timestamp.now(tz="Europe/Madrid").strftime("%H:%M")
    if after is None or after["fetched"] < t_start:
        return (f"Consulta forzada a las {hora}: no se pudieron descargar los datos "
                "diarios; se mantiene la última copia.")
    if before is None:
        return (f"Consulta forzada a las {hora}: último cierre "
                f"{after['last']:%d/%m/%Y}, tendencia diaria {after['trend']}.")
    cambios = []
    if after["last"] != before["last"]:
        cambios.append(f"nuevo cierre diario ({before['last']:%d/%m} → "
                       f"{after['last']:%d/%m})")
    if after["trend"] != before["trend"]:
        cambios.append(f"tendencia diaria {before['trend']} → {after['trend']}")
    if cambios:
        return f"Consulta forzada a las {hora}: " + "; ".join(cambios) + "."
    return (f"Consulta forzada a las {hora}: sin cambios (último cierre "
            f"{after['last']:%d/%m/%Y}, tendencia {after['trend']}).")


def force_daily(_=None):
    """Fuerza la consulta diaria: marca como caducados los datos diarios (si la
    descarga falla se conserva la última copia), recalcula y compara antes/después."""
    symbol = symbol_widget.value
    before = daily_trend_info(symbol)
    t_start = _time.time()
    with _LOCK:
        for k in list(S.ext):
            if k and k[0] in ("yf", "fng"):
                S.ext[k] = (0.0, S.ext[k][1])
        S.bundles.clear()
        S.sim_key = None
        S.sim = None
        S.view = None
    set_status("Consulta diaria forzada; descargando cierres diarios y "
               "recalculando…", BLUE, True)
    render_dashboard()
    update_daily_html(_daily_change_note(before, daily_trend_info(symbol), t_start))


def apply_health_status(c, forced=False):
    """Línea de estado según el autodiagnóstico; el botón «Forzar autodiagnóstico»
    sólo se muestra si hay avisos (dorado) o errores (rojo)."""
    n_err, n_warn = health_summary(c)
    if forced:
        if n_err:
            set_status(f"Autodiagnóstico forzado: {n_err} error(es) y {n_warn} "
                       "aviso(s), revisa el panel.", RED, bold=True)
        elif n_warn:
            set_status(f"Autodiagnóstico forzado: {n_warn} aviso(s) (sin errores).",
                       GOLD, bold=True)
        else:
            set_status("Autodiagnóstico forzado: sin incidencias.", GREEN, bold=True)
    elif n_err:
        set_status(f"Dashboard actualizado, pero el autodiagnóstico detecta "
                   f"{n_err} error(es) y {n_warn} aviso(s): revisa el panel.",
                   RED, bold=True)
    elif n_warn:
        set_status(f"Dashboard actualizado · autodiagnóstico: {n_warn} aviso(s) "
                   "(sin errores).", GOLD, bold=True)
    else:
        set_status("Dashboard actualizado correctamente · autodiagnóstico "
                   "sin incidencias.", GREEN, bold=True)
    diag_button.layout.display = "" if (n_err or n_warn) else "none"


def force_diag(_=None):
    """Repite el autodiagnóstico ahora: reevalúa la frescura con la hora actual,
    consulta al proveedor si hay velas más nuevas y vuelve a pintar el panel. No
    reentrena los modelos (para eso, «Actualizar ahora»)."""
    c = getattr(S, "view", None)
    if c is None:
        set_status("Aún no hay datos que revisar: pulsa «Actualizar ahora».",
                   GOLD, bold=True)
        return
    set_status("Forzando autodiagnóstico: consultando al proveedor…", BLUE, True)
    try:
        with _LOCK:
            raw, _gaps = download_data(c["symbol"])
        c["probe"] = {"last": raw.index[-1]}
    except Exception as exc:
        c["probe"] = {"error": f"{type(exc).__name__}: {exc}"}
    try:
        c["checked_at"] = _time.time()
        c.pop("health_items", None)
        set_output([HTML(f"<style>{base_css()}</style>"
                         + wrap(health_section_html(c)))], health_output)
        update_daily_html()
        apply_health_status(c, forced=True)
    except Exception as exc:
        set_status(f"No se pudo completar el autodiagnóstico: "
                   f"{type(exc).__name__}: {exc}", RED, bold=True)


def _render_dashboard_impl(_=None):
    set_status("Actualizando datos y modelos…")

    try:
        load_progress(2, "Leyendo parámetros…")
        params = read_params()
        symbol = symbol_widget.value
        asset = symbol.split("-")[0]
        capital, risk_pct = params["capital"], params["risk"]
        max_position_pct = params["maxpos"]

        mobile = bool(mobile_widget.value)
        UI.update(mobile=mobile, size=18 if mobile else 23,
                  min=135 if mobile else 10_000)

        if capital <= 0:
            raise ValueError("El capital debe ser positivo.")

        load_progress(4, f"Cargando datos de {symbol}…")
        bundle = get_bundle(symbol)
        df = bundle["df"]
        validation = bundle["validation"]
        breakeven = bundle["breakeven"]

        load_progress(68, "Calculando la señal…")
        st = live_state(bundle, params)
        st["capital"] = capital
        thr = st["threshold"]

        load_progress(70, "Simulando el backtest…", creep=80)
        sim = get_simulation(bundle, params, thr)
        bt, trades, metrics = sim["bt"], sim["trades"], sim["metrics"]

        S.metrics, S.validation, S.trades = metrics, validation, trades
        S.signal, S.signal_color = st["signal"], SIGNAL_COLORS[st["signal"]]
        S.view = {
            "symbol": symbol, "asset": asset, "df": df, "bt": bt,
            "trades": trades, "metrics": metrics, "validation": validation,
            "breakeven": breakeven, "st": st, "params": params,
            "gaps": bundle["gaps"], "sources": bundle["sources"], "capital": capital,
            "risk_pct": risk_pct, "max_pos": max_position_pct,
            "checked_at": _time.time(), "built_at": bundle.get("built_at"),
            "bundle_key": bundle["key"],
        }

        tg_note = notify_signal(symbol, st, bundle)

        price = st["price"]
        p_long, p_short = st["p_long"], st["p_short"]
        rsi, atr = st["rsi"], st["atr"]

        load_progress(82, "Construyendo los paneles…")
        blocks = build_blocks(S.view, mobile)
        load_progress(88, "Dibujando los gráficos…")
        price_fig, equity_fig, drawdown_fig = build_figures(
            df, bt, symbol, asset, chart_levels(st), mobile,
            trades=trades, capital=capital)
        forecast_fig = build_forecast_figure(df, st, symbol, asset, mobile)

        items = [
            HTML(f"<style>{base_css()}</style>" + wrap(blocks["top"])),
            HTML(wrap(blocks["kpis"], blocks["ta"], blocks["sent"], blocks["risk"], blocks["market"])),
            price_fig,
            HTML(wrap(blocks["validation"], blocks["backtest"],
                      blocks["curve"])),
            equity_fig,
            drawdown_fig,
            HTML(wrap(blocks["trades"], blocks["candles"], blocks["footer"])),
        ]
        load_progress(92, "Mostrando los resultados…")
        set_output([HTML(f"<style>{base_css()}</style>"
                         + wrap(blocks["forecast"])), forecast_fig],
                   forecast_output)
        set_output(items)
        S.view.pop("probe", None)          # la consulta al proveedor era de antes
        set_output([HTML(f"<style>{base_css()}</style>"
                         + wrap(blocks["health"]))], health_output)

        # ---------- Exportación ----------
        load_progress(95, "Guardando los archivos…")
        bt.to_csv(fname("backtest", symbol), index=False)
        trades.to_csv(fname("trades", symbol), index=False)
        drop_cols = [c for c in df.columns
                     if c.startswith("Target_") or c.startswith("Label_End_")]
        df.drop(columns=drop_cols).to_csv(fname("candles", symbol))

        flat_val = {}
        for side in ("LONG", "SHORT"):
            for k, v in validation[side].items():
                flat_val[f"Val_{side}_{k}"] = v

        pd.DataFrame([{
            **metrics, **flat_val,
            "Symbol": symbol, "Interval": INTERVAL, "Signal": S.signal,
            "ATR_Mult": ATR_MULTIPLIER, "Risk_Reward": RISK_REWARD,
            "Max_Hold_Bars": MAX_HOLD_BARS,
            "P_TP_Long": p_long, "P_TP_Short": p_short,
            "Bias_pp": st["bias"] * 100, "Bias": st["bias_label"],
            "Breakeven": breakeven, "Threshold": thr,
            "Last_Close": price, "RSI": rsi, "ATR": atr,
            "Capital_Input": capital, "Risk_Percent": risk_pct,
            "Max_Position_Percent": max_position_pct,
        }]).to_csv(fname("metrics", symbol), index=False)

        S.exported = symbol
        load_progress(98, "Generando el dashboard HTML…")
        write_full_dashboard_html(symbol)

        _dist = ATR_MULTIPLIER * atr
        _lvl = (f"Stop ±${_dist:,.0f} ({_dist / price:.2%}) · "
                f"TP ±${_dist * RISK_REWARD:,.0f} ({_dist * RISK_REWARD / price:.2%}) · "
                f"{ATR_MULTIPLIER:g}·ATR, R:R {RISK_REWARD:g} · "
                f"máx. {MAX_HOLD_BARS} velas")
        alert_html.value = f"""
        <div class="qd-alert" style="--c:{S.signal_color}">
          <b>{symbol} · {TF['label']} · {SIGNAL_ICONS[S.signal]} {S.signal}</b><br>
          Precio ${price:,.2f} · P(TP) LONG {p_long:.1%} · SHORT {p_short:.1%}
          · umbral {thr:.1%}<br>
          {_lvl}<br>
          Sesgo del modelo: <b>{st['bias_label']}</b> ({st['bias']*100:+.1f} pp;
          LONG {st['edge_long']*100:+.1f} pp / SHORT {st['edge_short']*100:+.1f} pp
          vs tasa base {st['base_long']:.1%} / {st['base_short']:.1%})<br>
          <small>Actualizado {pd.Timestamp.now().strftime('%Y-%m-%d %H:%M:%S')}.
          Alerta visual; no se envían órdenes.{tg_note}</small>
        </div>
        """

        update_daily_html()
        apply_health_status(S.view)

    except Exception as exc:
        set_status(f"Error al actualizar: {type(exc).__name__}: {exc}",
                   RED, bold=True)
        try:
            set_output([], forecast_output)   # no dejar una previsión obsoleta
            set_output([], health_output)
            diag_button.layout.display = "none"
            set_output([HTML(
                f"<style>{base_css()}</style>"
                f"<div class='qd-alert' style='--c:{RED}'>"
                "<b>No se pudo completar el análisis</b><br>"
                f"{type(exc).__name__}: {exc}</div>")])
        except Exception:
            pass


def render_dashboard(_=None):
    """Igual que antes, pero con indicador de carga visible mientras trabaja."""
    global _LOAD_MAP
    first = getattr(S, "view", None) is None
    nested = _LOAD_DEPTH > 0          # p. ej. dentro del ciclo automático (ya va por el 50 %)
    _LOAD_MAP = (50.0, 50.0) if nested else (0.0, 100.0)
    show_loading("Cargando datos y modelos…" if first else "Actualizando datos…")
    try:
        _render_dashboard_impl(_)
    finally:
        hide_loading()
        _LOAD_MAP = (0.0, 100.0)


# ============================================================
# 12. ACTUALIZACIÓN PERIÓDICA
# ============================================================

async def auto_refresh_loop():
    while S.auto:
        await asyncio.sleep(REFRESH_SECONDS)
        if not S.auto:
            break
        try:
            params = read_params()
            show_loading("Actualizando datos…")
            load_progress(2, "Descargando y entrenando los 3 activos…", creep=45)
            try:
                # El trabajo pesado (descarga + entrenamiento de los 3 activos)
                # va en un hilo para no bloquear el kernel.
                await asyncio.to_thread(monitor_all, params)
                load_progress(50, "Actualizando el panel…")
                render_dashboard()
            finally:
                hide_loading()
        except asyncio.CancelledError:
            raise
        except Exception as exc:
            set_status("Error en el ciclo automático: "
                       f"{type(exc).__name__}: {exc}", RED)


def start_auto_refresh(_):
    if S.task is not None and not S.task.done():
        set_status("La actualización automática ya está activa.", GOLD)
        return

    S.auto = True
    try:
        S.task = asyncio.create_task(auto_refresh_loop())
    except RuntimeError:
        S.task = asyncio.ensure_future(auto_refresh_loop())

    set_status(
        f"Actualización automática cada {REFRESH_SECONDS} s sobre "
        f"{', '.join(SYMBOLS)} (sólo reentrena con vela nueva). "
        "Mantén la sesión de Colab conectada.", GREEN)


def stop_auto_refresh(_):
    S.auto = False
    if S.task is not None:
        S.task.cancel()
        S.task = None
    set_status("Actualización automática detenida.", GOLD)


# ============================================================
# 13. ARRANQUE
# ============================================================

# Si se reejecuta la celda, cancelar tareas antiguas.
try:
    if _prev is not None and getattr(_prev, "task", None) is not None:
        _prev.auto = False
        _prev.task.cancel()
except Exception:
    pass

# Reloj de la temporalidad: cancelar el de una ejecución anterior y arrancar uno nuevo.
try:
    _old_clock = globals().get("_clock_task")
    if _old_clock is not None:
        _old_clock.cancel()
except Exception:
    pass

def apply_timeframe(selected):
    """Fija todas las constantes que dependen de la duración de la vela."""
    global INTERVAL, TF, BAR_MIN, BAR_HOURS, BAR_FREQ, VOL_WIN, VOL_MIN
    global MIN_BARS, MAX_HOLD_BARS, LABEL_HORIZON, WF_MIN_START
    global WF_REFIT_BARS, ANNUAL_PERIODS, CONFIG_VERSION
    global ATR_MULTIPLIER, RISK_REWARD
    INTERVAL = selected
    TF = TIMEFRAMES[selected]
    BAR_MIN = TF["minutes"]
    BAR_HOURS = BAR_MIN / 60.0
    BAR_FREQ = f"{BAR_MIN}min"
    VOL_WIN, VOL_MIN = TF["vol"]
    MIN_BARS = TF["min_bars"]
    ATR_MULTIPLIER = TF["atr"]
    RISK_REWARD = TF["rr"]
    MAX_HOLD_BARS = TF["hold"]
    LABEL_HORIZON = MAX_HOLD_BARS      # gap anti-fuga = horizonte de etiqueta
    WF_MIN_START = TF["wf_min"]
    WF_REFIT_BARS = max(1, int(round(TF["refit_h"] / BAR_HOURS)))
    ANNUAL_PERIODS = int(24 * 365 / BAR_HOURS)
    CONFIG_VERSION = make_config_version()


def on_timeframe_change(change):
    """Aplica la temporalidad y fuerza una descarga/reconstrucción completa."""
    if change.get("name") != "value":
        return
    selected = change.get("new", DEFAULT_INTERVAL)
    if selected not in TIMEFRAMES or selected == INTERVAL:
        return
    apply_timeframe(selected)
    update_timeframe_clock()
    with _LOCK:
        S.bundles.clear()
        S.sim_key = None
        S.sim = None
        S.view = None
    set_status(f"Temporalidad cambiada a {TF['label']}; recalculando datos y modelos…", BLUE, True)
    render_dashboard()

timeframe_widget.observe(on_timeframe_change, names="value")
refresh_button.on_click(render_dashboard)
daily_button.on_click(force_daily)
diag_button.on_click(force_diag)
start_button.on_click(start_auto_refresh)
stop_button.on_click(stop_auto_refresh)
telegram_test_button.on_click(send_test)

for _kind, _button in download_buttons.items():
    _button.on_click(lambda _b, k=_kind: download_file(k, user_click=True))

display(HTML(f"""
<style>{base_css()}</style>
<div class="qd qd-wrap" id="qd-top-anchor">
  <div class="qd-box">
    <div class="qd-brand"><span class="qd-live"></span>
      Quant Desk Pro · {VERSION}</div>
    <div style="font-size:26px;font-weight:600;letter-spacing:-.02em;margin-top:12px">
      Inteligencia de mercado cripto</div>
    <div class="qd-note" style="font-size:13px">
      HistGradientBoosting para LONG y SHORT · triple barrera real ·
      backtest vela a vela · validación cronológica</div>
  </div>
</div>
"""))

_wrap = widgets.Layout(flex_flow="row wrap", gap="10px")

display(panel("Parámetros", [
    widgets.HBox([symbol_widget,
                  widgets.HBox([timeframe_widget, timeframe_clock],
                               layout=widgets.Layout(align_items="center")),
                  capital_widget,
                  risk_widget,
                  max_position_widget], layout=_wrap),
    params_dashboard_text]))   # enlace en la esquina inferior derecha del módulo
display(panel("Umbral de señal", [auto_thr_widget, threshold_widget]))
display(forecast_output)      # Previsión de tendencia
display(panel("Acciones", [
    widgets.HBox([refresh_button, daily_button, start_button, stop_button],
                 layout=_wrap),
    daily_html]))
display(panel("Exportar", [
    widgets.HBox(list(download_buttons.values()), layout=_wrap)]))
display(widgets.HBox([status, diag_button],
                     layout=widgets.Layout(flex_flow="row wrap", gap="10px",
                                           align_items="center")),
        alert_html, health_output, status_js, loading_js, output)

# Reloj en vivo junto a la temporalidad
update_timeframe_clock()
try:
    _clock_task = asyncio.ensure_future(timeframe_clock_loop())
except Exception:
    _clock_task = None

# Primer análisis
render_dashboard()
try:   # fuera el hueco temporal reservado para el indicador de arranque
    _boot_display(_BootJS(
        "var _s=document.getElementById('qd-boot-spacer');if(_s){_s.remove();}"))
except Exception:
    pass


# ============================================================
# 14. IR AL INICIO DEL DOCUMENTO AL TERMINAR DE CARGAR
# ============================================================

def scroll_to_top(delays_ms=(0, 400, 1200, 2500)):
    """Lleva la vista al inicio del documento (cabecera del dashboard).
    Se repite con retardos porque el dashboard crece mientras se pinta;
    así el último desplazamiento ya ocurre con todo cargado."""
    js = """
    (function () {
      function goTop() {
        try {
          var a = document.getElementById('qd-top-anchor');
          if (a) { a.scrollIntoView({block: 'start', behavior: 'instant'}); }
        } catch (e) {}
        try { window.scrollTo(0, 0); } catch (e) {}
        try { document.documentElement.scrollTop = 0; document.body.scrollTop = 0; } catch (e) {}
        try { window.parent.scrollTo(0, 0); } catch (e) {}
        try {
          // Colab: contenedor con scroll del notebook
          var sc = (window.parent || window).document.querySelector(
              'colab-window, .notebook-container, #cells, .main-container');
          if (sc) { sc.scrollTop = 0; }
        } catch (e) {}
      }
      var delays = %s;
      delays.forEach(function (d) { setTimeout(goTop, d); });
    })();
    """ % json.dumps(list(delays_ms))
    try:
        display(Javascript(js))
    except Exception:
        pass


scroll_to_top()


# ============================================================
# 15. VENTANA CENTRAL AL PULSAR LONG = COMPRAR / SHORT = VENDER
# ============================================================

def install_panel_modal():
    """Al pulsar un panel LONG/SHORT se abre, pegado arriba y centrado en horizontal, un
    cuadro del mismo tamaño que la zona pulsada con su contenido.
    Se cierra con la X, pulsando fuera del cuadro o con Esc.
    El listener es delegado en el documento: sobrevive a cada re-render."""
    js = PANEL_MODAL_JS
    try:
        display(Javascript(js))
    except Exception:
        pass


install_panel_modal()
```
